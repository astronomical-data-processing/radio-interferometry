## 4.1 Fourier 直觉到干涉条纹的桥接<a id='visibility:sec:periodic_fourier'></a>


第 2 章已经系统讨论傅里叶级数、连续/离散变换、卷积与采样。本节只承担一次物理翻译，不重新推导这些结果；已经完成第二章核心路径的读者可以直接进入 [4.2 基线及其空间表示](4_2_the_baseline_and_its_representation_in_space.ipynb)。全书指数和归一化采用[第 2.0 节标准约定](../2_Mathematical_Groundwork/2_0_introduction.ipynb#math:sec:canonical_conventions)。

干涉仪中的一条投影基线对应一组复指数条纹。二维近似下，位于 $(l_0,m_0)$ 的点源满足

$$I(l,m)=S\,\delta(l-l_0)\delta(m-m_0)\quad\Longrightarrow\quad V(u,v)=S e^{-2\pi i(ul_0+vm_0)}.$$

这条变换对已经包含本章最需要的 Fourier 直觉：源通量决定可见度幅度，源位置决定相位斜率，而多个非相干亮度成分因测量方程的线性性在可见度中相加。

| 图像域结构 | Fourier/可见度域表现 | 干涉测量含义 |
|:---|:---|:---|
| 平滑的大尺度发射 | 集中在原点附近 | 需要短基线和零间距约束 |
| 紧致或尖锐结构 | 延伸到较高空间频率 | 长基线提高细节约束，但不补大尺度通量 |
| 整体位置平移 | 幅度不变，增加线性相位 | 相位中心和基线符号直接决定位置符号 |
| 有限边界或硬截断 | 产生振荡和旁瓣 | 视场窗、主波束和采样边界会进入响应 |

![天空与 Fourier 平面的对应](../2_Mathematical_Groundwork/figures/fourier_transform_sky_uv_pair.png)

**图 4.1.1** 天空亮度与二维 Fourier 域的对应。幅度描述各空间频率成分的强弱，相位记录它们相对于所选原点的位置关系。

这里的 `uv` plane 只是在小视场二维近似下的工作平面。完整测量仍依赖 $w$、频率、时间、偏振和方向响应；[4.6 节](4_6_the_fourier_approximation_van_cittert-zernike_theorem.ipynb)将给出这些项何时可以化为二维 Fourier 关系。
