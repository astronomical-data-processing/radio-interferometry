# 6.P 去卷积综合问题集

本问题集检验能否把 CLEAN 迭代、主循环/次循环、恢复图单位、宽带与多尺度模型、残差统计和源目录选择函数连接成一条可审计的推断链。第 1--3 题是本科高年级核心，共 60 分；第 4 题是进阶选读，共 18 分；第 5 题是课程训练，共 22 分。完整研究生路径总分 100 分。所有数值题必须写单位和归一化约定；所有算法题必须区分观测约束与模型预测。


## 1. 有符号 Högbom CLEAN（18 分）

考虑一维教学残图和峰值归一化 PSF：

$$R_0=[0,\,0.4,\,1.0,\,-0.5,\,0.1],\qquad B=[0.1,\,-0.2,\,1,\,-0.2,\,0.1],$$

PSF 峰位于索引 2，循环增益 $\gamma=0.2$，使用线性边界裁剪。

1. 写出 Högbom CLEAN 的 `AND` 停止条件，并说明为什么峰值搜索使用 $|R|$、分量振幅却保留 $R$ 的符号。（4 分）
2. 手算前两次迭代的峰位置、加入模型的分量和新残图。（8 分）
3. 若 CLEAN mask 只允许索引 1 至 3，说明 mask 改变的是测量数据、模型支撑还是二者。（3 分）
4. 解释为什么用 `roll` 平移 PSF 会在边缘制造错误，以及正确实现应怎样处理重叠区。（3 分）


## 2. 主循环与恢复图单位（22 分）

1. 比较 Högbom、Clark 和 Cotton-Schwab CLEAN 的次循环响应与主循环更新域。说明为什么 Cotton-Schwab 主循环要从观测可见度减去模型预测后重新成像。（8 分）
2. 写出恢复图公式

$$I_R=B_C*I_M+\epsilon R_D,$$

并解释三项的单位。某 clean beam 为 $2.5''\times2.0''$，dirty beam 有效主瓣为 $3.0''\times2.4''$；在相同高斯面积约定下计算 $\epsilon$。（5 分）
3. 某扩展源位置残差为 $0.30\,\mathrm{mJy\,dirty\,beam^{-1}}$。计算缩放后加入恢复图的数值，并说明忽略缩放会怎样偏置残余扩展通量。（4 分）
4. 说明为什么恢复图 Fourier 平面在未采样坐标上有连续值，却不能称为获得了更充分的观测采样。（5 分）


## 3. 多尺度、宽带与宽场模型（20 分）

1. 一个 FWHM 为 12 个像素的高斯展源同时用点源 CLEAN 和尺度集合 $[0,4,12,24]$ 的 multiscale CLEAN 处理。解释尺度归一化、scale bias 和 CLEAN mask 分别影响什么；为什么比较相同迭代数不等于比较相同计算成本。（6 分）
2. 区分 MFS 与 MT-MFS。若在参考频率附近采用 $I(\nu)\simeq I_0+I_1t$、$t=(\nu-\nu_0)/\nu_0$，某亮源测得 $I_0=0.80\,\mathrm{Jy\,beam^{-1}}$、$I_1=-0.56\,\mathrm{Jy\,beam^{-1}}$，估算谱指数并写出这一近似的边界。（6 分）
3. 说明二维 Fourier 成像近似需要哪些小视场或 $w$ 项条件；van Cittert-Zernike 定理本身是否只在小角近似下成立？（4 分）
4. 为同时含离轴强点源、弥散辐射和显著谱曲率的数据选择 deconvolver、宽场校正和主要 QA 产品，并给出理由。（4 分）


## 4. 停止准则与统计验证（进阶选读，18 分）

某次去卷积的训练可见度残差持续下降，但恢复图中的弱源数不断增加；图像残差直方图近似高斯，残差自相关却在 PSF 旁瓣尺度上有显著峰。

1. 为什么单一 RMS 或近似高斯直方图不能证明模型可信？列出至少四项互补诊断。（5 分）
2. 写出带噪声协方差的数据一致性项 $(Ax-d)^\dagger C_n^{-1}(Ax-d)$。解释为什么图像像素一般不独立，以及在哪里计算白化残差更直接。（5 分）
3. 说明循环增益、字典、mask 和 early stopping 中，哪些主要是数值步长，哪些形成模型假设或隐式正则化。（4 分）
4. 设计保留可见度验证：说明校准、数据分块、$uv$ 覆盖和信息泄漏应怎样处理。为什么反复选择同一个验证集上的最优参数仍会过拟合？（4 分）


## 5. 从高斯拟合到可发表目录（课程训练，22 分）

一幅 Jy/beam 恢复图的椭圆高斯波束为 $4''\times3''$。某源拟合峰值为 $2.0\,\mathrm{mJy\,beam^{-1}}$，观测 FWHM 为 $6''\times4''$，且源轴与波束轴对齐。

1. 计算高斯积分通量和两个去卷积 FWHM。若协方差差矩阵有非正特征值，应怎样报告？（6 分）
2. 搜索区域为 $1\,\mathrm{deg^2}$，波束为 $10''\times8''$。计算独立波束数，并估算理想高斯噪声中超过单边 $4\sigma$ 与 $5\sigma$ 的假峰数。使用 $P(Z>4)=3.167\times10^{-5}$、$P(Z>5)=2.867\times10^{-7}$。（6 分）
3. 一组注入实验在某通量箱注入 1000 个源，找回 780 个；输出 820 个候选，其中 760 个与注入源唯一匹配。计算完整率与可靠率，并解释为什么二者不能互换。（4 分）
4. 解释 Eddington bias、resolution bias 和主波束位置相关噪声怎样改变阈值目录。给出一个包含负图检验和注入恢复的目录验证方案。（6 分）


## 参考量与评分要点

**第 1 题：** 前两次都选择索引 2，分量分别为 $0.20$ 和 $0.16$；$R_1=[-0.02,0.44,0.80,-0.46,0.08]$，$R_2=[-0.036,0.472,0.64,-0.428,0.064]$。必须说明 mask 约束模型搜索而不新增数据，且边缘采用真实重叠裁剪。

**第 2 题：** $\epsilon=(2.5\times2.0)/(3.0\times2.4)\simeq0.694$，缩放残差约为 $0.208\,\mathrm{mJy\,clean\,beam^{-1}}$；不缩放会把该残差项高估约 $1/\epsilon\simeq1.44$ 倍。算法比较 8 分，公式与单位 5 分，数值 4 分，观测约束解释 5 分。

**第 3 题：** 一阶近似给出 $\alpha\simeq I_1/I_0=-0.70$；必须指出更宽频带、谱曲率、频率相关主波束和低 S/N 比值偏差。二维 Fourier 近似的限制不应误写成 VCZ 定理本身失效。

**第 4 题：** 满分答案需包含残差空间相关、局部 RMS、模型稳定性、可见度白化残差、注入恢复或独立保留数据中的至少四项。循环增益是步长；字典、约束、mask 和停止规则形成模型选择。验证阶段不得使用由完整数据求得的校准解。

**第 5 题：** $S_{\rm int}=4.0\,\mathrm{mJy}$；对齐轴下去卷积 FWHM 约为 $4.47''$ 和 $2.65''$。波束面积约 $90.65\,\mathrm{arcsec^2}$，独立波束数约 $1.43\times10^5$，故期望 $4\sigma$ 假峰约 4.53 个、$5\sigma$ 假峰约 0.041 个。完整率为 0.780，可靠率约为 0.927。选择偏差与验证设计各占 3 分。
