# 3.P 综合问题集

本问题集共 100 分，用于检验从星表方向到可见度相位的完整坐标变换链。每题均应写明单位转换、角度与时间约定、适用近似和主要中间量；仅给出最终数值不足以检验符号与方向约定。A、B 考查观测几何，C、D、G 考查天空方向与可见度相位的关系，E、F 考查时间标准与台站坐标。


## A. 时角与观测时刻（12 分）

某源赤经为 $\alpha=8^{\rm h}20^{\rm m}$，观测时地方恒星时为 $\mathrm{LST}=10^{\rm h}40^{\rm m}$。采用 $H=\mathrm{LST}-\alpha$，计算时角的小时数和角度，并判断天源位于子午圈东侧还是西侧。它距离 $H=+4^{\rm h}$ 还有多少恒星时？再用 $\Delta\mathrm{LST}/\Delta t_{\rm UT1}=1.00273790935$ 估算对应的 `UT1` 时间。

进一步说明：若误把 UTC 直接当作 LST，缺少哪些信息？回答中应提到日期对应的格林尼治恒星时、台站经度、`UT1-UTC`，以及平均/视恒星时的选择。


## B. 高度角限制与观测窗口（14 分）

台站纬度为 $\varphi=34^\circ$，天源赤纬为 $\delta=20^\circ$。使用

$$
\sin\mathcal{E}=\sin\varphi\sin\delta+\cos\varphi\cos\delta\cos H
$$

求天源满足高度角 $\mathcal{E}\geq30^\circ$ 时允许的最大 $|H|$，并换算为以上中天为中心的总观测窗口。随后计算上中天高度角，并检验结果是否与观测窗口相容。

说明仅满足几何高度角门限仍不足以保证数据质量的原因，可从大气路径长度、地平遮挡或天线阴影中任选一项讨论。


## C. 方向余弦与小角近似（16 分）

相位中心为 $(\alpha_0,\delta_0)=(12^{\rm h},30^\circ)$，目标位于 $\Delta\alpha=+20'$、$\Delta\delta=-10'$。这里 $\Delta\alpha$ 是赤经角本身，而不是已经乘过 $\cos\delta$ 的天球距离。由

$$
\begin{aligned}
l &= \cos\delta\sin(\alpha-\alpha_0),\\
m &= \sin\delta\cos\delta_0-\cos\delta\sin\delta_0\cos(\alpha-\alpha_0),\\
n &= \sin\delta\sin\delta_0+\cos\delta\cos\delta_0\cos(\alpha-\alpha_0)
\end{aligned}
$$

计算精确的 $(l,m,n)$，验证 $l^2+m^2+n^2=1$。再与 $l\simeq\cos\delta_0\Delta\alpha$、$m\simeq\Delta\delta$ 比较，给出两个近似的绝对误差，并解释为什么赤经偏移必须带上 $\cos\delta_0$。说明为什么本题可取 $n=+\sqrt{1-l^2-m^2}$，以及这个正号分支在全天球上为何不唯一。


## D. 星表误差到可见度相位（16 分）

一条投影长度 $B=5000\ {\rm km}$ 的 VLBI 基线在 $8.4\ {\rm GHz}$ 观测。假设位置误差完全沿基线投影方向，利用

$$
\Delta\phi=2\pi\frac{B}{\lambda}\Delta\theta
$$

计算 $\Delta\theta=0.10\ {\rm mas}$ 引起的相位误差。若位置误差为零均值高斯随机量，且相位均方根等于上述数值，使用 $\exp(-\sigma_\phi^2/2)$ 估计相干幅度保留比例。最后反求将相位误差限制在 $10^\circ$ 以内时，星表位置误差应小于多少 mas。

说明这一定量结果为什么使参考架、历元、自行和相位参考源位置成为 VLBI 数据处理的一部分，而不只是星表元数据。


## E. UT1-UTC 误用引起的相位误差（16 分）

某历元的 $\mathrm{UT1}-\mathrm{UTC}=+0.600\ {\rm s}$，处理程序却把 UTC 数值直接当成 UT1。取地球自转角速度 $\omega_\oplus=7.2921150\times10^{-5}\ {\rm rad\,s^{-1}}$，计算地球转角误差的弧度和角秒值。再对 $B=100\ {\rm km}$ 基线取最不利投影，用 $|\Delta\tau|\simeq B|\Delta\theta_\oplus|/c$ 估计延迟误差，并求 $1.4\ {\rm GHz}$ 的相位误差和相位周数。

说明为什么这是上限量级而不是所有基线共有的固定误差；再写出 `TT` 与 `TAI` 的固定关系，并说明闰秒为何应交给时间库处理。


## F. 地心地固基线向局部东-北-天顶坐标的转换（14 分）

参考台站的大地纬度和东经为 $(\varphi,\lambda)=(34^\circ,-107^\circ)$，另一台天线相对它的 ECEF 差向量为 $(\Delta X,\Delta Y,\Delta Z)=(100,-50,20)\ {\rm m}$。使用 3.5 节的旋转矩阵计算 $(E,N,U)$，并验证旋转前后向量长度不变。

解释为什么矩阵中的 $\varphi$ 应使用大地纬度而不是地心纬度；指出地心和地理两极分别存在哪些坐标退化。


## G. 小视场几何与 $w$ 项判据（12 分）

某可见度样本有 $w=10^5$（波长），源距参考方向 $\theta=1^\circ$。对径向偏移使用 $n=\cos\theta$，计算精确的

$$
|\Delta\phi_w|=2\pi|w|(1-n).
$$

再反求要求 $|\Delta\phi_w|<1\ {\rm rad}$ 时允许的最大视场半径，并与小角结果 $\theta_{\max}\simeq(\pi|w|)^{-1/2}$ 比较。解释为什么“$\sin\theta\simeq\theta$ 很准确”不能单独证明二维傅里叶近似成立。


## 参考核对量

- A：$H=2^{\rm h}20^{\rm m}=35^\circ$，天源位于子午圈西侧；还需 $1^{\rm h}40^{\rm m}$ 恒星时，对应约 $1^{\rm h}39^{\rm m}43.6^{\rm s}$ UT1。
- B：$|H|\simeq66.65^\circ=4.44^{\rm h}$，总窗口约 $8.89\ {\rm h}$；上中天高度角为 $76^\circ$。
- C：$(l,m,n)\simeq(0.00504674,-0.00290154,0.99998306)$；$l,m$ 近似的绝对误差分别约为 $8.41\times10^{-6}$、$7.34\times10^{-6}$。
- D：$\lambda\simeq0.03569\ {\rm m}$，相位误差约 $0.427\ {\rm rad}=24.45^\circ$，相干幅度约保留 $91.3\%$；$10^\circ$ 门限对应约 $0.0409\ {\rm mas}$。
- E：转角误差约 $4.375\times10^{-5}\ {\rm rad}=9.025''$；最不利延迟约 $14.59\ {\rm ns}$，相位约 $128.38\ {\rm rad}=20.43$ 周。
- F：$(E,N,U)\simeq(110.249,6.192,26.586)\ {\rm m}$；两边长度均约 $113.578\ {\rm m}$。
- G：$|\Delta\phi_w|\simeq95.696\ {\rm rad}$；精确和小角的 $1\ {\rm rad}$ 半径均约 $0.1022^\circ$。

核对量只用于检查数量级。提交时仍需给出公式、单位和物理解释。
