## 3.x 延伸阅读与参考文献


本章的延伸阅读分为三类：位置天文学、时间与参考系统的基础资料，射电干涉几何的核心教材，以及用于实现标准坐标变换和观测处理的软件文档。


### 位置天文学、时间与参考系统

* [Practical Astronomy with your Calculator or Spreadsheet](https://www.cambridge.org/core/books/practical-astronomy-with-your-calculator-or-spreadsheet/coordinate-systems/01F111D3505038668441D3715414C40F)  
  适合系统复习坐标系统、时间系统以及若干基础观测几何概念。它比本章更接近传统位置天文学教材，可作为坐标与时间变量的补充背景。
* [JPL Horizons Manual](https://ssd.jpl.nasa.gov/horizons/manual.html)  
  该官方手册系统说明参考架、观测者位置、视位置与星历服务之间的关系，并展示现代天文软件对参考系统和观测时刻的规范处理。
* [IERS Conventions (2010)](https://iers-conventions.obspm.fr/conventions_versions.php) 与 [IERS EOP 数据中心](https://datacenter.iers.org/eop.php)  
  精密时间、地球定向、ITRS/GCRS 变换与台站位移模型的规范来源。阅读本章 3.5 节时应重点区分 UTC、UT1、ERA 与 EOP。
* [ICRF3（Charlot et al. 2020）](https://doi.org/10.1051/0004-6361/202038368) 与 [ITRF2020](https://itrf.ign.fr/en/solutions/ITRF2020)  
  分别展示天球参考架和地球参考架如何由实际源位置、站坐标与速度实现。


### 射电干涉核心教材

* [Essential Radio Astronomy（NRAO）](https://www.cv.nrao.edu/~sransom/web/Ch3.html)  
  这套公开课程讲义将赤道坐标、时角、台站本地坐标和干涉仪几何置于统一的观测框架中，适合作为本章的补充材料。
* [Interferometry and Synthesis in Radio Astronomy](https://www.loc.gov/item/2019756474)  
  这是射电干涉测量领域的经典教材。其几何关系、两单元干涉仪和可见度响应等章节，将本章的 $(H,\delta)$ 与 $(l,m,n)$ 进一步扩展到 $(u,v,w)$ 坐标和可见度积分。
* [Tools of Radio Astronomy](https://link.springer.com/book/10.1007/978-3-642-39950-3)  
  适合把单天线观测、射电辐射基础和干涉测量几何放在更宽广背景下通读。


### 现代软件与实践框架

* [Astropy Coordinates Documentation](https://docs.astropy.org/en/stable/coordinates/index.html)  
  这份文档非常适合把本章公式落实到 Python 工作流中。`SkyCoord`、`ICRS`、`AltAz`、`EarthLocation` 等对象正是现代坐标变换实践中最常见的工具。
* [IAU SOFA](https://iausofa.org/)  
  IAU 维护的标准天文算法库，覆盖时间标准、岁差-章动、ERA、恒星时以及天球与地球参考系统变换。高精度软件应采用并验证此类标准实现。
* [FITS 天球坐标投影规范（Calabretta & Greisen 2002）](https://doi.org/10.1051/0004-6361:20021327)  
  说明 `SIN` 等天球投影和 FITS WCS 的数学定义，可用于核对 $(l,m,n)$ 与图像坐标轴约定。
* 各阵列的调度与观测准备软件文档  
  CASA、MeerKAT/EVN/VLBI 相关的观测准备工具通常都会在本章的基础上进一步加入阵列工程约束、扫描策略和指向模型。


### 课程与暑校材料

* [NRAO Synthesis Imaging Workshop Lectures](https://science.nrao.edu/science/meetings/2018/16th-synthesis-imaging-workshop/16th-synthesis-imaging-workshop-lectures)  
  这类课程讲义可用于衔接本章的观测几何与地球自转合成下的 $uv$ 覆盖。
* 各类 VLBI/射电干涉暑校与观测学校  
  这类材料通常综合讨论源目录、观测调度、相位中心、校准策略和成像流程。阅读时可重点关注时角覆盖（hour-angle coverage）、高度角限制（elevation limits）、相位参考（phase referencing）和宽视场成像（wide-field imaging）等主题。


#### 阅读建议

建议先阅读 Practical Astronomy 和 Astropy 文档，以巩固坐标、时间和软件对象的定义；随后结合 Interferometry and Synthesis in Radio Astronomy 或 NRAO 公开课程，理解赤经/赤纬、时角/地方恒星时和方位角/高度角在干涉几何中的作用；最后复习本章第 3.2 至 3.4 节，为后续的地球自转合成、相位中心和宽视场成像内容作准备。
