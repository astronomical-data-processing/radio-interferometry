## 1.13 延伸阅读与参考文献 <a id='science:sec:refs'></a>

第 1 章旨在建立射电科学、辐射机制、单碟望远镜与干涉阵之间的基本联系，其内容不能替代系统的射电天文学、辐射过程或干涉测量专著。本节按学习目的选列参考材料，用于补充本章内容并与后续章节衔接。

下列参考材料主要用于三个方面：补充天体物理背景，系统理解干涉测量原理，以及考察相关原理在实际阵列和典型科学问题中的应用。


### 1.13.1 射电科学与辐射物理基础

* [<cite data-cite='1979rpa..book.....R'>Radiative Processes in Astrophysics</cite> &#10548;](https://ui.adsabs.harvard.edu/abs/1979rpa..book.....R/abstract)：介绍热辐射、同步辐射、逆康普顿散射和谱线辐射的经典教材，可用于系统深化 1.2、1.3、1.5、1.6 和 1.7 节涉及的辐射物理。
* [Essential Radio Astronomy &#10548;](https://www.cv.nrao.edu/~sransom/web/Chapters.html)：开放的射电天文学教材，适合沿通量、亮温、谱线、脉冲星和望远镜灵敏度等主题复习，可作为本章核心内容的配套读物。
* [<cite data-cite='2013tra..book.....W'>Tools of Radio Astronomy</cite> &#10548;](https://ui.adsabs.harvard.edu/abs/2013tra..book.....W/abstract)：兼顾射电天文学物理背景与观测方法，可进一步联系辐射测量量、射电谱线和望远镜系统。
* [Distance measures in cosmology &#10548;](https://ui.adsabs.harvard.edu/abs/1999astro.ph..5116H/abstract)：集中说明光度距离、角直径距离、谱通量与 K 修正的频率约定，适合核对 1.2 节中近邻公式向宇宙学源推广时的边界。
* [<cite data-cite='1983QJRAS..24..267H'>The Determination of Cloud Masses and Dust Characteristics from Submillimetre Thermal Emission</cite> &#10548;](https://ui.adsabs.harvard.edu/abs/1983QJRAS..24..267H/abstract)：给出亚毫米热尘埃辐射中不透明度约定和质量估算的经典讨论，可用于理解 1.5 节修正黑体关系及温度、不透明度与气尘比带来的系统不确定性。

前两本教材适合用于系统巩固本章的科学基础：第一本侧重辐射过程及其天体物理表述，第二本侧重射电观测中的基本概念和实际语境。关于宇宙学距离与热尘埃质量估算的两篇文献，可在处理相应专题时查阅。


### 1.13.2 干涉测量与孔径合成基础

* [<cite data-cite='1999ASPC..180.....T'>Synthesis Imaging in Radio Astronomy II</cite> &#10548;](https://ui.adsabs.harvard.edu/abs/1999ASPC..180.....T)：射电干涉测量训练中最经典的讲义集之一，覆盖可见度、成像、标定和科学应用，适合作为本书后续章节的外部参照。
* [<cite data-cite='2017isra.book.....T'>Interferometry and Synthesis in Radio Astronomy</cite> &#10548;](https://ui.adsabs.harvard.edu/abs/2017isra.book.....T/abstract)：干涉测量领域的系统专著，适合将第 1 章中的双缝、可见度、孔径合成和单碟/阵列关系扩展为严格理论框架。

前者适合按专题查阅可见度、成像和标定，后者适合作为完整理论体系的系统参照；二者可分别承担专题复习和深入查阅的功能。


### 1.13.3 面向仪器与观测实践的延伸

* [<cite data-cite='2002ASPC..278..375S'>Short-Spacings Correction from the Single-Dish Perspective</cite> &#10548;](https://ui.adsabs.harvard.edu/abs/2002ASPC..278..375S)：可与 1.10 节配合阅读，用于理解单碟数据与干涉阵数据联合使用的必要性，以及短间距补偿在实际成像中的作用。
* [<cite data-cite='1998AJ....115.1693C'>The NRAO VLA Sky Survey</cite> &#10548;](https://ui.adsabs.harvard.edu/abs/1998AJ....115.1693C)：经典射电巡天实例，集中展示阵列构型、角分辨率、灵敏度、成像流程与科学产出之间的联系。
* [<cite data-cite='mailloux2005phased'>Phased Array Antenna Handbook</cite> &#10548;](https://books.google.co.za/books/about/Phased_Array_Antenna_Handbook.html?id=v-htQgAACAAJ)：可用于补充 ASKAP 等相控阵馈源系统涉及的阵元理论与波束形成技术。

建议在完成第一册第 5 至第 8 章及第二册基础内容后阅读本组材料。掌握成像、主波束、短间距和基本处理流程后，可以更准确地理解这些原理在实际数据处理中的实现方式。


### 1.13.4 与本章案例直接相关的专题阅读

* [<cite data-cite='1996A&ARv...7....1C'>Cygnus A</cite> &#10548;](https://ned.ipac.caltech.edu/level5/Carilli/Car_contents.html)：可作为 1.4、1.6 和 1.8 节中同步辐射、射电星系与谱老化内容的专题延伸，并将教材中的典型射电源与具体天体物理问题联系起来。
* [<cite data-cite='ska2015advancing'>Advancing Astrophysics with the Square Kilometre Array</cite> &#10548;](https://pos.sissa.it/215/)：适合作为 1.11 节之后的科学背景阅读。SKA-Low 与 SKA-Mid 的当期建设和设计参数应以 [SKAO 望远镜页面 &#10548;](https://www.skao.int/en/explore/telescopes) 为准。
* [Hydrogen Epoch of Reionization Array &#10548;](https://reionization.org/)：补充 1.11 节中从 PAPER 到 HERA 的实验演进、阵列设计和再电离 21 cm 功率谱目标。

案例文献有助于建立以科学问题为出发点的仪器认识。阵列设计、观测频段和数据处理流程并非彼此独立的技术环节，而是由具体科学目标和测量误差要求共同约束。


### 1.13.5 建议的阅读顺序

建议按以下顺序阅读：

1. 首先阅读 [Essential Radio Astronomy &#10548;](https://www.cv.nrao.edu/~sransom/web/Chapters.html) 或 [<cite data-cite='2013tra..book.....W'>Tools of Radio Astronomy</cite> &#10548;](https://ui.adsabs.harvard.edu/abs/2013tra..book.....W/abstract)，巩固射电辐射与观测的基本概念和术语。
2. 学习第 4 至第 8 章时，以 [<cite data-cite='2017isra.book.....T'>Interferometry and Synthesis in Radio Astronomy</cite> &#10548;](https://ui.adsabs.harvard.edu/abs/2017isra.book.....T/abstract) 作为干涉测量理论的系统参照。
3. 掌握成像和校准基础后，再阅读 [<cite data-cite='1999ASPC..180.....T'>Synthesis Imaging in Radio Astronomy II</cite> &#10548;](https://ui.adsabs.harvard.edu/abs/1999ASPC..180.....T) 以及短间距、巡天和阵列案例等专题材料。

这一顺序使学习者先建立物理概念，再形成方法框架，最后进入专题与实践，并与本书由科学背景逐步过渡到干涉测量理论和数据处理的结构保持一致。
