## 9.y 综合实践问题集（100 分）

本问题集检验学习者能否把数据检查、校准、成像、自校准、产品 QA 和科学测量组织成一条可审查的证据链。题目不要求记忆软件参数；每个结论必须给出数量级、诊断证据或适用边界。

### 共同案例

一次 1.4 GHz 连续谱观测使用 27 面天线，最长和最短物理基线分别为 11 km 和 35 m。可用带宽为 800 MHz，原始积分时间为 4 s，拟平均到 2 MHz 通道。目标场中有一个离相位中心 12 arcmin 的紧致源、一个角尺度约 15 arcmin 的弥散候选体，以及 60 mJy 的自校准紧致模型。理论图像噪声目标为 12 microJy/beam。

除非题目另有说明，取 $c=3.00\times10^8\,\mathrm{m\,s^{-1}}$、$\omega_\oplus=7.292\times10^{-5}\,\mathrm{rad\,s^{-1}}$，并定义 $\operatorname{sinc}(x)=\sin(x)/x$。


### 1. 数据巡检与异常分类（10 分）

巡检发现：A）与 7 号天线相关的所有基线在 04:10--04:18 振幅同时下降 35%；B）1.315--1.320 GHz 在大部分基线上出现短时宽带增强；C）只在目标场长基线上出现随 $uv$ 距离增大而平滑降低的振幅；D）带通校准源相位对频率存在稳定线性斜率。

1. 将 A--D 分别归入“优先 flag”、“优先校准”或“优先改进天空模型”，每项给出一条跨基线、时间、频率或观测场的判据。（8 分）
2. 说明为什么把 C 当作坏数据删除会产生选择性偏差。（2 分）

### 2. 平均展宽的数量级（12 分）

对最坏投影方向，使用

$$R_{\rm bw}\simeq\operatorname{sinc}\!\left(\pi\,\Delta\nu\,B\theta/c\right),\qquad R_t\simeq\operatorname{sinc}\!\left(\pi\,\omega_\oplus\Delta t\,(B/\lambda)\theta\right).$$

1. 计算 11 km 基线、12 arcmin 离轴角处的 $R_{\rm bw}$、$R_t$ 和联合衰减。（6 分）
2. 若要求带宽展宽损失不超过 2%，使用小量近似 $\operatorname{sinc}(x)\simeq1-x^2/6$ 估算允许的最大 $\Delta\nu$。（4 分）
3. 说明为什么对中心源测得无损不能验证这个平均方案。（2 分）

### 3. 校准链与退化（10 分）

设观测包含已知通量标度源、强带通源和每 12 min 插入一次的相位校准源。

1. 给出 flag、delay/bandpass、复增益、通量标度传递和 apply 的顺序，说明每一步在测量方程中估计的量。（5 分）
2. 指出仅用点源可见度解复增益时的两个退化，并说明参考天线和通量标度源分别解决什么。（3 分）
3. 列出两项 apply 后必须重新检查的诊断，不得以“任务成功返回”作为 QA。（2 分）


### 4. 从科学目标反推成像参数（12 分）

1. 用 $\theta_{\rm res}\simeq\lambda/B_{\max}$ 估计最高角分辨率，提出一个使恢复波束每轴约有 4--5 个像素的 cell size。（3 分）
2. 若需要覆盖直径 30 arcmin 的主波束，计算最小图像像素数，并解释为什么实际 imsize 还需要考虑网格化效率、w-term 和视场外亮源。（3 分）
3. 用 $\theta_{\rm MRS}\simeq0.6\lambda/B_{\min}$ 估计最大可恢复尺度，判断 15 arcmin 弥散候选体的总通量密度是否可信。（3 分）
4. 在“紧致源定位”和“弥散体总通量密度”两个目标之间，说明 weighting、uv taper 和 multiscale 应如何分别取舍。（3 分）

### 5. 自校准可行性与停止准则（10 分）

设单基线、4 s、2 MHz 样本的热噪声为 1 Jy。用下式估算 60 s、800 MHz 解区间的天线相位解信噪比：

$$\mathrm{S/N}_{\rm sol}\simeq {S_{\rm model}\over\sigma_{\rm bl}}\sqrt{(N_{\rm ant}-1){\Delta t_{\rm sol}\over\Delta t_{\rm int}}{\Delta\nu_{\rm sol}\over\Delta\nu_{\rm chan}}}. $$

1. 计算 $\mathrm{S/N}_{\rm sol}$，判断是否足以开始 phase-only 自校准。（4 分）
2. 说明这个估计为什么对展源模型、大量 flag 或天线灵敏度不等的数据过于乐观。（3 分）
3. 给出三条停止自校准的证据，其中至少一条必须来自保留数据或模型独立诊断。（3 分）

### 6. 图像与残差 QA（10 分）

最终图像的峰值为 100 mJy/beam，残差 RMS 为 14 microJy/beam，最大残差为 0.70 mJy/beam；亮源周围仍有与 PSF 旁瓣方向一致的环状结构。

1. 分别计算基于全局峰值/RMS 和峰值/最大残差的两种动态范围，说明两者对局部伪影的敏感性不同。（4 分）
2. 将 14 microJy/beam 与理论值比较，列出至少三个可能的超额噪声或结构来源。（3 分）
3. 为该产品写一句不超过 60 字的限制声明，不得只写“图像质量良好”。（3 分）


### 7. 源表选择函数（12 分）

在 40 个注入源的实验中得到：

| 峰值阈值 | 检测数 | 与真值匹配数 |
|---:|---:|---:|
| $3\sigma$ | 45 | 35 |
| $4\sigma$ | 28 | 27 |
| $5\sigma$ | 22 | 22 |

1. 计算每个阈值的 completeness 和 reliability。（4 分）
2. 若发布目录要求 reliability $\geq0.95$，选择保留完整度最高的阈值，并报告牺牲的注入源比例。（3 分）
3. 解释为什么检测边界的峰值通量密度容易上偏，并指出注入样本在通量密度、角尺度、位置上应如何覆盖真实母体。（3 分）
4. 说明为什么在有真实吸收或强负碗结构时，负源计数不能直接当作正源伪检数。（2 分）

### 8. 谱线与偏振量的边界（10 分）

1. H I 静止频率为 1420.4058 MHz，一条谱线的质心在 1418.0000 MHz。用 radio convention $v=c(\nu_0-\nu)/\nu_0$ 计算速度，并说明为什么发布时必须同时声明参考系和 velocity convention。（4 分）
2. 偏振数据覆盖 $\lambda^2=0.03$--$0.09\,\mathrm{m^2}$。用 $\delta\phi\simeq2\sqrt{3}/\Delta\lambda^2$ 估计 RMSF 宽度，用 $\pi/\lambda_{\min}^2$ 估计最大 Faraday-thick 尺度。（4 分）
3. 若 $Q=0.30$、$U=0.40$ mJy/beam，两者噪声均为 0.10 mJy/beam，计算高信噪比近似 $P_0=\sqrt{Q^2+U^2-\sigma^2}$ 和偏振角 $\chi=\tfrac12\operatorname{atan2}(U,Q)$。（2 分）

### 9. 短间距与联合成像（8 分）

1. 根据第 4 题的 MRS 结果，为 15 arcmin 弥散候选体选择“仅干涉阵”、“合并紧凑配置”、“feather 单碟”或“joint deconvolution”中的至少一个补救路径，并说明选择依据。（4 分）
2. 列出 feather 前必须核对的四项契约：至少包括通量标度、坐标/像素网格、单碟波束和 Fourier 重叠区。（4 分）

### 10. 可复现产物与结论降级（6 分）

为这次处理设计一个最小发布包。列出必须保留的数据身份、flag 版本、校准/成像参数、软件环境、诊断图、测量表、校验和和限制声明。然后写出一条“若无法取得短间距数据，如何降级 15 arcmin 弥散体结论”的发布规则。


### 判分与数值核对

每题的满分答案都应包含“结果 + 依据 + 边界”。只写软件任务名称、只报一个数字或只作定性描述，原则上不超过该小题一半分值。单位、角度弧度转换、sinc 定义或速度约定错误但推理链完整时可给过程分；隐藏关键假设不应给满分。
