## 1.10 单碟望远镜的局限与价值

到这里为止，我们已经知道射电源可以非常复杂，干涉测量也已经在概念上出现了。那么接下来的自然问题就是：既然单台望远镜也能接收射电信号，为什么射电天文学还必须发展干涉阵？这个问题不能只回答“因为分辨率不够”，还必须进一步说明两件事。第一，单碟望远镜确实存在根本性的尺度和工程限制；第二，单碟并没有因此失去价值，相反，它在许多科学任务中仍然不可替代。

因此，本节不应被理解为“单碟已经过时”的声明，而应被理解为一种更成熟的仪器观：单碟望远镜和干涉阵在本质上测量的是同一类电磁波，但它们对空间尺度、视场、亮温、绝对零点和成像流程的敏感方式并不相同。后续真正进入可见度、成像和定标时，这种差别会反复出现，并决定许多实际观测策略。


### 1.10.1 单碟望远镜并没有过时

“单碟”是一个方便的历史称呼，但从更一般的角度说，这里讨论的是**单阵元望远镜**。这个阵元可以是一面可转向的大口径反射面，也可以是低频阵中的单个偶极子或一个先经过波束形成的台站。之所以先把这个概念讲清楚，是因为后面讨论干涉阵时，阵列中的每一个阵元本身往往仍然是一台小型望远镜。

单阵元系统在射电天文学中仍然承担着若干基础任务：测量总功率、恢复极大尺度的弥散辐射、提供绝对亮温或零点信息、开展大面积快速扫描，以及为某些谱线项目和时变源项目提供稳定的单站观测。也就是说，干涉阵并不是把单碟“淘汰掉”，而是在单碟难以继续扩展的地方，用另一种方式把角分辨率推高。

这三幅图想要传达的核心并不是“哪一种看起来更先进”，而是：单阵元和多阵元只是射电望远镜的两种组织方式。单碟适合直接测总功率，阵列适合通过基线合成高分辨率。理解这两类系统的互补性，是学习射电干涉成像的前提。


### 1.10.2 收集面积、口径与角分辨率为什么会失控

单台望远镜最直观的优势，是用有限效率把几何面积转换成有效收集面积。若把望远镜近似为圆形孔径，则有

$$A_{\rm eff}=\eta_a\frac{\pi D^2}{4},$$

这里 $D$ 是口径，$\eta_a<1$ 是包含照明、遮挡、表面误差等损失的孔径效率。更大的 $A_{\rm eff}$ 通常意味着更高的灵敏度，但系统温度也同样关键；常用的等效系统通量密度为

$$\mathrm{SEFD}=\frac{2k_BT_{\rm sys}}{A_{\rm eff}},$$

它越小，望远镜对同一通量密度源的信噪比越高。单碟的衍射极限角分辨率大致满足

$$\Delta \theta \approx 1.22\frac{\lambda}{D}$$

其中 $\lambda$ 是观测波长。这个公式看似简单，却几乎决定了射电望远镜的命运。原因在于：射电波长通常比可见光长得多，因此若想获得相同角分辨率，就必须把口径做得大得多。

这里必须区分两个常被混淆的量。**有效收集面积与系统温度**共同控制灵敏度；**口径尺度**控制角分辨率。虽然在单碟里增大口径通常会同时改善二者，但它们并不是同一回事。到了干涉阵，二者甚至会以不同方式分离开来：阵列灵敏度取决于各阵元的有效面积、系统温度、带宽和相关效率，而最高角分辨率主要由最长基线决定。


| 望远镜类型 | 角分辨率 <br> $\Delta \theta$ | 可见光 <br> $\lambda$ = 500 nm | 红外 <br> $\lambda$ = 10 $\mu$ m | 射电 EHF <br> $\lambda$ = 10 mm <br> 30 GHz | 射电 UHF <br> $\lambda$ = 1 m <br> 300 MHz |
|:---:|:---:|:---:|:---:|:---:|:---:|
| 业余级 | 0.8'' | 15 cm | 3 m | 3 km | 300 km |
| 自动后随 | 0.25'' | 50 cm | 10 m | 10 km | 100 km |
| 小型科研 | 0.12'' | 1 m | 21 m | 21 km | 2100 km |
| 大型科研 | 0.015''（15 mas） | 8 m | 168 m | 168 km | 16800 km |


**表 1.10.1**：在给定观测波长下，为实现某个角分辨率所需的望远镜口径 $D$。


表中的数量级已经足够说明问题。若在 300 MHz 附近想达到接近光学望远镜的角分辨率，所需“口径”会轻易膨胀到几十、几百甚至上千公里。这显然不可能通过建造一面真正连续的反射面来实现。下面的交互小程序进一步展示这一数量级关系：只要给定波长和目标分辨率，就可以立即看出口径需求会有多快失控。


```python
def aperture_diameter(wavelength_m=1.0, angular_resolution_deg=15e-3 / 3600):
    """Print the circular-aperture diameter for a target resolution."""
    frequency_mhz = 299_792_458.0 / wavelength_m / 1e6
    diameter_m = 1.22 * wavelength_m / np.radians(angular_resolution_deg)
    print(f"Frequency: {frequency_mhz:.3f} MHz (wavelength {wavelength_m:.3g} m)")
    print(f"Required diameter: {diameter_m:.3g} m")
    print(f"Resolution: {angular_resolution_deg * 3600:.3g} arcsec")


ipywidgets.interact(
    aperture_diameter,
    angular_resolution_deg=(15e-3 / 3600, 10, 1e-5),
    wavelength_m=(0.5e-6, 1, 1e-7),
)
```

### 1.10.3 为什么不能无限放大单碟

既然单碟分辨率由 $\lambda/D$ 决定，一个最自然的想法就是不断把 $D$ 做大。但真正的工程实现会很快遇到多重瓶颈。

首先，大型可转向结构必须在重力作用下保持足够高的面形精度和指向精度。口径越大，自重引起的挠曲越严重，结构设计、驱动系统、基础设施和维护成本都会急剧上升。其次，观测频率越高，允许的面形误差就越小。若反射面均方根误差为 $\sigma$，Ruze 近似给出高频孔径效率损失

$$\frac{\eta}{\eta_0}=\exp\left[-\left(\frac{4\pi\sigma}{\lambda}\right)^2\right].$$

因此在毫米和亚毫米波段，反射面误差必须远小于波长本身，“做得大”和“做得准”会同时变得非常困难。传统单馈源一次形成一束，但多馈源接收机和相控阵馈源（PAF）可以同时形成多束并提高巡天速度；真正没有改变的是，单台反射面内部不会自动产生阵元间的多条独立基线，也就不能天然获得干涉阵的二维空间频率采样。


大型单碟的历史很好地体现了这些约束。可转向的 300 英尺绿岸望远镜曾经是工程壮举，但也说明大口径结构始终面临机械与材料极限。阿雷西博则采取了另一条路线：把反射面固定在天然地形中，以牺牲天空可达范围换取更大口径。中国的 FAST 采用主动反射面和馈源舱系统，把这一思路进一步推进，但它同样不是“什么都能看到、什么频率都能轻松做到”的万能方案。也就是说，单碟并不是不能继续发展，而是每前进一步都需要付出极高的工程代价。


![口径与分辨率缩放](figures/reflector_scaling_schematic.png)


**图 1.10.4a**：单孔径衍射尺度随口径的缩放。长波观测若追求亚角秒分辨率，会迅速要求数百米到数百千米的等效口径；阴影区提醒大型可动结构的工程难度会同步上升。

### 1.10.4 单碟为什么仍然不可替代

讲完局限之后，更重要的是讲清它的价值。单碟望远镜最大的优势，是它直接测量天空的**总功率**。这件事在干涉阵中并不自动成立，因为干涉阵天然测量的是阵元之间的相关信号，而不是零基线上的绝对总功率。用后面会正式引入的语言说，干涉阵通常缺少 $B=0$ 的“零间距”测量，因此会对足够大尺度的平滑结构不敏感。

这带来一条极其重要的实践结论：许多扩展射电源即使很亮，也可能被干涉阵“分辨掉”。银河系弥散同步辐射、分子云大尺度 CO 发射、星系盘外延 H I 气体，甚至太阳和附近延展星云中的部分结构，都可能需要单碟数据来恢复真正的总通量和低空间频率信息。正因为如此，现代毫米和厘米波工作流中经常要做单碟与干涉阵联合成像，或至少做短间距补偿。


单碟的另一个重要价值在于大面积扫描效率和光谱稳定性。对于巡测、总功率映射、宽视场弥散结构研究，以及某些要求严格绝对标定的谱线项目，单碟往往更直接、也更稳健。干涉阵虽然分辨率高，但并不天然适合所有科学问题。这里最重要的判断是：更高角分辨率不等于更完整的信息，特别是当科学目标本身具有大尺度、低面亮度结构时。

**图 1.10.6**：单碟与干涉阵对空间尺度的响应示意。单碟包含零间距信息，能够测量总功率和大尺度结构；干涉阵由有限的最短与最长基线采样空间频率，因此会同时面对最大可恢复尺度和最高角分辨率两个边界。


### 1.10.5 从单碟到合成孔径：概念上发生了什么

既然单碟不能无限放大，而我们又仍然希望获得更高角分辨率，那么思路就必须改变。关键想法不是继续造一面更大的连续反射面，而是把原本连续孔径中的信息改为**离散采样并数字合成**。下面的卡通示意图并不追求工程细节，而是帮助建立一种非常重要的直觉：干涉阵不是完全不同的一类仪器，它可以被看成是把“连续孔径”拆成了许多可管理的小孔径，然后再在电子学和计算中重建它们之间的关系。


![cartoon_1.png](figures/cartoon_1.png)


**图 1.10.7a**：一个简化的单碟望远镜模型。入射平面波（红色虚线）沿着光线追迹路径（青色）被抛物面反射并汇聚到焦点处的接收机。


先从最普通的单碟开始。远处天体发出的波前可近似看成平面波，被反射面汇聚到焦点处的接收机。此时所有光学或电磁路径都在硬件中天然完成叠加，观测者直接得到总信号。


![cartoon_2.png](figures/cartoon_2.png)


**图 1.10.7b**：可以把整面反射镜看成由多个较小的反射面组成，每个子反射面都有自己通向焦点的光路。


如果把连续反射面想象成由许多小子孔径组成，那么单碟就等价于对孔径进行连续采样的极限情况。在这个意义上，单碟可以被看成一种“连续孔径”系统，而后面的干涉阵则是“稀疏孔径”系统。后续第 4 章会把这种直觉正式转写到孔径平面与可见度空间语言中。


![cartoon_3.png](figures/cartoon_3.png)


**图 1.10.7c**：每个子反射面各自接收信号，随后这些信号在数字域重新组合，以恢复原本应在主焦点处形成的总信号。


一旦每个子孔径都能单独接收并数字化记录信号，我们就不必再依赖反射面把它们被动汇聚到同一个焦点。对射电波来说，这是可行的，因为信号频率足够低，可以被电子学采样和后处理。于是，“光学汇聚”被替换成了“数字合成”。


![cartoon_4.png](figures/cartoon_4.png)


**图 1.10.7d**：原本属于同一面大反射镜的子阵元被布置到地面上，并配有各自的指向系统。通过加入适当的时间延迟，可以在数字域中恢复原始总信号。


接下来更关键的一步是：这些子孔径不必继续排成完整抛物面。只要我们知道它们之间的几何延迟，并能在电子学上做相应补偿，就可以把阵元分散到地面上形成阵列。这一步并不会自动给出图像，但它已经把有效尺度从单台望远镜口径，扩展到了阵列的最大尺度。


![cartoon_5.png](figures/cartoon_5.png)


**图 1.10.7e**：用相关器硬件替代波束形成器后，就可以生成天空图像。


如果只是把多个阵元延迟对齐后直接相加，那对应的是**波束形成**：系统会朝某个方向形成一束等效主波束，得到的仍然更像“一个数”或“一路总信号”。而一旦对各阵元信号做两两相关，就会得到**可见度**。相关器之所以在现代干涉测量中如此核心，就是因为它把阵列从“会看一个方向”推进到“能够恢复二维图像结构”。


![cartoon_6.png](figures/cartoon_6.png)


**图 1.10.7f**：射电干涉阵并不需要覆盖原始“大望远镜”的每一个位置，因此可以构造出直径与地球尺度相当的合成望远镜。


最后，也是最反直觉但最重要的一点：干涉阵并不需要在整个大孔径上连续铺满阵元。即使只在少量离散位置采样，只要基线分布合适，仍然能合成出远大于单阵元的角分辨率。代价则是孔径不再连续，从而导致灵敏度、旁瓣、空间频率覆盖以及大尺度结构恢复都变成需要认真处理的问题。


### 1.10.6 干涉阵解决了什么，又带来了什么代价

干涉阵最直接解决的问题，是把最高角分辨率从单阵元口径提升到最长基线尺度。若最长基线为 $B_{\max}$，最高分辨率近似按 $\theta_{\rm res}\sim\lambda/B_{\max}$ 缩放；若最短投影基线为 $B_{\min}$，最大可恢复尺度常用数量级估计 $\theta_{\rm MRS}\sim 0.6\lambda/B_{\min}$。系数 0.6 不是普适常数，实际 MRS 还取决于投影基线分布、权重、观测时长、mosaic 策略和源形态。与此同时，单阵元口径 $D_{\rm ant}$ 决定主波束视场，近似为 $\theta_{\rm PB}\sim\lambda/D_{\rm ant}$。这三个角尺度分别对应“看得多细”“能恢复多大结构”和“一次看多大天区”，它们不是同一个量。

因此，干涉阵并不是“更大的单碟”。它把分辨率提升到最长基线尺度，但也引入了缺零间距、大尺度通量恢复不全、视场受主波束限制、成像质量依赖 $uv$ 覆盖、相位稳定性和延迟模型要求更高等新问题。数据产品也不再是直接得到一幅图，而是必须经过成像、权重选择、去卷积和定标后才能成为科学图像。

现实中的高质量射电成像几乎总是在不同仪器能力之间做平衡：单碟负责总功率和大尺度结构，干涉阵负责高角分辨率和结构分解，二者联合后才能更接近真实天空。这一判断会在后续短间距、成像权重、主波束和校准章节中反复出现。


### 1.10.7 本节小结

单碟望远镜的收集面积与口径平方相关，而角分辨率近似按 $\lambda/D$ 缩放。由于射电波长很长，若只靠单碟获得高角分辨率，所需口径会很快进入工程上不可实现的范围。干涉阵通过离散阵元和相关处理把分辨率提升到最长基线尺度，这是现代高分辨率射电科学得以成立的关键。

但大型单碟仍然极其重要，因为它们能测总功率、恢复大尺度结构并提供零间距信息。现代射电观测不是“单碟对抗干涉阵”，而是两类系统互补配合。带着这一判断进入下一节，再看现代射电干涉阵的类型和设计差异，就更容易理解为什么不同阵列会在频率范围、阵元类型、视场和基线分布上做出完全不同的选择。
