## 9.20 偏振与法拉第诊断

第 9.9 节给出基础 IQUV 图、噪声、派生偏振量和泄漏上限；第 9.11 节给出校准证据、脏法拉第谱、RMSF 与算法验证。本节不再调整偏振掩膜或 RM-CLEAN 参数，而是讨论这些测量在何种前景、红移、几何和退偏振模型下能够约束磁场与热电子等离子体。

这一测量链比连续谱总强度测量更易受到系统误差影响。总强度 $I$ 的小幅标定误差通常表现为通量尺度偏差；偏振数据中同量级的误差却可能将未偏振强源泄漏为虚假的 $Q/U/V$ 信号。常量 EVPA 旋转会改变本征偏振角和投影磁场方向，但只有随频率变化的残余旋转才会直接偏置理想 RM 斜率。法拉第诊断还依赖 $\lambda^2$ 覆盖、通道平均和方向相关波束；这些环节未得到验证时，观测到的 RM 结构可能来自仪器效应或选择效应。


### 9.20.1 Stokes 参数与复偏振

线偏振通常用

$$
\mathcal P(\lambda^2)=Q(\lambda^2)+iU(\lambda^2)=L(\lambda^2)e^{2i\chi(\lambda^2)}
$$

表示，其中

$$
L=\sqrt{Q^2+U^2},\qquad p=\frac{L}{I},\qquad \chi=\frac{1}{2}\operatorname{atan2}(U,Q).
$$

偏振角前面的 $1/2$ 来自线偏振的自旋二性质：电场方向旋转 $180^\circ$ 后代表同一个线偏振状态，因此 $Q/U$ 平面中的相位为 $2\chi$。这也是偏振角出现 $n\pi$ 模糊的根源。若只将 $Q$ 和 $U$ 视为两幅普通图像，而没有保持复数 $\mathcal P=Q+iU$ 的相位关系，频率合成、平滑或掩膜操作便可能破坏其物理信息。偏振分数 $p=L/I$ 还要求图像具有共同波束、相容的频谱采样，且 $I$ 显著为正；它不是在任意两幅图上逐像素相除即可得到的无偏估计量。

馈源基也会影响校准过程的表述。在线偏振基中，平行手相关主要约束 $I$ 与馈源坐标中的 $Q$，交叉手相关携带 $U$ 与 $V$；在圆偏振基中，平行手相关主要约束 $I$ 与 $V$，交叉手相关携带 $Q$ 与 $U$。馈源基不改变天空 Stokes 参数，却会改变误差进入数据的路径。Stokes $V$ 是独立的带符号圆偏振量，不属于 $\mathcal P=Q+iU$；其手性约定、增益泄漏和波束偏斜（beam squint）必须单独验证。

EVPA 描述电场方向，并不直接等同于磁场方向。只有在光学薄同步辐射、本征 EVPA 已完成法拉第去旋转且不存在显著波束退偏振时，天球投影磁场方向通常取 EVPA 旋转 $90^\circ$；光学厚同步辐射、其他辐射机制或传播转换均可能改变这一关系。

**图 9.20.1：偏振观测量的派生链。** 图中左侧给出从 Stokes 数据立方到偏振物理量的派生链。中上图显示 $Q/U$ 平面中的复偏振向量，其相位对应 $2\chi$ 而不是 $\chi$。右上图显示理想外部法拉第屏的情形：$\chi$ 随 $\lambda^2$ 近似线性变化。下方面板强调，$I$、线偏振分数、偏振角和圆偏振 $V$ 描述不同的物理量；将它们合并为单一的“偏振强弱”会丢失关键信息。


### 9.20.2 偏振强度偏差和角度误差

由于 $L=\sqrt{Q^2+U^2}$ 永远非负，即使真实线偏振为零，带噪声的 $Q$ 和 $U$ 也会给出正的 $L$。在 $Q/U$ 独立、同方差高斯噪声下，$L$ 服从 Rice 分布；更一般的协方差情形仍有正定偏差，但不再是标准 Rice 分布。高信噪比时常用近似去偏差公式

$$
L_{\rm deb}\simeq \sqrt{\max(L^2-\sigma_{QU}^2,0)},
$$

但这个公式只在 $L/\sigma_{QU}$ 足够大且 $Q$、$U$ 噪声近似相同、独立、高斯时才有明确意义。低信噪比偏振测量更适合用似然、上限或后验分布表达，而不是把负根号截断为零后继续当作普通通量使用。

偏振角的不确定度也由偏振信噪比控制。令 $C_{QU}=\mathrm{Cov}(Q,U)$，一阶误差传播给出

$$
\sigma_\chi^2\simeq\frac{U^2\sigma_Q^2+Q^2\sigma_U^2-2QUC_{QU}}{4L^4}.
$$

在 $Q/U$ 独立、同方差且高信噪比时，它才化为常用近似

$$
\sigma_\chi \simeq \frac{1}{2}\frac{\sigma_{QU}}{L}\quad {\rm rad}.
$$

这些关系说明，偏振角精度不由总强度 $I$ 的信噪比直接决定，而由 $Q/U$ 中的偏振信号和协方差决定。一个明亮但偏振分数很低的源，偏振角可以比一个较暗但强偏振的源更不稳定。低信噪比时角度分布不再近似高斯，应回到 $Q/U$ 似然。RM 拟合还会组合各通道噪声、主波束和频率相关 leakage。


![莱斯偏差与偏振角不确定度](figures/practical_polarization_ricean_angle_uncertainty.png)

**图 9.20.2：莱斯偏差与偏振角不确定度。** 左图显示真实 $L=0$ 时测得的 $L$ 仍呈正分布，真实 $L=4\sigma$ 时测量分布也不以简单高斯方式围绕真值对称。右图给出独立同方差噪声下的高信噪比角度误差近似。常量系统角度偏差只平移 EVPA；跨频率的相对角度误差才会转化为 RM 斜率或曲率偏差。


### 9.20.3 校准误差引起的旋转、平移与假偏振

转换到天空 Stokes 后，残余校准误差常表现为两类效果：复线偏振旋转，或 Stokes $I$ 向偏振的加性 leakage。但它们在相关器数据中的来源依赖馈源基。圆偏振馈源的残余 R--L phase 可直接旋转 $Q/U$；线偏振馈源的残余 X--Y phase 主要混合 $U/V$，而平行手相对增益会影响 $I/Q$ 分离。若等效天空 EVPA 旋转随频率变化，它才会在 $\chi(\lambda^2)$ 中模拟或偏移 RM。

Cross-hand delay 是两个正交偏振通道之间的相对群延迟，会使交叉手相位随频率变化；常量 phase 的天空效应则由馈源基和 Stokes 变换决定。D-term 描述馈源非理想正交造成的 on-axis leakage，方向相关主波束还会产生 off-axis Mueller leakage。由于许多源的真实偏振只有几个百分点，百分之一量级的残余 leakage 已可能支配弱信号。

绝对 EVPA 校准将天空 $Q/U$ 坐标固定到声明的角度零点。缺少该校准时，部分偏振幅度仍可能可用，但本征 EVPA 和投影磁场方向不可信；与频率无关的角度零点不会改变理想的 RM 斜率。相反，交叉手延迟、频变泄漏、校准源自身的 RM、方向相关波束和电离层残差均可能制造频率结构，必须先依据第 9.11 节的校准证据排除，才能进入物理解释。


![偏振校准误差链](figures/practical_polarization_calibration_error_chain.png)

**图 9.20.3：偏振校准误差链。** 图中将偏振校准误差概括为天空 $Q/U$ 旋转和加性泄漏；这是转换到斯托克斯参量后的等效图像，并不表示所有馈源基中的交叉手相位都会直接旋转 $Q/U$。实际分析中，亮源附近的低偏振结构、主波束边缘的偏振增强或不同频段 RM 不一致，均应首先检查馈源基、交叉手、泄漏和波束环节。


### 9.20.4 法拉第深度、旋转量与 $\lambda^2$ 关系

Faraday rotation 来自磁化等离子体中左右圆偏振传播相位速度的差异。在常用单位下，沿视线的 Faraday depth 可写为

$$
\phi=0.812\int_{\rm source}^{\rm observer} n_e(\ell)\,B_\parallel(\ell)\,d\ell\quad {\rm rad\,m^{-2}},
$$

其中 $n_e$ 的单位为 $\mathrm{cm^{-3}}$，$B_\parallel$ 的单位为 $\mu\mathrm{G}$，路径长度单位为 pc；这里将积分方向取为从源到观测者，并规定指向观测者的磁场为正。采用其他路径方向或傅里叶符号约定时，$\phi$ 的符号会相应改变，必须随结果一并说明。$\phi$ 只约束 $n_eB_\parallel$ 的带符号路径积分；密度、磁场强度、路径长度和磁场反转彼此退化，单独一张 RM 图不能唯一确定磁场强度。

若偏振源位于一个外部、均匀的法拉第薄旋转屏之后，则

$$
\chi(\lambda^2)=\chi_0+{\rm RM}\,\lambda^2,
$$

且 RM 可以近似等同于该屏在观测坐标中的 Faraday depth。对红移为 $z_j$ 的薄屏，静止系贡献按 $\phi_j/(1+z_j)^2$ 进入观测 RM。多个简单前景近似相加时，

$$
{\rm RM}_{\rm obs}\simeq {\rm RM}_{\rm iono}+{\rm RM}_{\rm MW}+\sum_j\frac{\phi_j}{(1+z_j)^2}.
$$

因此把源内贡献还原为 $({\rm RM}_{\rm obs}-{\rm foreground})(1+z_{\rm src})^2$ 只适用于前景已估计且主导剩余旋转位于源红移处的简化情形。若发射区和旋转介质混合，或视线中存在多个不同红移的分量，单一修正因子没有这种含义，$\chi$ 也不再严格线性。

更一般的表达是

$$
\mathcal P(\lambda^2)=\int_{-\infty}^{+\infty} F(\phi)\,e^{2i\phi\lambda^2}\,d\phi,
$$

其中 $F(\phi)$ 为法拉第色散函数。RM 综合（RM synthesis）试图利用有限、离散且含噪声的 $\mathcal P(\lambda^2)$ 采样约束该函数；第 9.11 节已经说明，直接变换得到的是与 RMSF 卷积后的脏法拉第谱，而不是唯一的反演结果。


### 9.20.5 RMSF 对物理解释的限制

第 9.11 节已定义 RMSF，并由经过数据标记的实际 $\lambda^2$ 采样计算法拉第分辨率、最大敏感厚度和通道平均限制。本节只讨论这些算法量对物理解释的限制：

| RMSF/采样性质 | 允许的物理结论 |
| --- | --- |
| 主瓣 FWHM $\delta\phi$ | 间隔小于或接近主瓣的分量通常不可唯一分离；一个峰不等于一个薄屏 |
| 旁瓣 | 次峰可能来自采样混叠；需比较 dirty/恢复谱、空白视线和注入结果 |
| 最大敏感法拉第厚度 | 更宽的 $F(\phi)$ 会被衰减或解析掉；未见宽结构不等于介质不存在 |
| 通道平均响应与 $|\phi|$ 范围 | 高法拉第深度信号可能在单通道内退偏振，形成深度相关选择函数 |
| $\phi$ 网格与搜索范围 | 细网格只过采样响应；大范围找峰需要考虑相关 bins 和搜索试验 |

即使两个峰间隔超过名义 FWHM，低信噪比、强旁瓣和不等振幅仍可能使分解不稳定。物理分量数应通过观测空间的 $Q/U(\lambda^2)$ 残差、模型比较和模拟恢复共同判断，不能只数 RM-CLEAN components。


![法拉第旋转测量函数](figures/practical_faraday_rmsf_lambda2_coverage.png)

**图 9.20.4：法拉第旋转测量函数。** 图中比较了宽连续、窄带和带缺口的 $\lambda^2$ 采样。较宽覆盖产生较窄的 RMSF 主瓣；窄覆盖使法拉第分辨率降低；采样缺口会增强旁瓣，使多个法拉第分量更容易相互污染。RM 综合不是简单的频率维傅里叶变换，而是具有明确采样函数和响应函数的反演问题。


### 9.20.6 复杂法拉第结构与退偏振

法拉第薄分量在 $F(\phi)$ 中近似为窄峰；法拉第厚结构在 $\phi$ 空间中具有有限宽度。多个薄分量可能在 RMSF 主瓣内混合为单个峰，也可能通过旁瓣产生假峰。法拉第厚结构在长波长处容易发生退偏振，因此总偏振强度下降不一定表示磁场变弱，也可能说明视线内或波束内存在较强的法拉第深度弥散。

退偏振可以来自多种机制。Bandwidth depolarization 是一个通道内部旋转过多，导致通道平均后 $Q/U$ 抵消。Beam depolarization 是一个合成波束内包含不同 RM 或本征角的结构，空间平均后偏振减弱。Internal Faraday rotation/dispersion 则来自发射区和旋转介质混合，使不同深度的辐射相互抵消。

两个简化模型能帮助区分量级。以下把内禀频谱趋势吸收到 $\mathcal P_0$ 或预先校正；若 beam 内外部 Faraday depth 近似服从均值 $\phi_0$、标准差 $\sigma_\phi$ 的高斯分布，则 Burn 外部色散模型为

$$
\mathcal P(\lambda^2)=\mathcal P_0\exp(2i\phi_0\lambda^2)\exp(-2\sigma_\phi^2\lambda^4).
$$

若发射和旋转均匀混合成总 Faraday 厚度为 $\Phi$ 的理想 slab，则

$$
\mathcal P(\lambda^2)=\mathcal P_0\,\operatorname{sinc}(\Phi\lambda^2)\exp(i\Phi\lambda^2),\qquad \operatorname{sinc}x=\frac{\sin x}{x}.
$$

前者给出近似 $\lambda^4$ 指数衰减，后者产生零点和振荡；有限频带内二者与多 thin 分量仍可能退化。比较模型前还必须把通道平均、频变 $uv$ 覆盖、共同 beam、主波束 leakage 和总强度谱纳入前向预测，否则仪器分辨率变化也会伪造偏振分数随频率下降。


![复杂法拉第结构](figures/practical_faraday_complexity_interpretation.png)

**图 9.20.5：复杂法拉第结构。** 左图显示多个法拉第分量和一个法拉第厚成分经过 RMSF 后混合为脏法拉第谱。右图比较单一薄屏、多分量结构和法拉第厚介质的 $|\mathcal P(\lambda^2)|$ 行为。RM-CLEAN 可以减弱 RMSF 旁瓣，但不能自动使物理模型唯一化；当 $\lambda^2$ 覆盖有限、信噪比不高或存在法拉第厚结构时，多个模型可能同样符合数据。


### 9.20.7 与真实软件工作流的对应

CASA 等校准/成像工具、CARTA 等检查工具以及 RM-Tools、pyrmsynth 等 Faraday 工具分别对应 9.9 和 9.11 的数据产品与算法层。本节使用它们的版本化输出，不以更换 RM-CLEAN 参数来追求更容易解释的物理图像。若结果来自 pipeline，也必须引用校准源模型、parallactic angle、leakage、频带 flag、Mueller beam 和电离层证据。

物理解释报告应引用 9.9 的 IQUV/noise/leakage 身份和 9.11 的校准/RMSF/dirty spectrum，而不复制两章清单。本节新增的记录是：电离层、银河系和其他前景处理；各 Faraday 屏的红移；EVPA 到磁场方向的辐射机制假设；$n_e$、路径长度、场反转和 filling factor 的退化；外部色散、slab 或多分量模型；beam/bandwidth depolarization；模型比较、适用频带和不能排除的替代解释。


### 9.20.8 本节结论

偏振分析的核心对象是复线偏振 $\mathcal P=Q+iU$，而 Stokes $V$ 需单独解释。线偏振强度、EVPA、偏振分数和 RM 都从带噪声、受校准和采样影响的 Stokes 产品派生。$L$ 的正定偏差、$\chi$ 的 $n\pi$ 模糊、馈源基相关误差、leakage 和有限 $\lambda^2$ 覆盖，使偏振图比总强度图更依赖完整误差叙述。

本节的交付物不是一张去噪 RM 图，而是带前景与红移约定、统计和系统误差、RMSF 选择效应、退偏振模型及替代解释的物理限制。Faraday depth 约束的是 $n_eB_\parallel$ 路径积分；从它推断磁场仍需独立电子密度、几何和辐射机制信息。教材训练的目标是知道数据实际约束了什么，也知道哪些磁场叙述仍不能由当前频带唯一确定。
