## 1.4 射电波段<a id='science:sec:radio_regime'></a>


传统的天文图景主要建立在可见光观测之上，包括恒星、行星、银河和明亮星云。然而，可见光只占电磁波谱中很窄的范围，不同波段对应不同的能量尺度、辐射机制和天体环境。完整理解天体及其演化过程，需要综合比较各波段所提供的互补信息。

射电波段对冷气体、磁场、相对论粒子、传播介质和时间变化现象具有独特的诊断能力。因此，射电天文学并非光学天文学在低能端的简单延伸，而是研究另一组物理成分和过程的重要途径。射电波段的定义不仅涉及频率和波长范围，还涉及该波段能够测量的物理量及其对应的观测条件。

从数量级上看，地面射电天文学常处理从兆赫到太赫兹附近的电磁波，对应从数十米到亚毫米的波长。由 $\lambda\nu=c$ 可知，同一观测频段既可用频率表示，也可用波长表示。单个射电光子的能量较低，$h\nu/k\simeq 0.048\,{\rm K}\,(\nu/{\rm GHz})$；在 $1$ GHz 和 $100$ GHz 处，光子能量对应的温度分别约为 $0.05$ K 和 $5$ K。射电接收系统因而通常以电场、电压或功率的连续测量为基础，并进一步获得频谱、偏振和相干性信息。这一探测特点是理解接收机噪声、相关器和可见度的物理基础。

### 1.4.1 射电天空与光学天空的差异

伽马射线和 X 射线主要探测高温等离子体、激波与极端引力环境；可见光观测常直接记录恒星光球和电离气体的辐射；射电波段则对冷气体、非热连续谱、低能跃迁和传播效应较为敏感。不同波段具有不同的物理选择效应，因此同一天体在多波段图像中可以呈现显著不同的形态。

射电图像中的亮结构不一定对应光学图像中最明亮的恒星或星系主体，其主要辐射区可能是喷流、射电瓣、冷气体盘、磁层或受尘埃遮蔽的恒星形成区。这种差异来自波段选择的物理过程，而非图像显示方式：射电观测使磁场、低温气体、等离子体传播效应和相干辐射过程成为可测量对象。下表列出三个具有代表性的科学窗口；相关辐射机制和天体案例将在 1.5--1.8 节进一步讨论。

### 1.4.3 射电窗口及其物理边界

地基射电观测的可用频率范围受到地球电离层和中性大气的共同限制。二者具有显著的频率选择性，从而在低频端和高频端分别形成不同的观测边界。

低频端的主要限制来自电离层。电离层中存在自由电子，它们使电磁波在传播时具有等离子体色散关系。当频率接近或低于等离子体频率

$$
\nu_p={1\over 2\pi}\left({n_e e^2\over m_e\epsilon_0}\right)^{1/2}
\simeq 9\,\sqrt{n_e/{\rm cm}^{-3}}\ {\rm kHz},
$$

电磁波就难以像在真空中那样穿过介质，而会发生强烈折射、反射或吸收。实际电离层电子密度随高度、昼夜、太阳活动和地磁条件变化，因此地基射电天文学的低频边界不是一个固定常数；但在数量级上，十几 MHz 以下的天文观测通常十分困难，更低频段的宇宙信号需要借助空间平台或月球背面环境进行研究。

高频端的限制主要来自中性大气，尤其是水汽和氧气分子的吸收线及其连续不透明度。毫米和亚毫米波段的光学深度对可降水汽含量十分敏感，因此观测站址通常选择高海拔、低水汽且气象条件稳定的地区。射电窗口的低频端主要受等离子体传播控制，高频端主要受分子吸收和大气热辐射控制。图 1.4.2 给出了大气透过率随频率变化的示意。


<img src="figures/atmospheric_radio_windows.png" alt="Schematic atmospheric radio transmission" width="800">


**图 1.4.2**：地面射电窗口的概念化天顶透过率。低频端受电离层截止限制，高频端受水汽连续吸收和分子谱线影响；实际透过率强烈依赖海拔、可降水量、天气与仰角，因此曲线只用于展示结构和数量级。<a id='science:fig:atm_opa'></a> <!--\label{science:fig:atm_opa}-->


射电窗口并非透过率均匀的连续频段，而是由多个大气透明度不同的子窗口组成。地基射电天文学的主要工作频率大致位于 $10$ MHz 至 $1$ THz；更高频率逐渐进入远红外和红外波段，更低频率则受到电离层的强烈限制。

这种频率选择性直接影响望远镜设计和站址选择。低频阵列需要处理电离层相位扰动、强射频干扰以及大视场引起的方向相关校准问题；厘米波段具有较稳定的地面观测条件，相关的干涉测量技术体系较为成熟；毫米和亚毫米阵列则通常建设在高海拔、低水汽环境，例如阿塔卡马高原。因此，观测频段在很大程度上决定了主要传播介质、噪声背景和校准误差来源。

大气透明也不等于频谱在工程上可用。广播、移动通信、卫星下行、雷达和望远镜自身电子设备都会产生射频干扰（RFI），其强度可比天文信号高许多个数量级。国际与国家频谱管理为若干射电天文频段提供保护，但保护带只覆盖部分频率，也不能自动消除带外泄漏、卫星旁瓣或非法发射。因此，选频还要同时检查目标谱线的红移范围、当地和卫星 RFI 环境、保护频段及接收机动态范围；偏远站址、射电静默区、实时监测和后处理旗标都是观测系统的一部分。


### 1.4.4 波段命名与观测语言

射电波段的字母命名具有明显的历史沿革，其字母顺序与频率顺序并不完全一致。这些名称源于雷达、通信和射电天文的不同技术传统，随后沿用于望远镜接收系统。波段代号不是严格的自然边界，而是对频率范围、典型接收机配置和常见科学任务的约定性简称。

下面列出一些常见波段。不同台站、接收机和文献中可能采用略有差别的边界，但表中的数量级已经足以支持本书后续讨论。

| L | 1 - 2 |
| S | 2 - 4 |
| C | 4 - 8 |
| X | 8 - 12 |
| Ku | 12 - 18 |
| K | 19 - 26.5 |
| Ka | 26.5 - 40.0 |
| Q | 40.0 - 50.0 |


**表 1.4.1**：射电天文学中常见的观测波段。<a id='science:tab:bands'></a> <!--\label{science:tab:bands}-->

例如，文献中的 L 波段通常指约 $1$--$2$ GHz；毫米波观测则位于更高频率，对大气条件、相位稳定性和天线表面精度更为敏感。使用波段代号时，仍应明确具体频率范围，并结合波长、角分辨率、主波束、接收机噪声和科学目标解释其观测含义。


### 1.4.5 本节小结

射电波段对低能、低温、磁化和等离子体环境具有独特的观测灵敏度，使冷气体、喷流、射电瓣、传播介质效应和相干辐射源等光学波段不易探测的成分进入可观测范围。相应的主要测量量包括气体速度场、谱指数、亮温、偏振和时间变化。

射电观测同时受到大气窗口、传播介质和目标角尺度的限制。电离层与水汽分别构成低频和高频观测的重要约束，目标的空间尺度则影响单碟望远镜与干涉阵的适用范围。后续各节将在此基础上讨论热辐射、同步辐射和谱线辐射等主要物理机制。
