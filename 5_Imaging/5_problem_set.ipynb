# 5.P 成像综合问题集

本问题集检验能否把空间频率、采样、网格化、权重和宽场效应转化为可核查的成像决策。除数值结果外，每题都要求写出采用的近似、单位和判断依据；只给软件参数名而不说明物理理由，不能得到满分。

总分 100 分。建议先独立完成前五题，再使用末尾的参考量和评分要点自检。

## 1. 从阵列尺度到图像网格（20 分）

某 25 m 口径阵列在 $1.5\,\mathrm{GHz}$ 观测，最长和最短有效基线分别为 $36\,\mathrm{km}$ 与 $680\,\mathrm{m}$。采用

$$
\theta_{\rm res}\simeq {\lambda\over B_{\max}},\qquad
\theta_{\rm PB}\simeq 1.02{\lambda\over D},\qquad
\theta_{\rm MRS}\simeq 0.6{\lambda\over B_{\min}}.
$$

1. 计算波长、名义分辨率、主波束 FWHM 和最大可恢复尺度，角度分别用角秒或角分表示。
2. 若每个合成波束用 3 至 5 个像素采样，给出合理的 cell 范围。
3. 若要覆盖一个主波束直径，估算图像边长并选择一个便于 FFT 的 imsize。
4. 说明增大 imsize 为什么不能恢复超过 $\theta_{\rm MRS}$ 的结构。

## 2. 权重不是审美选择（20 分）

同一数据得到下列三组结果：

| 权重 | 恢复波束 | RMS | PSF 最大旁瓣 |
|:--|:--|--:|--:|
| natural | $1.8''\times1.4''$ | $8\,\mu\mathrm{Jy\,beam^{-1}}$ | 0.18 |
| Briggs | $1.3''\times1.0''$ | $10\,\mu\mathrm{Jy\,beam^{-1}}$ | 0.11 |
| uniform | $0.95''\times0.75''$ | $16\,\mu\mathrm{Jy\,beam^{-1}}$ | 0.09 |

1. 对一个未分辨、峰值约 $80\,\mu\mathrm{Jy}$ 的点源，分别计算三幅图的峰值信噪比，并选择检测图。
2. 对尺度约 $6''$ 的低面亮度晕，说明为什么点源 RMS 最低不等于积分通量一定最可靠。
3. 设计一组至少包含两幅图的成像产物，并说明每幅图承担的科学测量任务。
4. 列出比较这些图时必须共同报告的三个 QA 量。

## 3. 网格化的计算量与误差（20 分）

一个卷积网格器使用 $7\times7$ 像素支撑的可分离卷积核，过采样因子为 64，需要处理 $2\times10^7$ 个可见度样本。

1. 每个偏振和频率平面约需多少次加权网格累加？利用可分离性后，一维核权重的求值或查表次数可降到多少，为什么网格写入次数并不会同样下降？
2. 说明过采样因子 64 为什么主要增加查找表大小和分数像素精度，而不应把每个样本的网格支撑扩大 64 倍。
3. 分别说明核支撑过小、过采样不足和忘记图像域 gridding correction 会产生什么类型的误差。
4. 给出一个不用真实天空模型也能执行的 gridder/degridder 一致性测试，并写出采用 NumPy FFT 归一化时应满足的伴随内积关系。

## 4. $w$ 项何时不可忽略（20 分）

小角度下，忽略非共面项造成的相位误差量级可写成

$$
|\Delta\phi_w|\simeq \pi |w|\theta^2.
$$

某数据的 $|w|_{\max}=10^5$ 波长。

1. 若要求 $|\Delta\phi_w|<0.1\,\mathrm{rad}$，计算允许的视场半径。
2. 计算半径 $15'$ 处的相位误差，并判断普通二维 FFT 是否足够。
3. 比较 faceting、W-stacking 和 W-projection 分别在什么数据特征下较有吸引力。
4. 说明只检查目标源位置，而忽略视场内离轴强源，为什么仍可能得到错误结论。

## 5. 从科学目标反推成像方案（20 分）

一组 $1$ 至 $2\,\mathrm{GHz}$ 连续谱数据同时包含 $1.2''$ 紧致核、约 $30''$ 的弥散晕，以及距离相位中心 $12'$ 的强源。科学目标是测量紧致核位置、弥散晕积分通量和带内谱指数。

写出一份不超过 15 行的成像方案，至少明确：

- cell、视场和图像大小的选择依据；
- 至少两组 weighting/taper 产物及各自用途；
- 是否需要宽场与宽带算法，以及依据；
- deconvolver、尺度列表、mask 和停止准则；
- 必须保存的 PSF、残图、主波束和频谱 QA 产品；
- 如何检验 $30''$ 结构接近最大可恢复尺度时的通量损失；
- 哪些失败迹象会使谱指数或弥散通量只能作为上限。

## 参考量与评分要点

**第 1 题：** $\lambda\simeq0.200\,\mathrm{m}$，$\theta_{\rm res}\simeq1.15''$，$\theta_{\rm PB}\simeq28.0'$，$\theta_{\rm MRS}\simeq36.4''$。cell 约为 $0.23''$ 至 $0.38''$；若取 $0.30''$，一个主波束约需 5600 像素，可选择 6144 或更大的高效尺寸。公式、单位、数值和 MRS 解释各占 5 分。

**第 2 题：** 三幅图的点源信噪比分别约为 10、8、5，natural 图最适合检测；弥散源还受每波束表面亮度、短基线权重、波束相关噪声和缺短间距影响。数值 6 分，物理解释 6 分，双产品设计 5 分，QA 量 3 分。

**第 3 题：** $7\times7$ 支撑仍需约 $9.8\times10^8$ 次加权网格累加；可分离性把一维 tap 的求值或查表降到约 $2.8\times10^8$ 次，再组合成二维权重，但不能消除 49 个网格点的更新。误差分析 8 分，计算量 6 分，过采样解释 3 分，一致性测试 3 分。可接受的测试包括点源往返、与直接 DFT 的小规模对照；最有诊断力的是 $\langle G\mathbf v,\mathbf x\rangle=\langle\mathbf v,D\mathbf x\rangle/(N_lN_m)$ 的伴随性测试。

**第 4 题：** 0.1 rad 限制给出 $\theta\simeq1.94'$；在 $15'$ 处误差约为 $5.98\,\mathrm{rad}$，不能忽略。两个数值各 5 分，算法比较 6 分，离轴源解释 4 分。

**第 5 题：** 没有唯一参数表。满分方案必须把每个参数连接到一个科学量或误差项，并明确高分辨率图与低面亮度图的职责不同；必须讨论宽带主波束、$w$ 项、MT-MFS 模型边界，以及用多配置/短间距数据、注入源或不同 taper 检验通量恢复。遗漏 QA 与失败降级规则时最高不超过 12 分。
