# 第1章：利用干涉阵开展射电科学<a id='science:sec:intro'></a>

本章介绍射电干涉测量所依托的物理概念、科学问题和观测系统。射电观测能够探测冷气体、磁场、相对论粒子、等离子体传播以及快速时变现象；这些信息分别通过连续谱、谱线、偏振和时间结构表现出来。许多天体同时包含从致密核心到大尺度弥散结构的多层次空间尺度，因此观测系统必须在角分辨率、面亮度灵敏度、频谱和时间分辨率以及总通量恢复之间进行取舍。

本章按照“物理过程—观测量—仪器方法”的顺序展开。首先介绍电磁辐射的基本测量量和辐射转移，再讨论热连续谱、同步辐射、谱线辐射及典型射电源；随后比较单碟望远镜与干涉阵，说明可见度、孔径合成和阵列设计分别解决什么问题。完整的坐标、测量方程、成像和校准推导留到后续章节处理。本章要建立的是一条可复用的判断链：由科学问题确定观测量，再由观测量确定频率、基线、分辨率和校准需求。


本章不对后续理论作完整推导，而是给出理解这些推导所需的科学语境。完成本章后，应能形成三项基本判断：不同辐射机制由哪些观测量诊断；单碟与干涉阵分别保留哪些空间尺度和信息；科学目标如何约束阵列的频率、基线、分辨率与校准要求。这些判断构成后续傅里叶变换、采样、成像和校准章节的物理前提。


#### 本章学习目标

完成本章后，读者应当能够：

1. 区分光度、通量密度、比强度、亮温和图像中的 $\mathrm{Jy\,beam^{-1}}$，并说明各自适合回答什么问题；
2. 用辐射转移、谱指数、谱线轮廓、偏振和时间变化判断主要辐射机制及其适用边界；
3. 把科学目标翻译成频率覆盖、角分辨率、频谱/时间分辨率、偏振和空间尺度响应需求；
4. 区分角分辨率、主波束、最大可恢复尺度、点源灵敏度和面亮度灵敏度；
5. 说明单碟、连接阵和 VLBI 各自保留或丢失什么信息，并据此给出基本观测方案。


#### 本章概念地图

本章可以读成一条从科学问题到干涉测量方法的链条。这里不要求一开始就掌握每个技术细节，但需要先看清概念之间的方向关系：先问宇宙中有什么物理过程值得研究，再决定要测哪些辐射量，随后才会发现单碟在角分辨率和空间尺度响应上不够用，进而自然走向干涉阵与孔径合成。

#### 本章结构

第 1.1 到 1.8 节先从射电科学本身出发，介绍本章后续所有观测讨论所依赖的物理背景，包括辐射测量量、辐射转移、热与非热连续谱、谱线辐射以及典型天文射电源。第 1.9 到 1.11 节则把重心转向仪器与方法，说明干涉测量的物理直觉、单碟与干涉阵的关系，以及现代射电阵列的主要类型；第 1.12 节用定量问题检验这些概念能否转化为可核查的观测判断。这样安排，是为了在进入后续正式理论之前，先把“为什么要这样测”“测出来的量代表什么”“不同阵列为什么会有不同设计”这三件事放在同一幅图景里。

本章采用两层阅读路径。**核心路径**关注定义、诊断关系和观测设计判断；标明“选读”“拓展内容”或位于附录中的内容，用于连接宇宙学、完整辐射过程、仪器工程和专题实践，不作为首次阅读的先决条件。若希望严格按“科学需求 $\rightarrow$ 单碟边界 $\rightarrow$ 干涉测量实现”的概念顺序学习，可以在读完 1.8 后先读 1.10，再回到 1.9；编号顺序则保留现有课程结构。

读完本章之后，第一册会按方法主线展开：第 2 章介绍干涉测量所需的数学工具，第 3 章讨论位置天文学与坐标系统，第 4 章进入可见度空间，第 5 至第 6 章讲成像与去卷积，第 7 章讨论观测系统，第 8 章讨论校准；第二册再把这些原理组织成实践工作流。换句话说，第 1 章负责建立科学与仪器背景，后续章节和实践册负责把这套背景转化为完整的干涉测量方法论。
