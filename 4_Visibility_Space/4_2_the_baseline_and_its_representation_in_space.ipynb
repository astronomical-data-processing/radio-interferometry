## 4.2 基线及其空间表示<a id='visibility:sec:baseline_space'></a>


上一节建立了干涉条纹与空间频率之间的关系。本节进一步讨论干涉测量的几何基础。基线是连接两台天线的三维有向量，其长度和方向由阵列布局及天线顺序共同确定。

对于给定的天源方向，干涉仪的空间频率坐标由物理基线在相位中心局部坐标架中的投影确定。地球自转过程中，地面物理基线近似保持不变，但它相对于天源方向的投影不断变化，由此形成 $uv$ 轨迹，并增加傅里叶平面中的采样。

本节依次在本地坐标、地固赤道坐标和相位中心局部坐标中表示同一条基线。本地坐标给出天线之间的物理分离，地固赤道坐标将基线与地球自转轴联系起来，$(u,v,w)$ 坐标则给出基线相对于相位中心的测量坐标。


坐标变换过程如下：首先在本地直角坐标中以两台天线的位置差定义物理基线；随后将坐标轴对齐到东、北和天方向，得到局部东-北-天（ENU）坐标；再利用台站纬度将基线转换到与地球自转轴相关的 $(X,Y,Z)$ 坐标架；最后以相位中心方向为 $w$ 轴，转换到 $(u,v,w)$ 坐标架。各步变换只改变基线的坐标分量，不改变物理基线本身。


<a id="vis:sec:421"></a> <!---\label{vis:sec:421}--->
### 4.2.1 物理基线

物理基线是两个天线单元在三维空间中的有向几何分离量。投影基线则是物理基线在相位中心局部坐标架中的分量。在国际地球参考架或局部台站坐标架中，物理基线可近似视为常量；对于固定在天球参考架中的天源，投影基线会随地球自转而变化。

高精度台站坐标还需考虑板块运动、固体地球潮汐、海潮负荷、大气负荷、天线参考点和参考架历元。极移及其他地球定向参数主要进入地球参考系统与天球参考系统之间的姿态变换。本章将台站坐标视为固定量，重点讨论坐标变换和投影关系。


#### 4.2.1.1 本地基线向量
<a id="vis:sec:4211"></a> <!---\label{vis:sec:4211}--->

设本地直角坐标架 $\mathcal{R}$ 的原点为 $O$，正交基底为 $\mathcal{B}=(\mathbf{\hat e}_x,\mathbf{\hat e}_y,\mathbf{\hat e}_z)$。两台天线的位置向量分别为

$$
\mathbf{a}_1=\vec{OA}_1=x_1 \mathbf{\hat{e}_x} + y_1 \mathbf{\hat{e}_y} + z_1 \mathbf{\hat{e}_z},
$$

$$
\mathbf{a}_2=\vec{OA}_2=x_2\mathbf{\hat{e}_x} + y_2 \mathbf{\hat{e}_y} + z_2 \mathbf{\hat{e}_z}.
$$

本书把与相关积 $V_{12}=\langle v_1v_2^*\rangle$ 配对的有向基线约定为

$$
\mathbf{b}_{12} = \mathbf{a}_1 - \mathbf{a}_2 = \vec{A_2A_1}.
$$

该方向约定必须与后文的负指数傅里叶约定配套使用。其他资料可能采用相反的基线方向，并相应改变指数符号。交换天线顺序后有 $\mathbf b_{21}=-\mathbf b_{12}$；在理想标量条件下，同时满足 $V_{21}=V_{12}^*$。紧凑阵列可以使用统一的本地测量坐标，长基线阵列则必须先将两台站坐标统一到同一 ITRF 版本和历元，计算 ECEF 差向量后再转换到所需的局部坐标架。

**图 4.2.1：本地三维直角坐标系中的天线位置向量 $\mathbf a_1$ 与 $\mathbf a_2$。**

#### 4.2.1.2 $(E,N,U)$ 局部东-北-天坐标架

在前面的本地表达式中，$\mathbf{\hat e}_x$、$\mathbf{\hat e}_y$ 和 $\mathbf{\hat e}_z$ 的方向仍是任意指定的。通常采用东-北-天坐标架（east-north-up，ENU）：$E$ 指向大地东，$N$ 指向大地北，$U$ 沿参考椭球法线向外。原点可以取阵列中心，也可以取某一面参考天线。这里的 $U$ 是大地天方向；若进行角秒级指向计算，还需区分参考椭球法线与受重力场影响的天文铅垂线。

由此，基线向量仍在台站附近的局部切平面中表示，但三个坐标分量具有明确的地理含义。紧凑阵列通常可以直接使用 ENU 坐标描述天线布局；跨大陆 VLBI 阵列则应先在地心地固坐标中描述台站位置，再转换到各台站或阵列中心附近的局部坐标架。


#### 4.2.1.3 地平坐标系 $(\mathcal{A}, \mathcal{E})$ 中的基线表示


地面观测者通常用方位角 $\mathcal{A}$ 和高度角 $\mathcal{E}$ 描述天空方向；同样，基线的空间方向也可用这两个角表示。以下将有向基线的起点作为地平坐标原点，并沿用 $\mathbf b_{12}=\vec{A_2A_1}$ 的约定，即基线方向由天线 2 指向天线 1。$E$、$N$ 和 $U$ 的方向沿用前述定义。

采用本书的方位角约定：$\mathcal{A}$ 从北向东增加，$\mathcal{E}$ 从地平线向天顶增加。若基线长度为 $|\mathbf{b}|$，则它在东-北-天（ENU）坐标中的分量为

$$
\mathbf{b}_{\rm ENU}
=
|\mathbf{b}|
\begin{bmatrix}
\sin \mathcal{A} \cos \mathcal{E}\\
\cos \mathcal{A} \cos \mathcal{E}\\
\sin \mathcal{E}
\end{bmatrix}.
$$

矩阵的三行依次给出东、北和天方向分量。若两台天线处于同一水平面，基线的 $U$ 分量为零，对应 $\mathcal{E}=0^\circ$。


<a id="vis:fig:422"></a> <!--\label{vis:fig:422}--></a>
<img src='figures/enu_alt_az.png' width=40%>


**图 4.2.2：地平坐标 $(\mathcal{A},\mathcal{E})$ 与直角坐标 $(E,N,U)$ 之间的关系。**

<div class=warn>
<b>约定说明：</b> 不同软件或工程文档可能采用不同的方位角零点和正方向。本章采用从北向东增加的约定；程序实现时必须明确记录并保持方位角定义一致。
</div>


#### 4.2.1.4 地固赤道坐标架 $(X,Y,Z)$ 中的基线表示

现在将 ENU 中的基线转换到与地球自转轴相关的地固赤道坐标架 $(X,Y,Z)$。设阵列参考点的大地纬度为 $\varphi_a$：$Z$ 轴指向约定地球北极，$X$ 轴位于赤道面内并指向本地子午圈与赤道面的交点，$Y$ 轴补成右手系。该阵列中心化坐标架随地球转动，不是 ICRS/GCRS 惯性参考架；它用于将固定在地面的基线接入时角变换。

ENU 坐标面与天赤道面之间的相对取向由台站纬度决定。若只描述基线向量在 $(X,Y,Z)$ 坐标架中的方向，可形式化地引入球面角 $(H_b,\delta_b)$，其中 $H_b$ 向西为正；这两个角仅用于参数化基线方向，不表示真实天源。后文的 $(H_0,\delta_0)$ 则分别表示相位中心的时角和赤纬。


<a id="vis:fig:413"></a> <!--\label{vis:fig:413}--></a>
<img src='figures/xyz_XYZ.png' width=40%>


**图 4.2.3：地固赤道坐标架 $(X,Y,Z)$。** $T$ 表示当地天顶，$(X,Y)$ 位于垂直于地球自转轴的赤道面内，$(N,S,E,W)$ 所在平面为天线局部地平面。

将 $(\mathcal{A},\mathcal{E},\varphi_a)$ 转换到 $(X,Y,Z)$ 坐标架后，基线可写为

$$
\begin{bmatrix}
X\\
Y\\
Z
\end{bmatrix}
=
|\mathbf{b}|
\begin{bmatrix}
\cos \varphi_a \sin \mathcal{E} - \sin \varphi_a \cos \mathcal{E} \cos \mathcal{A}\\
\cos \mathcal{E} \sin \mathcal{A} \\
\sin \varphi_a \sin \mathcal{E} + \cos \varphi_a \cos \mathcal{E} \cos \mathcal{A}
\end{bmatrix}.
$$

同一个向量也可以写成赤道形式

$$
\begin{bmatrix}
X\\
Y\\
Z
\end{bmatrix}
=
\begin{bmatrix}
|\mathbf{b}| \cos \delta_b \cos H_b\\
-|\mathbf{b}| \cos \delta_b \sin H_b\\
|\mathbf{b}| \sin \delta_b
\end{bmatrix}.
$$

两种写法描述同一条基线：前者采用方位角、高度角和台站纬度，后者采用基线方向的球面参数 $(H_b,\delta_b)$。二者的等价关系可由第 3 章介绍的天文三角形推导。

<div class=advice>
<b>说明：</b> 这个转换本质上是把局部地平坐标绕东西方向倾斜到天赤道坐标。完整球面三角推导可参考附录 [A.1](../0_Introduction/2_Appendix.ipynb)。
</div>


### 4.2.2 $(u,v,w)$ 空间

现在将基线从 $XYZ$ 坐标架转换到与相位中心直接相关的 $(u,v,w)$ 坐标架。设相位中心方向为 $\mathbf s_0$，在观测时刻的时角和赤纬分别为 $(H_0,\delta_0)$。定义 $w$ 轴指向 $\mathbf s_0$，$u$ 轴位于局部切平面内并指向赤经增加方向，$v$ 轴补成右手系并指向赤纬增加方向。于是 $(u,v)$ 张成垂直于相位中心方向的局部平面。

$u$ 和 $v$ 是成像测量方程中的二维空间频率坐标，$w$ 记录基线沿相位中心视线方向的分量。在窄视场且 $w(n-1)$ 可忽略时，测量方程可近似为二维傅里叶变换；宽视场条件下必须保留 $w$ 项。


从 $XYZ$ 到 $(u,v,w)$ 的变换可以看成两次旋转。先绕 $Z$ 轴旋转，使新坐标系的一个轴对齐到相位中心所在时角；再绕新的 $X'$ 轴旋转，使 $w$ 轴指向相位中心。图 4.2.4 给出了这组旋转的几何关系。


<a id="vis:fig:414"></a> <!--\label{vis:fig:414}--></a>
<img src='figures/coordinates2b.png' width=50%>


**图 4.2.4：$(X,Y,Z)$ 坐标架与 $(u,v,w)$ 坐标架之间的旋转关系。**

由 $(X,Y,Z)$ 到 $(u,v,w)$ 的变换可写为

$$
\begin{pmatrix}
u\\
v\\
w
\end{pmatrix}
=
\mathcal{R}_{X'}^{\frac{\pi}{2}-\delta_0}
\mathcal{R}_{Z}^{\frac{\pi}{2}-H_0}
\begin{pmatrix}
X\\
Y\\
Z
\end{pmatrix}.
$$

按照线性代数中的通常约定，右侧旋转先作用。因此先执行绕 $Z$ 轴的旋转，再执行绕 $X'$ 轴的旋转。


第一步旋转矩阵为

$$
\mathcal{R}_{Z}^{\frac{\pi}{2}-H_0}
=
\begin{pmatrix}
\sin H_0 & \cos H_0 & 0 \\
-\cos H_0 & \sin H_0 & 0 \\
0 & 0 & 1
\end{pmatrix}.
$$


第二步旋转矩阵为

$$
\mathcal{R}_{X'}^{\frac{\pi}{2}-\delta_0}
=
\begin{pmatrix}
1& 0& 0\\
0 & \sin \delta_0 & \cos \delta_0  \\
0 &-\cos \delta_0 & \sin \delta_0
\end{pmatrix}.
$$


两次旋转相乘后，得到常用的 $XYZ\rightarrow uvw$ 转换矩阵：

$$
\begin{pmatrix}
u\\
v\\
w
\end{pmatrix}
=
\begin{pmatrix}
\sin H_0 & \cos H_0 & 0\\
-\sin \delta_0 \cos H_0 & \sin\delta_0\sin H_0 & \cos\delta_0\\
\cos \delta_0 \cos H_0 & -\cos\delta_0\sin H_0 & \sin\delta_0
\end{pmatrix}
\begin{pmatrix}
X\\
Y\\
Z
\end{pmatrix}.
$$

矩阵的三行可以分别理解为：基线在相位中心切平面东西方向上的分量、南北方向上的分量，以及沿相位中心视线方向上的分量。


理论测量方程通常将物理基线除以观测波长，使指数项无量纲。定义

$$
\mathbf{b}_{\lambda}=\frac{\mathbf{b}}{\lambda}.
$$

于是

$$
\begin{pmatrix}
u\\
v\\
w
\end{pmatrix}
=
\frac{1}{\lambda}
\begin{pmatrix}
\sin H_0 & \cos H_0 & 0\\
-\sin \delta_0 \cos H_0 & \sin\delta_0\sin H_0 & \cos\delta_0\\
\cos \delta_0 \cos H_0 & -\cos\delta_0\sin H_0 & \sin\delta_0
\end{pmatrix}
\begin{pmatrix}
X\\
Y\\
Z
\end{pmatrix}.
$$

此时 $u$、$v$ 和 $w$ 是无量纲的波长数。在把局部角坐标视为弧度的二维近似中，也可将其理解为空间角频率。数据格式未必采用波长数存储；例如 CASA Measurement Set 的 UVW 列通常以米给出，成像器再按每个频道的 $\lambda$ 缩放。读取数据时必须检查列单位，不能仅凭变量名判断。

上述变换矩阵是正交矩阵，因此提供一个很有用的实现检查：

$$
u^2+v^2+w^2={X^2+Y^2+Z^2\over\lambda^2}.
$$

若数值变换不保持这个长度，通常说明矩阵转置、时角符号、基线顺序或单位至少有一项不一致。


### 4.2.3 本节小结
