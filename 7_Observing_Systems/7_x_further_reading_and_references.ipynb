## 7.x 延伸阅读与参考文献

第 7 章的跨度较大：Jones/RIME 属于测量方程的数学语言，接收机和相关器属于信号链工程，主波束、极化、传播效应和 RFI 又同时具有观测、仪器和数据处理含义。因此，本章的延伸阅读更适合按问题组织，而不是按书目顺序堆叠。

比较稳妥的阅读路径是先回到 `7.1` 与 `7.2`，把 Jones 语言、亮度矩阵、相关矩阵和 RIME 的符号体系固定下来；随后阅读 `7.3` 到 `7.5`，把模拟链路、数字相关器和主波束视为同一条系统响应链；最后进入 `7.6` 到 `7.9`，区分极化响应、传播统计、RFI 与灵敏度权重。这样安排的好处，是每种效应都会落到明确的模型层，而不是变成孤立名词。


### 7.x.1 RIME、Jones 语言与方向依赖效应

本章理论骨架的核心文献是 Smirnov 关于 RIME 的系列论文。第一篇 [*Revisiting the radio interferometer measurement equation. I. A full-sky Jones formalism*](http://adsabs.harvard.edu/abs/2011A%26A...527A.106S) 给出全空域 Jones 形式，是把标量测量方程推广为矩阵测量方程的重要入口。第二篇 [*Calibration and direction-dependent effects*](http://adsabs.harvard.edu/abs/2011A%26A...527A.107S) 把校准和方向相关效应放进同一框架。第三篇 [*Addressing direction-dependent effects in 21 cm WSRT observations of 3C 147*](http://adsabs.harvard.edu/abs/2011A%26A...527A.108S) 展示这一框架怎样进入真实观测。第四篇 [*A generalized tensor formalism*](http://adsabs.harvard.edu/abs/2011A%26A...531A.159S) 则进一步把符号扩展为更一般的张量形式。

这些论文和本章的对应关系很直接。`7.1` 提供 Jones 向量和 Jones 矩阵，`7.2` 建立基本 RIME，`7.5` 到 `7.7` 则给出方向相关项、传播项和极化项的物理例子。阅读时应先问某个效应是否满足确定性、线性、乘性和非退偏的 Jones 假设；若满足，再判断它位于链的哪一侧，以及怎样依赖方向、频率、时间和极化。


### 7.x.2 偏振、馈源与全相干标定

极化部分若只停留在 Stokes 参数定义，会很容易显得抽象。真正需要建立的是系统认识：馈源基底如何把极化信息编码进相关积，视差角怎样旋转馈源坐标系，D-term 又怎样把 Stokes I 泄漏成假偏振。Hamaker, Bregman & Sault (1996) 的 [*Understanding radio polarimetry. I. Mathematical foundations*](http://adsabs.harvard.edu/abs/1996A%26AS..117..137H) 是这一方向的经典文献，重点在于用线性代数语言描述极化和相干矩阵。Hamaker (2000) 的 [*Understanding radio polarimetry. IV. The full-coherency analogue of scalar self-calibration*](http://adsabs.harvard.edu/abs/2000A%26AS..143..515H) 则把全相干校准与极化保真度联系起来。

这些文献与 `7.6` 的关系尤其紧密。线馈源和圆馈源并不是两个互不相干的约定，而是同一电场在不同基底下的表达；D-Jones 也不是一个经验修正项，而是馈源响应不完美时自然出现的非对角耦合。


### 7.x.3 数字相关器、接收机与工程实现

`7.3` 和 `7.4` 涉及的许多问题属于信号链工程。Richard G. Lyons 的 *Understanding Digital Signal Processing* 适合补足采样、量化、窗函数、FFT、滤波和频谱泄漏等数字信号处理基础。Paul Horowitz 与 Winfield Hill 的 *The Art of Electronics* 虽然不是射电干涉测量教材，但对模拟前端、噪声、增益链、阻抗匹配和实际电路直觉非常有帮助。

这两类材料可以把“公式上的相关器”拉回硬件和软件中的相关器。ADC 位数会影响量化效率，PFB 窗函数会改变通道间泄漏，模拟前端的噪声温度会进入 SEFD，corner turn 和相关矩阵规模会决定数据搬运和算量瓶颈。现代干涉仪中的数字相关器不是抽象黑盒，而是一连串带有代价、近似和资源约束的设计决策。


### 7.x.4 主波束、传播与 RFI

Thompson, Moran & Swenson 的 [*Interferometry and Synthesis in Radio Astronomy, 3rd ed.*](https://doi.org/10.1007/978-3-319-44431-4) 是干涉测量、天线响应、灵敏度和传播误差的综合参考；Wilson, Rohlfs & Hüttemeister 的 [*Tools of Radio Astronomy, 6th ed.*](https://doi.org/10.1007/978-3-642-39950-3) 对接收机、系统温度、辐射计方程和大气传播提供了更宽的观测背景。

对流层相位可继续阅读 Carilli & Holdaway (1999) 的 [*Tropospheric phase calibration in millimeter interferometry*](https://doi.org/10.1029/1999RS900048)，低频电离层方向相关校准可读 Intema et al. (2009) 的 [*Ionospheric calibration of low frequency radio interferometric observations using the peeling scheme*](https://doi.org/10.1051/0004-6361/200811094)。RFI 自动标记的代表性入口是 Offringa et al. (2010) 的 [*Post-correlation radio frequency interference classification methods*](https://doi.org/10.1111/j.1365-2966.2010.16471.x) 与 Offringa et al. (2012) 的 [*A morphological algorithm for improving radio-frequency interference detection*](https://doi.org/10.1051/0004-6361/201118497)。阅读这些工作时要区分可校准相位、统计去相干、加性污染和前端非线性，它们不能共享同一个修正模型。


### 7.x.5 通道化、数据格式与灵敏度核算

Price (2016) 的 [*Spectrometers and Polyphase Filterbanks in Radio Astronomy*](https://arxiv.org/abs/1607.03579) 适合继续理解通道间隔、滤波器响应、谱泄漏和 PFB。Measurement Set 的列、形状和约定应查阅 casacore 的 [*MeasurementSet definition version 2.0*](https://casacore.github.io/casacore-notes/229.html)，尤其要核对 DATA、FLAG、WEIGHT、WEIGHT_SPECTRUM、SIGMA 和 UVW，而不能凭列名猜测粒度。

灵敏度计算应回到明确的 SEFD、相关效率、有效噪声带宽、有效积分时间和逐样本逆方差权重。第 7.9 节给出的同质与异质阵列公式是自然权重热噪声基线；稳健权重、taper、flag、相关通道和校准残差都应作为偏离该基线的可审计因素。


### 7.x.6 与第 8 章校准的接口

进入第 8 章之前，第 7 章最值得反复回看的内容是若干接口知识。`7.1` 与 `7.2` 提供 Jones 矩阵、亮度矩阵、可见度矩阵和 RIME；`7.5` 说明主波束为什么属于方向相关增益；`7.6` 说明馈源旋转和 D-term 为什么会改变极化测量；`7.7` 区分可由天线 Jones 表示的传播响应与需要统计模型的散射；`7.8` 说明 flag 不是简单后处理，而会直接影响校准和成像权重；`7.9` 则提供检查权重和实测噪声的定量基线。

本章最终要形成的判断是：观测系统不会忠实搬运天空信号，而会持续地重写、混合、衰减、旋转、污染和丢弃信息。校准的任务，就是尽可能把这些系统效应显式建模，并从数据中分离出可解释的天空信号。
