## 1.8 天文射电源<a id='science:sec:astronomical_radio_sources'></a>

射电天文学的研究对象覆盖从太阳系、银河系到河外宇宙的广泛尺度，包括太阳和行星、电离气体、分子云、超新星遗迹、脉冲星、恒星形成星系、活动星系核以及宇宙学背景辐射。对这些天体的分类应同时考虑辐射机制、观测特征及其对望远镜、频段、分辨率和数据处理方法的要求。

射电数据对气体、磁场、相对论粒子和传播效应具有独特的诊断能力，能够显示可见光波段不显著或不可见的结构。本节按照观测诊断量组织典型射电源，并说明不同物理对象对频率覆盖、角分辨率、时间分辨率、谱线、偏振和空间尺度响应提出的不同要求。


### 1.8.1 射电源的观测诊断框架

射电源通常不能仅凭一张图像分类。连续谱和谱线首先提供两种互补信息：连续谱强调宽带辐射机制、谱指数和亮温，谱线则强调成分、速度和激发状态。对连续谱源，通量密度常写成 $S_\nu\propto\nu^\alpha$，不同谱指数能够区分自由-自由辐射、光学薄同步辐射和同步自吸收等情况。亮温则把面亮度转化为温度单位，若远高于普通热平衡温度，就会指向非热辐射、相干辐射或受激放大。

按辐射机制，可将射电信号分为四类：热连续谱、非热连续谱、相干或受激辐射以及谱线辐射；色散、法拉第旋转、散射和闪烁属于传播效应，不应视为独立的发射机制。同一天体源可以同时包含多种辐射成分，因此其物理分类应依据多种观测量的联合证据。

形态、角尺度、偏振和时间变化也是重要诊断量。致密核心、脉泽斑点和脉冲星定位需要长基线；超新星遗迹、星系盘、太阳和 CMB 等延展结构需要短间距覆盖与较高的面亮度保真度；同步辐射常具有线偏振，某些磁层辐射或相干辐射则可表现出强圆偏振；喷流演化、太阳暴发、脉冲星和 FRB 还要求较高的时间分辨率。射电源分类的物理基础，是建立谱形、亮度、形态、偏振和时间变化与源环境之间的对应关系。

表 1.8.1 概括了主要观测量、物理诊断、观测要求及其适用边界。每一项观测量都对应特定的物理问题，同时受到相应退化关系和系统误差的限制。

| 观测量 | 主要诊断 | 首要观测要求 | 需要警惕的退化或失效 |
|:---|:---|:---|:---|
| $S_\nu$、$I_\nu$、$T_b$、谱指数 $\alpha$ | 总辐射、热/非热机制、吸收 | 宽带、共同 $uv$ 覆盖、共同恢复波束 | 波束稀释、主波束色度、谱标度不一致 |
| 线心、线宽、$\int T_bdv$ | 速度场、温度/湍流、柱密度和质量 | 足够细的通道、稳定带通、线外连续谱 | 速度约定、连续谱扣除、光学深度、缺短间距 |
| $Q/U/V$、RM、Zeeman 分裂 | 磁场几何、视线磁场、传播介质 | 全 Stokes、宽 $\lambda^2$ 覆盖、极化校准 | 泄漏、退偏、$\pi$ 角模糊、线形混合 |
| $DM$、散射尾、闪烁、频率漂移 | 电离电子柱密度和小尺度传播结构 | 高频率/时间分辨率、动态谱或电压数据 | 通道内残余色散、散射展宽、时间平均 |
| 形态、角尺度和时间变化 | 结构层级、运动、暂现源和定位 | 合适的长短基线、采样节奏和触发链 | 分辨掉扩展结构、混淆、相位失稳 |

三种常见视线积分分别对应不同物理诊断：$EM=\int n_e^2dl$ 主要控制自由-自由发射和吸收，$DM=\int n_edl$ 描述色散延迟，$RM\propto\int n_eB_\parallel dl$ 则联系电子柱密度与视线磁场。它们不是源结构的直接成像量，而是传播路径和介质状态的积分量。

### 1.8.2 太阳与空间天气

太阳是地球上最亮、同时也是最复杂的射电天体。它的射电辐射既包括较平稳的宁静太阳成分，也包括耀斑、日冕物质抛射、激波和磁重联等活动过程产生的剧烈暴发。对射电干涉观测而言，太阳尤其重要，因为它在同一个目标中叠加了热辐射、非热辐射、偏振、快速时变和传播效应。

宁静太阳在不同频段对应不同高度和光深条件下的色球层、过渡区与日冕等离子体；暴发期的太阳射电辐射则常由高能电子束、等离子体辐射、回旋脉泽辐射等机制主导。观测上，太阳射电爆发的通量密度可远高于典型河外源，频率漂移、偏振和形态演化又携带磁场拓扑与能量释放信息。由于源结构变化快、亮度高、动态范围要求高，太阳成像往往需要高时间分辨率、高频率分辨率以及专门的校准策略；它既服务于基础太阳物理，也与空间天气预报密切相关。


### 1.8.3 选读：行星与磁层辐射

太阳系行星在射电波段同样重要，但辐射机制并不单一。类地行星和巨行星的大气、表面及地下结构可以通过热辐射来探测；具有强磁场的行星，尤其是木星，还会产生显著的非热磁层辐射。行星热辐射帮助约束温度结构、化学组成和表面介电性质，磁层辐射则揭示行星磁场、带电粒子分布以及与太阳风的相互作用。

某些行星的低频辐射具有较强圆偏振，可作为磁场诊断量。对系外行星而言，搜寻类似木星磁层辐射的低频信号，是探测行星磁场的潜在方法。此类目标同时涉及行星环境与空间等离子体物理，并根据频段、灵敏度和角分辨率需求选择单碟望远镜、低频阵列或长基线干涉阵。


### 1.8.4 电离氢区、分子云与脉泽源

恒星形成区包含多种射电辐射过程。H II 区的自由-自由辐射直接示踪年轻大质量恒星周围的电离气体，分子云中的转动跃迁谱线则可测量冷气体的温度、密度、化学组成和速度场。H II 区通常表现为热连续谱源，其谱指数和光学深度随频率变化；分子云研究则主要依赖 CO 及其同位素、致密气体示踪线和复杂有机分子跃迁。

脉泽（maser）源由分子跃迁的受激辐射放大产生，具有很高亮温、很窄线宽和很小角尺度。水、羟基和甲醇脉泽可用于研究恒星形成区、晚型恒星包层和活动星系核邻近环境。脉泽斑点通常需要高角分辨率乃至 VLBI 观测；结合其视向速度和自行还可重建三维运动学信息。


### 1.8.5 超新星遗迹

超新星遗迹是非热射电辐射的典型来源。超新星爆炸产生的激波压缩并放大磁场，同时将粒子加速到相对论能量，进而产生同步辐射。许多超新星遗迹具有壳层结构、显著偏振和空间变化的谱指数，可用于研究同步辐射与宇宙线加速。

超新星遗迹通常是延展源，可呈壳层、中心填充或混合形态。射电辐射能够约束激波与星际介质的相互作用、宇宙线加速、磁场结构和能量反馈；与光学或 X 射线图像联合分析，还可区分激波前后的热与非热成分。这类目标体现了干涉阵与单碟望远镜的互补性：长基线可分辨精细壳层，缺少短间距则会造成大尺度面亮度结构缺失，因此较高角分辨率并不必然对应更完整的结构信息。


### 1.8.6 脉冲星与中子星

脉冲星是高速自转、具有强磁场的中子星，也是射电天文学的重要发现。其射电发射通常具有相干辐射特征，亮温可远高于一般热源和非相干连续谱源。脉冲星观测除平均通量密度外，还重点测量脉冲轮廓、周期、色散量、散射展宽、偏振和单脉冲统计。

这些观测量既反映脉冲星磁层物理，也记录信号穿过电离星际介质时产生的传播效应。色散量给出沿视线自由电子柱密度，散射尾反映小尺度电子密度起伏，偏振位置角变化可用于研究磁层几何。脉冲星观测不同于常规二维成像，通常更重视时间分辨率、相位稳定性、去色散处理和周期折叠；原始电压、时间序列或折叠脉冲轮廓本身均可构成科学数据产品。


### 1.8.7 选读：宇宙微波背景

宇宙微波背景（CMB）是射电和毫米波天文学中的重要热辐射背景。其平均谱形高度接近理想黑体，主要宇宙学信息包含在微弱温度起伏、极化结构和次级各向异性中。CMB 不同于孤立的局域天体源，而是覆盖全天的大尺度背景场。

CMB 观测更重视波束控制、系统噪声、扫描策略、前景分离、绝对校准和大角尺度保真度，而非单纯追求高角分辨率。它表明射电观测对象也可以是覆盖大范围天区、需要统计方法和精密系统控制才能提取的背景场。


### 1.8.8 射电星系、类星体与活动星系核

射电星系和类星体是由活动星系核（AGN）驱动的典型河外射电源。其能量通常来自中心超大质量黑洞吸积，并通过喷流输运到宿主星系及更大尺度环境。射电辐射多由同步辐射主导，因而常具有负谱指数和显著线偏振。

活动星系核具有多样的射电形态：致密核心通常需要 VLBI 分辨，喷流可延伸至 kpc 乃至 Mpc 尺度，双瓣结构和多波段反馈过程则跨越更大的空间范围。射电图像与谱指数分布可用于研究喷流能量学、电子老化、磁场结构以及 AGN 对宿主星系和星系团环境的反馈。由于核心、喷流、热点和射电瓣跨越多个角尺度，这类目标通常需要不同阵列构型、观测频率和成像策略的联合应用。

### 1.8.9 恒星形成星系

与由黑洞活动主导的射电星系不同，恒星形成星系的射电辐射更多反映恒星形成本身及其反馈。其总射电发射通常是两部分混合：一部分来自 H II 区的热自由-自由辐射，另一部分来自超新星遗迹和宇宙线电子产生的非热同步辐射。低频更容易由同步辐射主导，高频厘米波到毫米波则可能逐渐增加自由-自由和尘埃连续谱贡献。

这种混合辐射使射电波段成为估计恒星形成率的重要工具，尤其适用于尘埃严重遮蔽的星系。通过多频数据分离热与非热成分，还可进一步约束大质量恒星形成、宇宙线传播和星系磁场结构。星暴星系具有较高面亮度、较高超新星率和复杂的多尺度结构，通常同时包含致密辐射区和平滑盘面成分，因此对角分辨率和面亮度灵敏度均有要求。

### 1.8.10 时变射电天空与快速暂现源

除相对稳定的连续谱源和谱线源外，射电天空还包含大量时变目标，包括耀斑星、X 射线双星、磁星、脉冲星单脉冲、行星磁层事件和快速射电暴（FRB）。对这些目标而言，时间结构本身是主要科学信号，而非成像产品的附属信息。

快速暴发现象通常要求高时间分辨率、高频率分辨率和对传播效应的精细校正。冷等离子体色散使低频脉冲晚到；若频率用 GHz、色散量 $DM=\int n_e\,dl$ 用 $\mathrm{pc\,cm^{-3}}$，两频率之间的群延迟近似为

$$\Delta t(\mathrm{ms})=4.15\,DM\left[\left(\frac{\nu_1}{\mathrm{GHz}}\right)^{-2}-\left(\frac{\nu_2}{\mathrm{GHz}}\right)^{-2}\right],\qquad \nu_1<\nu_2.$$

例如，$DM=500\,\mathrm{pc\,cm^{-3}}$ 的脉冲在 1.2 GHz 与 1.5 GHz 之间的到达时间相差约 0.52 s，远大于 FRB 常见的毫秒级宽度。若保留复电压，可利用冷等离子体逆传递函数进行相干去色散；若只保留通道化强度，则需对足够细的频率通道作平移和叠加，以实现非相干去色散，通道内残余色散无法恢复。时间或频率分辨率不足都会展宽脉冲并降低峰值信噪比。散射也可形成频率相关的非对称尾部，但其频率依赖和视线效应不同于色散。宿主星系定位还需要干涉阵的角分辨率和快速触发能力。时变射电观测因而将成像、波束形成、实时搜索、去色散和后续定位组织为一体化工作流程。


### 1.8.11 从射电源到干涉测量需求

不同射电源对观测系统提出不同要求。AGN 核、脉泽斑点和部分脉冲星定位需要高角分辨率，因而依赖长基线干涉阵；超新星遗迹、星系盘、太阳和 CMB 相关结构需要良好的短间距覆盖和面亮度保真度；太阳暴发、脉冲星和 FRB 等强时变源需要高时间分辨率以及专门的校准、去色散或搜索流程；谱线源则要求高频率分辨率和严格的频谱标定。

射电源的物理性质说明了干涉阵存在的科学动机，也说明单一阵列构型难以覆盖全部科学目标。不同源即使采用相同的成像框架，也会要求不同的频率设置、基线分布、校准策略和科学产品。后续关于可见度、采样、成像和校准的讨论均建立在这些观测需求之上。


#### 综合示例：由科学问题确定观测设置

考虑一个红移为 $z=0.05$ 的星系，其中同时存在致密 AGN 核、约 $2'$ 的低面亮度同步辐射晕、H I 21 cm 谱线和 1--2 GHz 偏振信号。该示例不指定具体观测设施，而是逐项分析科学目标对应的观测能力：

| 科学问题 | 必要能力 | 结论边界 |
|:---|:---|:---|
| 分辨或定位约 $0.2''$ 的 AGN 核 | 足够长的基线和稳定相位 | 未分辨核的尺寸只能给上限，亮温通常只能给下限；位置误差还需结合信噪比和标定 |
| 恢复约 $2'$ 的同步晕 | 短基线、合适 taper，并视情况加入单碟或总功率 | 负碗、通量随 taper 改变或无 overlap 时，大尺度通量不可直接解释 |
| 测量 H I 速度场 | 约 $1.353$ GHz 的频率设置、连续谱扣除和足够细的通道 | 线宽还会受到仪器响应、未分辨速度梯度和光深影响 |
| 测量 RM | 全 Stokes、宽 $\lambda^2$ 覆盖、频率分辨率和泄漏校准 | $\chi(\lambda^2)$ 非线性时，单一 Faraday 屏模型不再充分 |

合理的观测方案通常需要设计相互补充的数据产品：长基线约束核区位置，短基线和单碟数据恢复同步辐射晕及绝对亮度尺度，谱线设置提供速度信息，宽带全 Stokes 数据用于偏振分析。比较谱指数或 RM 时，还必须匹配有效 $uv$ 覆盖和恢复波束，否则频率之间的结构差异可能来自采样差异。后续第 4--9 章将分别讨论这些环节的数学基础和实践实现。
