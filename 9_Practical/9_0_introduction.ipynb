# 射电干涉数据处理实践

本册是《射电干涉孔径合成基础》的实践教材，面向高年级本科实践课、研究生课程、暑期学校和研究训练。第一册建立辐射、傅里叶分析、可见度、成像、观测系统和校准原理；本册将这些原理落实为数据检查、校准、成像、科学测量、质量控制和可复现项目。

本册沿用原第 9 章的 `9.x` 文件编号，以保持现有链接和课程材料兼容。该编号属于文件路径兼容标识，不表示本册仍是第一册的第 9 章；读者应按本页给出的阅读路径选择相应的实践单元。


## 先修要求与环境边界

读者应理解第一册第 4 至第 8 章的可见度、成像、去卷积、RIME 和校准主线；只完成基础实践时，也可以在遇到公式时回查相应章节。本书默认环境仍由仓库根目录 `requirements.txt` 提供，真实轻量样例不要求 CASA 或 `python-casacore`。

CASA、WSClean、DP3/DDFacet、CARTA、PyBDSF、SoFiA 和大型公开归档数据属于课程扩展环境。使用这些工具时，应单独记录软件版本、数据身份、参数和运行日志，不把平台相关依赖加入基础教材环境。

## 页面类型

本书把页面标为四类。**讲授页**解释工作流背后的物理和统计判断；**实验页**包含可运行代码、样例数据和定量产物；**参考页**整理软件生态、特殊体制或拓展算法；**项目页**给出里程碑、量规和提交物。层级标签中的**核心**适合基础教学主线，**提高**供专题选读，**课程**用于综合作业。页面类型不同，不意味着科学标准不同，但读者不必把参考页或课程页当作线性必修内容。完整分类同时记录在 `book_manifest.yaml`，并由测试检查。

## 谱线测量链

谱线专题按照数据产品、统计算法和物理解释分为三个层次。第 9.13 节属于提高层次；基础课程可以从第 9.7 节直接进入第 9.19 节的受限物理量练习，但不能省略第 9.7 节的数据立方质量控制。

| 页面 | 核心任务 | 交付物 |
| --- | --- | --- |
| [9.7](9_7_spectral_line_processing.ipynb) | 谱线数据立方的坐标、波束、噪声和掩膜溯源 | 数据立方、噪声、波束、基础掩膜与诊断产品 |
| [9.13](9_13_advanced_spectral_line_analysis.ipynb) | 相关噪声中的三维信号检测与运动学模型检验 | 选择函数、候选与测量掩膜、前向模型和三维残差 |
| [9.19](9_19_spectral_line_physical_interpretation.ipynb) | 谱线测量向柱密度、质量和动力学量转换的假设 | 包含统计与系统误差、适用尺度和失效条件的物理结果 |


## 偏振测量链

偏振专题按照基础 IQUV 产品、校准与法拉第反演、物理解释分为三个层次。第 9.11 节属于提高层次；基础课程可在第 9.9 节后使用单一法拉第薄案例进入第 9.20 节，但不能省略符号约定、偏振泄漏和偏振统计检查。

| 页面 | 核心任务 | 交付物 |
| --- | --- | --- |
| [9.9](9_9_polarization_imaging.ipynb) | IQUV 产品的波束、噪声、符号和泄漏审查 | IQUV、$Q/U$ 协方差、派生偏振量与泄漏上限 |
| [9.11](9_11_polarization_calibration_and_rm_synthesis.ipynb) | 校准和 $\lambda^2$ 采样对法拉第结构恢复的限制 | 校准证据、脏法拉第谱、RMSF 与算法停止条件 |
| [9.20](9_20_polarization_faraday_interpretation.ipynb) | 法拉第测量的前景、红移和退偏振模型 | 包含模型退化、系统误差和替代解释的磁化等离子体约束 |


## 短间距测量链

短间距专题首先建立缺失尺度与输入数据的诊断方法，再进行定量的方法选择和结果验收。第 9.10 节是基础内容；需要报告总通量、柱密度、质量、速度场或功率谱时，还应完成第 9.21 节，不能以图像外观代替传递函数验证。

| 页面 | 核心任务 | 交付物 |
| --- | --- | --- |
| [9.10](9_10_short_spacing_and_feather.ipynb) | 缺失尺度与羽化合并输入的一致性 | 科学尺度、输入定义、可信重叠区与不可恢复尺度的组合前诊断 |
| [9.21](9_21_short_spacing_mosaic_joint_imaging.ipynb) | 镶嵌、多配置、羽化合并和联合去卷积新增约束的定量验收 | 响应函数与最大可恢复尺度、方法选择、通量恢复、残差、拆分与注入检验和限制声明 |


## 宽带宽场测量链

宽带宽场专题按科学产品、已知成像算子和未知方向响应分成三层。前两层没有通过时，不应直接增加方向相关校准自由度；需要谱指数、偏振或高动态范围科学量时，三层都要给出相应证据。

| 页面 | 核心任务 | 交付物 |
| --- | --- | --- |
| [9.8](9_8_wideband_and_widefield_imaging.ipynb) | 参考频率宽带产品及其协方差、主波束语义与失效区域 | $I_0$、谱指数/曲率、协方差、主波束、噪声和选择函数的分频段验证 |
| [9.12](9_12_direction_dependent_widefield_imaging.ipynb) | 已知 $w$ 项、A 项和指向信息的一致前向与伴随算子 | 相位预算、算子选择、A 项来源、位置相关 PSF 和灵敏度及注入检验 |
| [9.22](9_22_wideband_widefield_algorithm_boundaries.ipynb) | 未知方向相关效应的归因、受约束求解与过拟合检验 | 固定、求解和忽略项清单，规范与正则化，训练/留出比较，科学量稳定性和停止条件 |

## 学习成果与结论边界

完成基础路径后，读者应能说明每一步改变的是数据、模型、权重、数据标记还是图像解释，并能依据残差、噪声、通量、位置或选择函数判断结果。完成课程项目后，读者还应能提交数据清单、处理参数、运行记录、质量控制证据、误差预算和限制声明。

实践训练不要求每个数据集都产生可发表图像。识别数据不适合某个科学问题、拒绝没有绝对标尺的 Jy 结论，或在证据不足时降低结论等级，都是合格且重要的学习成果。
