## 4.7 闭合量、短间距缺失与拼接观测

第 4.6 节建立了可见度与天空亮度之间的傅里叶关系。本节进一步讨论实际数据中反复出现的三个问题：哪些结构信息不依赖天线增益，短基线不足为何会导致扩展源通量缺失，以及拼接观测（mosaicking）为何需要同时从像平面和可见度平面加以理解。

这三个问题共同表明：可见度不是一组可以任意插值的图像像素，而是具有特定几何、仪器响应和采样结构的复测量量。闭合相位利用三条基线的相位和抵消天线相位误差；短间距缺失说明 $uv$ 平面中心区域承载总通量和大尺度结构信息；拼接观测则通过多个指向改变主波束加权，使像平面和傅里叶空间中的信息得到重新组合。

### 4.7.1 闭合相位与闭合振幅

若只考虑方向无关天线增益，基线 $ij$ 上的观测可见度可写为

$$
V_{ij}^{\rm obs}=g_i g_j^\ast V_{ij}^{\rm true},
$$

其中 $g_i=|g_i|e^{i\phi_i}$ 是天线 $i$ 的复增益。单条基线相位包含源结构相位和两个天线相位之差，因此并不能直接等同于天空结构。三台天线形成闭合三角形时，三条基线相位相加得到

$$
\Psi_{ijk}=\arg V_{ij}^{\rm obs}+\arg V_{jk}^{\rm obs}+\arg V_{ki}^{\rm obs}.
$$

把每条基线中的天线相位展开，$\phi_i-\phi_j+\phi_j-\phi_k+\phi_k-\phi_i$ 正好抵消，剩下的只与真实可见度相位有关。这就是闭合相位；实际记录的闭合相位按模 $2\pi$ 解释。类似地，四台天线可以构造闭合振幅，例如

$$
C_{ijkl}= {|V_{ij}^{\rm obs}|\,|V_{kl}^{\rm obs}|\over |V_{ik}^{\rm obs}|\,|V_{jl}^{\rm obs}|},
$$

在理想天线增益模型下，天线振幅因子也会抵消。闭合量不需要知道每个天线的绝对相位或振幅，因此在自校准、VLBI 和质量诊断中非常重要。

![闭合相位三角形](figures/closure_phase_triangle.png)

**图 4.7.1：闭合相位三角形。** 三台天线构成一个闭合三角形。每条基线相位都含有两个天线相位，但三条基线相位求和后，天线相位项相互抵消。闭合相位保留源结构信息，同时对基于天线的相位误差不敏感。

闭合量的适用范围具有明确限制。它只能抵消可分解为天线因子的误差，不能抵消基线相关误差、方向相关误差、偏振泄漏、时间或频率平均误差，以及真实天空模型不完备引起的结构偏差。热噪声也会直接传递到闭合量中，低信噪比三角形的闭合相位因而不稳定。因此，闭合量主要用于检验模型是否与数据中不依赖天线增益的部分一致，不能替代校准和成像。

在实际处理中，若自校准后普通残差明显减小，但闭合相位残差没有相应改善，问题通常不在天线相位解本身，而可能来自天空模型、方向相关项或数据选择。相反，若闭合量稳定而天线增益解剧烈波动，则可能表明求解间隔过短、参考天线选择不当或增益模型自由度过高。

闭合量也会消去部分物理信息：天空整体平移产生的基线相位斜率在三角和中抵消，因此仅靠闭合相位不能恢复绝对天体测量位置；整体通量缩放在闭合振幅比中抵消，因此闭合振幅不能确定绝对通量尺度。对于 $N$ 面天线，独立闭合相位数为 $(N-1)(N-2)/2$，独立闭合振幅数为 $N(N-3)/2$。这些不变量具有重要诊断价值，但不能无损替代原始可见度。

![闭合量诊断时间序列](figures/closure_diagnostic_timeseries.png)

**图 4.7.2：闭合量诊断时间序列。** 天线相位漂移会使单条基线相位随时间明显变化，而闭合相位仍可保持稳定。在真实数据中，闭合量常用于区分基于天线的校准问题与源结构或模型问题。

### 4.7.2 短间距缺失与总通量

傅里叶关系给出一个直接结论：$V(0,0,0)$ 对应主波束加权的表观天空总通量。靠近原点的短投影基线约束大角尺度结构，远离原点的长基线约束小角尺度结构。真实互相关通常没有零物理间距，自相关又混有接收机和地面噪声，因此零间距总功率需要独立测量和标定。若阵列没有足够短的有效基线，扩展源总通量和背景水平就不可靠。

常用的最大可恢复尺度近似写作

$$
\theta_{\rm MRS}\simeq \eta {\lambda\over B_{\min}},
$$

其中 $B_{\min}$ 应理解为数据中具有有效权重的最短**投影**基线，$\eta$ 取决于阵列、权重、拼接观测和软件实现。天线口径、遮挡标记与投影会使它不同于天线中心之间的最短物理距离。最大可恢复尺度不是硬截止，而是依赖源形态的经验指标；短间距缺失常表现为扩展源周围的负碗效应（negative bowl）、总通量低估和背景水平被迫接近零。

![短间距缺失的傅里叶空间视图](figures/missing_short_spacing_fourier_view.png)

**图 4.7.3：短间距缺失的三种等价视角。** 左图显示 $uv$ 平面中心区域未被采样；中图显示扩展源可见度的主要功率集中在低空间频率；右图展示短间距滤波后，宽广发射被削弱并伴随负碗结构。

短间距缺失不是去卷积算法单独能够解决的问题。算法可以在先验约束下选择某个看似合理的大尺度模型，但未被测量的傅里叶分量不能凭空从数据中恢复。单碟总功率、紧凑阵列配置、多配置数据合并和联合去卷积的价值，正是补充不同空间频率范围的约束。若科学目标是测量扩展星系的总 H I 通量、星形成区的弥散分子气体或星系团的低面亮度晕，短基线和单碟信息有时比最高角分辨率更重要。

### 4.7.3 拼接观测的像平面与可见度平面表述

单个指向观测到的是经主波束加权的天空。若主波束为 $A(l,m)$，则表观亮度近似为 $A I$；在傅里叶空间中，这对应于真实可见度与 $\widetilde A$ 的卷积。对于直径为 $D$ 的天线，$\widetilde A$ 与孔径照明的自相关有关，并具有约 $D/\lambda$ 的有限支持。因此，一条名义基线实际上混合其附近一段空间频率，而不是只测量一个无限窄的空间频率点。

移动指向中心会给这段卷积核施加已知的相位调制；充分采样的拼接观测因而既能扩大视场，也能在每条物理基线附近分离更多傅里叶信息。这正是拼接观测不同于普通图像拼接之处。它能把最低可用空间频率从大约 $B_{\min}/\lambda$ 向下扩展一个孔径尺度，但若仍未覆盖原点，就不能恢复真正的零间距信息。联合拼接成像会同时考虑主波束、噪声和重叠区域；涉及总通量的科学测量通常仍需要紧凑阵列配置或单碟数据。

![拼接观测的像平面与傅里叶空间视图](figures/mosaic_image_uv_dual_view.png)

**图 4.7.4：拼接观测的像平面与傅里叶空间视图。** 在像平面中，不同指向给出经不同主波束加权的表观天空；在傅里叶空间中，主波束乘法对应孔径函数卷积，不同指向引入不同的相位调制。拼接观测可以改善视场覆盖和部分低空间频率约束，但不能替代真正的零间距总功率测量。

### 4.7.4 教学案例：扩展星系的通量缺失

附近星系的 H I 发射或低频连续谱晕是常见实例。长基线数据可以给出清晰的细节结构和较高的峰值信噪比，但与单碟或低分辨率巡天相比，测得的总通量可能明显偏低。此时，仅增加 CLEAN 迭代次数通常不能解决问题，反而可能把残余旁瓣和噪声误解释为大尺度成分。

诊断应从傅里叶空间的采样范围开始。首先检查最短基线对应的最大可恢复尺度是否小于目标角尺度；随后查看 $uv$ 权重是否过度偏向长基线，以及是否采用了进一步降低短基线贡献的均匀加权；再比较不同锥削（taper）和不同阵列配置下的积分通量。若短基线覆盖本身不足，应将结论限定为“高空间频率结构可靠，而总通量和弥散成分不完整”。若科学问题依赖总通量，则需要加入紧凑阵列配置、单碟数据、拼接观测或联合成像约束。

闭合量也可用于辅助诊断。若紧致源结构的闭合相位与模型一致，而扩展通量仍然缺失，问题可能主要来自采样范围，而不是天线相位校准；若闭合相位本身与模型不一致，则还必须检查天空模型或方向相关误差。

### 4.7.5 本节小结

闭合量、短间距缺失和拼接观测都是可见度空间分析的直接延伸。闭合相位和闭合振幅揭示了不依赖天线增益的结构信息；短基线和零间距决定扩展源总通量及大尺度结构是否得到测量；拼接观测则把主波束加权、多指向覆盖和傅里叶空间卷积联系起来。进入第 5 章讨论成像时，权重、栅格化、视场大小和去卷积策略都应依据这些可见度约束加以理解。
