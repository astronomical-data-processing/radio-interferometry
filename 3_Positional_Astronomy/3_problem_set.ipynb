# 第 3 章定量问题集

本问题集检查从星表方向到观测相位的完整坐标链。每题都应写出单位转换、采用的角度约定和中间量；只给最终数值不足以判断符号是否正确。A、B 适合本科阶段基础训练，C、D 要求把球面几何与干涉测量相位联系起来。


## A. 时角与观测时刻

某源赤经为 $\alpha=8^{\rm h}20^{\rm m}$，观测时地方恒星时为 $\mathrm{LST}=10^{\rm h}40^{\rm m}$。采用 $H=\mathrm{LST}-\alpha$，计算时角的小时数和角度，并判断天源位于子午圈东侧还是西侧。若地方恒星时每恒星小时增加一小时，该源距离 $H=+4^{\rm h}$ 还有多长时间？

进一步说明：若误把 UTC 直接当作 LST，缺少了哪两类信息？回答中应提到台站经度以及日期对应的格林尼治恒星时。


## B. 仰角限制与观测窗口

台站纬度为 $\phi=34^\circ$，天源赤纬为 $\delta=20^\circ$。使用

$$
\sin h=\sin\phi\sin\delta+\cos\phi\cos\delta\cos H
$$

求天源在仰角 $h\geq30^\circ$ 时允许的最大 $|H|$，并换算成以过中天为中心的总观测窗口。然后计算过中天仰角，并检查结果是否与窗口解相容。

讨论一个实际限制：仅满足几何仰角门限为什么还不足以保证数据质量？可从大气程长、地面遮挡或天线阴影中任选一个因素说明。


## C. 方向余弦与小角近似

相位中心为 $(\alpha_0,\delta_0)=(12^{\rm h},30^\circ)$，目标位于 $\Delta\alpha=+20'$、$\Delta\delta=-10'$。这里 $\Delta\alpha$ 是赤经角本身，而不是已经乘过 $\cos\delta$ 的天球距离。由

$$
\begin{aligned}
l &= \cos\delta\sin(\alpha-\alpha_0),\\
m &= \sin\delta\cos\delta_0-\cos\delta\sin\delta_0\cos(\alpha-\alpha_0),\\
n &= \sin\delta\sin\delta_0+\cos\delta\cos\delta_0\cos(\alpha-\alpha_0)
\end{aligned}
$$

计算精确的 $(l,m,n)$，验证 $l^2+m^2+n^2=1$。再与 $l\simeq\cos\delta_0\Delta\alpha$、$m\simeq\Delta\delta$ 比较，给出两个近似的绝对误差，并解释为什么赤经偏移必须带上 $\cos\delta_0$。


## D. 星表误差到可见度相位

一条投影长度 $B=5000\ {\rm km}$ 的 VLBI 基线在 $8.4\ {\rm GHz}$ 观测。假设位置误差完全沿基线投影方向，利用

$$
\Delta\phi=2\pi\frac{B}{\lambda}\Delta\theta
$$

计算 $\Delta\theta=0.10\ {\rm mas}$ 引起的相位误差。若位置误差是零均值高斯量，其相位 RMS 等于该值，使用 $\exp(-\sigma_\phi^2/2)$ 估计相干幅度保留比例。最后反求要把相位误差限制在 $10^\circ$ 以内时，星表位置误差应小于多少 mas。

说明这一定量结果为什么使参考架、历元、自行和相位参考源位置成为 VLBI 数据处理的一部分，而不只是星表元数据。


## 参考核对量

- A：$H=2^{\rm h}20^{\rm m}=35^\circ$，天源位于子午圈西侧；到 $H=+4^{\rm h}$ 还需约 $1^{\rm h}40^{\rm m}$。
- B：$|H|\simeq66.65^\circ=4.44^{\rm h}$，总窗口约 $8.89\ {\rm h}$；过中天仰角为 $76^\circ$。
- C：$(l,m,n)\simeq(0.00504674,-0.00290154,0.99998306)$；两个小角近似的绝对误差均约为 $8\times10^{-6}$。
- D：$\lambda\simeq0.03569\ {\rm m}$，相位误差约 $0.427\ {\rm rad}=24.45^\circ$，相干幅度约保留 $91.3\%$；$10^\circ$ 门限对应约 $0.0409\ {\rm mas}$。

核对量只用于检查数量级。提交时仍需给出公式、单位和物理解释。
