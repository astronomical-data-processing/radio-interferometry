## 1.7 谱线辐射<a id='science:sec:line_emission'></a>

连续谱主要描述辐射强度、谱形和亮温，谱线则能够进一步约束气体成分、视向速度、激发状态和密度。星系旋转曲线、分子云运动学、恒星形成前体的密度结构、气体柱密度以及外流和吸积速度场等重要结果，均建立在谱线测量基础上。

谱线观测与连续谱观测具有不同的数据组织和分析要求。连续谱通常通过宽带平均提高灵敏度，谱线观测则必须保留频率方向的细结构。频率偏移对应视向速度，线宽反映热运动、湍流或大尺度速度梯度，线强还取决于物质丰度、激发条件、光学深度和背景辐射。谱线分析由此形成从微观能级跃迁、三维数据立方体到宏观气体动力学解释的完整链条。


### 1.7.1 从微观跃迁到辐射转移

第 [1.3 节](1_3_radiative_transport.ipynb) 已讨论发射系数 $\varepsilon_\nu$ 和吸收系数 $\kappa_\nu$ 在辐射转移中的作用。谱线强度同时受到能级布居、激发条件、背景辐射和光学深度控制；下面给出的 Einstein $A/B$ 系数进一步建立宏观线强与微观跃迁之间的联系。

考虑两个能级 $1$ 和 $2$，高能级向低能级的自发辐射概率由 $A_{21}$ 描述，受激发射和受激吸收分别由 $B_{21}$ 与 $B_{12}$ 描述。这些系数并不是任意常数；在热平衡条件下，它们必须满足

<a id='science:eq:07_001'></a><!--\label{science:eq:07_001}-->
$$
 g_1 B_{12}=g_2 B_{21},
\qquad
 A_{21}={8\pi h\nu_0^3\over c^3}B_{21},
$$

其中 $g_1$ 和 $g_2$ 是能级简并度，$\nu_0$ 是静止跃迁频率。这里的 Einstein $B$ 系数按谱能量密度定义，因此 $A_{21}/B_{21}=8\pi h\nu_0^3/c^3$；若文献改用平均比强度 $J_\nu$ 定义 $B$，相应公式会移动一个 $4\pi/c$ 因子，但可观测的发射率和吸收系数不变。

真实谱线不是无限窄的单一频率尖峰。热运动、湍流、自然展宽、碰撞展宽，以及视线方向上的系统速度梯度，都会把谱线扩展成有限宽度的轮廓。通常引入归一化线轮廓函数 $\varphi(\nu)$，满足 $\int\varphi(\nu)d\nu=1$。把能级布居和线轮廓写入辐射转移表达式，可得

<a id='science:eq:07_002'></a><!--\label{science:eq:07_002}-->
$$
\kappa_\nu={h\nu_0\over c}N_1B_{12}
\left(1-{g_1N_2\over g_2N_1}\right)\varphi(\nu),
\qquad
\varepsilon_\nu={h\nu_0\over 4\pi}N_2A_{21}\varphi(\nu).
$$

括号中的项体现了受激发射对净吸收的抵消。如果上能级布居相对于下能级过高，净吸收甚至可以变为负值，对应 maser 放大。若两能级布居可用激发温度 $T_{\rm ex}$ 描述，则

<a id='science:eq:07_003'></a><!--\label{science:eq:07_003}-->
$$
{N_2\over N_1}={g_2\over g_1}\exp\left(-{h\nu_0\over kT_{\rm ex}}\right).
$$

这些公式表明，谱线强度不仅取决于原子或分子的丰度，还取决于能级布居、密度、温度、背景连续谱和光学深度。同一跃迁在不同环境中可能表现为发射或吸收；同一分子不同跃迁之间的强度比通常能对激发条件提供比单条谱线总强度更强的约束。


### 1.7.2 线轮廓中的速度信息

谱线最有用的特征之一，是静止频率通常已由实验室测量得非常精确，因此观测频率的偏移可以直接转化为速度。若采用射电速度定义，视向速度为

<a id='science:eq:07_004'></a><!--\label{science:eq:07_004}-->
$$
{V_{\rm radio}\over c}={\nu_0-\nu\over \nu_0}.
$$

在低速极限下，这与普通多普勒公式一致。对宇宙学源，首先应使用

$$\nu_{\rm obs}=\frac{\nu_0}{1+z}.$$

此时不能把上面的非相对论射电速度解释为真实退行速度；事实上代入后有 $V_{\rm radio}/c=z/(1+z)$。高红移或高精度工作应报告红移及所用频率/速度参考系，并明确射电、光学或相对论速度约定。对某个系统速度附近的小范围内部运动，上式仍可作为局部近似。频率分辨率和速度分辨率之间满足近似关系

$$
\Delta v\simeq c{\Delta\nu\over \nu_0}.
$$

例如，H I 21 cm 线在 $1.420$ GHz 附近，$5\,\rm km\,s^{-1}$ 的速度分辨率对应约 $24$ kHz 的频率分辨率。若通道太宽，窄线会被稀释，速度结构会被平滑；若通道足够细，则可以区分热展宽、湍流展宽、旋转、外流或多个速度分量。


<img src="figures/line_profile_velocity.png" alt="line_profile_velocity.png" width="900">


**图 1.7.1**：谱线轮廓与速度信息。左图示意热展宽、非热运动和合成线宽；右图示意 H I 21 cm 附近频率轴与射电速度轴之间的对应关系。


线宽的解释需要谨慎。对温度为 $T$、粒子质量为 $m$ 的 Maxwell 速度分布，一维热速度弥散和高斯半高全宽为

$$\sigma_{\rm th}=\sqrt{\frac{kT}{m}},\qquad \mathrm{FWHM}_{\rm th}=2\sqrt{2\ln2}\,\sigma_{\rm th}.$$

若热运动、微观湍流和仪器响应都可近似为相互独立的高斯过程，则方差相加，例如 $\sigma_{\rm obs}^2=\sigma_{\rm th}^2+\sigma_{\rm nt}^2+\sigma_{\rm inst}^2$。同样温度下轻元素线更宽；未分辨速度梯度不一定是高斯，也可能造成额外展宽。旋转盘、外流、吸积流或相互作用星系则可能形成多峰或不对称线形。因此，线轮廓不是一个附属细节，而是谱线观测中最直接的动力学信息之一。


### 1.7.3 H I 21 cm 超精细结构谱线<a id='science:sec:atomic_hydrogen'></a>

中性氢的 21 cm 超精细跃迁，是射电天文学中最具标志性的谱线之一。它来自氢基态中电子自旋与质子自旋相对取向的变化，静止频率为 $1.420405751\,\rm GHz$。这条线的自发跃迁概率极低，$A_{10}\simeq 2.85\times10^{-15}\,\rm s^{-1}$，单个原子的平均跃迁时间约为千万年量级。它之所以仍然能被观测到，是因为宇宙中的中性氢极其丰富，星系盘、外延气体、潮汐结构和星系间相互作用环境中都可能积累出可测的总信号。

在一条视线上可用单一或已知的自旋温度 $T_{\rm spin}$ 描述吸收气体时，中性氢柱密度与真实光学深度满足

<a id='science:eq:07_005'></a><!--\label{science:eq:07_005}-->
$$
N_{\rm H\,I}=1.823\times10^{18}
\left({T_{\rm spin}\over {\rm K}}\right)
\int \tau(v)\,{dv\over {\rm km\,s^{-1}}}
\quad {\rm cm^{-2}}.
$$

若 $T_{\rm spin}$ 随速度变化，应把它放到积分号内。吸收观测还必须考虑背景连续源覆盖因子 $f_c$；忽略云自身发射时，连续谱归一化线形满足 $S(v)/S_c=(1-f_c)+f_ce^{-\tau(v)}$，因此未知的 $f_c$ 会与 $\tau$ 退化。

在许多外星系 H I 发射工作中，21 cm 线可近似看作光学薄，此时连续谱和背景扣除后的线亮温 $T_{b,\rm line}\simeq T_{\rm spin}\tau$，上式化为更常用的亮温积分形式：

<a id='science:eq:07_006'></a><!--\label{science:eq:07_006}-->
$$
N_{\rm H\,I}=1.823\times10^{18}
\int {T_{b,\rm line}(v)\over {\rm K}}\,{dv\over {\rm km\,s^{-1}}}
\quad {\rm cm^{-2}}.
$$

该公式将谱线亮温积分直接转换为中性氢柱密度，从而同时约束气体的空间分布和总量。对于未分辨或只需测量总量的河外 H I 源，光学薄质量还可由积分通量估算：

$$\frac{M_{\rm H\,I}}{M_\odot}=2.356\times10^5\,\frac{D_L^2(\mathrm{Mpc})}{1+z}\int S(v)\,dv\; (\mathrm{Jy\,km\,s^{-1}}).$$

这里 $D_L$ 是光度距离，速度积分必须与所采用的红移和速度定义一致。结合速度信息，可以构建星系旋转曲线，研究潮汐尾、气体剥离、吸积和外流，并利用大半径处近似平坦的旋转曲线约束暗物质分布。

光学薄近似具有明确的适用边界。在低温致密的 H I 云中，自吸收会导致柱密度低估；若视线背景存在强连续谱，同一跃迁还可能表现为吸收线。H I 21 cm 谱线的微观物理较为明确，并能在大范围空间尺度上稳定示踪中性气体。


### 1.7.4 数据立方体、通道图和矩图

谱线干涉观测的基本数据产品是三维数据立方体，其中两个轴为空间坐标，第三个轴为频率或速度。沿速度轴切片可得到通道图，每幅通道图表示特定速度范围内的辐射空间分布；沿速度轴积分得到零阶矩图（moment 0），通常表示积分强度或柱密度；强度加权速度构成一阶矩图（moment 1），用于描述速度场；二阶矩图（moment 2）则常用于描述线宽或速度弥散。


<img src="figures/line_cube_workflow.png" alt="line_cube_workflow.png" width="850">


**图 1.7.2**：谱线数据立方体和常见派生产品。通道图保留速度切片信息，零阶矩图表示积分强度，一阶矩图表示速度场，二阶矩图或线宽图用于描述速度弥散。


从数据立方体构造科学量需要明确的处理条件。生成矩图前通常需要建立掩膜，以避免噪声在大量空通道中累积；多峰谱线或低信噪比区域的一阶矩可能缺乏明确的物理含义；对旋转盘而言，位置-速度图（PV diagram）通常比简单的一阶矩图更能显示动力学结构。因此，谱线分析除成像外，还必须处理频率轴定义、噪声统计、掩膜选择和运动学模型之间的关系。


### 1.7.5 其他重要射电谱线<a id='science:sec:other_line_emission'></a>

除 H I 21 cm 外，射电波段还包含大量原子和分子谱线。CO 转动线是冷分子气体的常用示踪线，例如 CO(1-0) 和 CO(2-1) 的静止频率分别为 $115.271$ GHz 和 $230.538$ GHz。由于 H$_2$ 在冷分子云中缺少易于观测的低能电偶极跃迁，CO 常用于间接估计分子气体质量；但 CO 到 H$_2$ 的转换因子依赖金属丰度、辐射场、密度和动力学环境，并非普适常数。HCN、HCO$^+$、CS 等分子具有较高临界密度，常用于示踪致密分子气体，但其线强也更容易受到激发条件和光学深度影响。

射电复合线产生于高主量子数电子与离子复合后的跃迁，可用于研究 H II 区和电离气体的温度、密度与速度场。脉泽（maser）谱线则由布居反转引起的受激辐射放大产生，常见跃迁包括 OH、H$_2$O 和 CH$_3$OH 脉泽。脉泽具有很高亮温和很窄线宽，可用于研究恒星形成区、晚型恒星包层和活动星系核附近的精细速度结构。水脉泽结合 VLBI 观测还可用于测量几何距离和核区盘运动。

谱线还可以通过 Zeeman 分裂提供视线磁场信息。其频率分裂系数依赖具体跃迁；当分裂小于本征线宽时，信号通常不表现为两条清楚分开的谱线，而是在 Stokes $V$ 中近似呈现与 Stokes $I$ 频率导数相关的弱特征。因此可靠测量需要高频谱分辨率、稳定的全 Stokes 标定，并同时排除波束 squint、泄漏和复杂速度分量。

不同谱线具有不同的微观机制，但均需保留频谱结构，并联合分析线强、线宽、中心速度、线形、空间分布、偏振以及多跃迁线强比。单一谱线的总通量通常不足以唯一确定气体的物理状态。


### 1.7.6 谱线与连续谱数据处理的差异

谱线数据处理具有独立于连续谱成像的参数和质量控制要求。谱线观测通常需要较高的频谱分辨率，因此数据量较大，单通道噪声也较高。带通（bandpass）标定必须具有足够稳定性，否则频率方向的增益起伏可能被误认为宽而浅的谱线或吸收结构。若目标同时具有强连续谱和弱谱线，还需要在 $uv$ 域或图像域进行连续谱扣除，避免连续谱残差进入谱线数据立方体。

干涉阵谱线观测同样受到空间采样限制。扩展气体在某些速度通道中可能具有较大角尺度，缺少短基线或总功率数据会导致大尺度通量缺失，并对柱密度、积分强度和速度场产生系统影响。频率轴上的 RFI、边带结构、Hanning 平滑、通道平均和重采样还会改变有效频谱分辨率与噪声相关性。因此，谱线成像需要独立的参数选择和质量控制流程，并应明确数据立方体、矩图、速度场与柱密度估计之间的关系。


### 1.7.7 本节小结

谱线辐射将微观能级跃迁与宏观气体物理联系起来，是研究物质成分、温度、密度、速度场和柱密度的重要工具。线轮廓包含热运动、湍流、旋转、外流和多速度分量等动力学信息。H I 21 cm 是最基础的射电谱线之一，可同时约束中性气体分布、速度场和柱密度；分子线、复合线和脉泽线则分别提供冷分子气体、电离气体和高亮温特殊环境的观测窗口。

对干涉测量而言，谱线观测要求频谱分辨率、bandpass 稳定性、连续谱扣除、短间距恢复和数据立方体分析同时成立。进入后续章节后，这些要求会逐渐转化为具体的相关器设置、成像参数、权重选择和科学产品构造方法。
