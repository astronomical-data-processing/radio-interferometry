# 第 6 章：成像中的去卷积<a id='deconv:sec:intro'></a>


在综合孔径成像中，成像与去卷积构成相互关联的两个步骤。成像步骤使用选定的权重和成像算子，将可见度映射到图像域；去卷积步骤则利用该算子的点源响应，即点扩散函数（PSF），以及关于天空结构的先验，求解欠定的反问题。去卷积得到的是与观测数据、噪声模型和模型先验相容的天空模型，而不是由脏图像唯一确定的“真实天空”。

上一章已经说明，最终图像会受到阵列构型、带宽、观测时长、指向赤纬、权重和网格化方式的共同影响。只有在校准解与成像参数固定、成像算子保持线性，并且可以采用窄视场标量和平移不变近似时，才能用一幅由算子而非天空决定的 PSF，将脏图像表示为表观天空与 PSF 的卷积。宽视场项、时变响应和方向相关响应会产生位置相关的点源响应；自校准会使校准解依赖天空模型，自动掩膜和数据依赖的停止准则也会使重建映射具有非线性。因此，去卷积只能在给定成像算子的条件下分离可辨识的天空结构与仪器响应，不能保证将二者完全分离。

“将脏图像变为洁净图像”只是对去卷积结果的非正式描述。严格地说，去卷积利用已知成像算子和源形态先验，通过 CLEAN 及其变体构造天空模型、残图和恢复图，并利用残差统计与参数稳定性检验该解是否足以支持具体的科学测量。

本章以第 4 章的测量方程和第 5 章的成像算子为先修内容。核心内容为 6.1--6.4，依次讨论天空模型、点源 CLEAN、主循环与次循环以及残图质量；6.5 作为源搜索与选择函数的课程训练；6.6 使用正则化语言统一讨论模型先验、过拟合和残差统计。章末习题分为 60 分核心题、18 分提高题和 22 分课程训练题，完整研究生学习路径总分为 100 分。图像单位和归一化沿用[第 5 章的约定表](../5_Imaging/5_0_introduction.ipynb#imaging:sec:normalization_units)。
