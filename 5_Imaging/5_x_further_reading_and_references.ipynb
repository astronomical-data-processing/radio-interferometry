## 5.x 延伸阅读与参考文献 <a id='imaging:sec:refs'></a>


### 综合成像总论

1. [<cite data-cite='taylor1999synthesis'>Synthesis Imaging in Radio Astronomy II（Lecture 7）</cite>](https://ui.adsabs.harvard.edu/abs/1999sira.conf.....T)：系统讨论成像、加权和采样函数，是本章基础内容的重要参考。
2. [<cite data-cite='thompson2008interferometry'>Interferometry and Synthesis in Radio Astronomy（Chapter 10）</cite>](https://doi.org/10.1002/9783527617845)：可与 *Synthesis Imaging in Radio Astronomy II* 第 7 讲配合阅读，作为综合孔径成像理论的补充。

### 权重函数

1. [<cite data-cite='briggs1995high'>High Fidelity Deconvolution of Moderately Resolved Sources</cite>](https://www.aoc.nrao.edu/dissertations/dbriggs/)：提出鲁棒权重，并对不同权重方案进行了较系统的讨论。

### 宽视场效应

1. [<cite data-cite='taylor1999synthesis'>Synthesis Imaging in Radio Astronomy II（Lecture 19）</cite>](https://ui.adsabs.harvard.edu/abs/1999sira.conf.....T)：进一步讨论宽视场成像问题。
2. [<cite data-cite='cornwell2008noncoplanar'>The noncoplanar baselines effect in radio interferometry: The W-projection algorithm</cite>](https://doi.org/10.1109/JSTSP.2008.2005290)：详细介绍 $w$ 投影算法。
3. [<cite data-cite='cornwell1992radio'>Radio-interferometric imaging of very large fields: The problem of non-coplanar arrays</cite>](https://ui.adsabs.harvard.edu/abs/1992A%26A...261..353C)：概述非共面基线问题和分面成像方法，并讨论其他相关技术。
4. [<cite data-cite='bhatnagar2008correcting'>Correcting direction-dependent gains in the deconvolution of radio interferometric images</cite>](https://doi.org/10.1051/0004-6361:20079284)：讨论 $A$ 投影以及方向相关增益如何进入成像算子。
5. [<cite data-cite='offringa2014wsclean'>WSCLEAN</cite>](https://doi.org/10.1093/mnras/stu1368)：从现代实现角度讨论 $w$ 分层、宽视场成像性能和计算取舍。

### 宽带成像

1. [<cite data-cite='rau2011multi'>A multi-scale multi-frequency deconvolution algorithm for synthesis imaging</cite>](https://doi.org/10.1051/0004-6361/201117104)：讨论 MT-MFS 的测量方程、泰勒展开项及误差范围。

### 网格化与反网格化

1. [<cite data-cite='rhodes1970spheroidal'>On the Spheroidal Functions</cite>](https://cdm16009.contentdm.oclc.org/digital/collection/p13011coll6/id/61576)
2. [<cite data-cite='tan1986aperture'>Aperture-synthesis mapping and parameter estimation</cite>](https://ethos.bl.uk/OrderDetails.do?uin=uk.bl.ethos.384529)：进一步讨论网格化与反网格化卷积函数。
3. 本章代码中的伴随内积测试应与任何生产级网格化算子共同使用；该测试能够发现索引、复共轭和 FFT 归一化错误，但不能单独证明抗混叠误差已经足够小。
