## 8.5 校准退化、模型不完备与解可信度

第 8 章前面把校准写成测量方程和最小二乘求解问题，并介绍了方向无关外场校准、方向无关自校准和方向相关校准。本节是全章的解可信度标准页：前文关于模型错误、留出残差和科学量的简短预警，都以这里的完整判据为准。真实处理中，求出增益解不等于校准正确，拟合残差降低也不等于科学量更准确；可信度来自标尺、参考规范、模型完整性、解间隔、转移稳定性、残差结构和最终科学量等多条证据的一致性。

本节集中讨论三个问题。第一，校准方程存在退化和规范自由度，必须由参考天线、通量标尺或相位约束固定。第二，天空模型不完备会使自校准把真实结构或模型缺陷吸收到增益中。第三，解可信度需要分级表达：可用于候选成像、可用于图像质量改善、可用于科学测量，所需证据并不相同。


![校准规范自由度](figures/calibration_degeneracy_gauge_freedom.png)

图 8.5.1 展示校准退化的基本思想。某些增益和天空模型的共同变换不会改变可见度拟合，必须由参考天线、外部通量/偏振标尺等约束固定。相位中心是同时定义数据相位与天空坐标的坐标约定，不应与增益全局相位规范混为同一个自由度。退化不是求解器故障，而是测量方程本身允许的自由度。


### 8.5.1 校准退化与规范选择

以方向无关增益为例，观测可见度可写成

$$
V_{pq}^{\rm obs}=g_p g_q^* V_{pq}^{\rm model}+n_{pq}.
$$

若所有天线增益同时乘上共同相位因子 $e^{i\phi}$，则 $g_p g_q^*$ 不变，固定天空模型也无法确定这个增益相位零点，因此需要参考天线或等价约束。幅度尺度退化则只在天空通量也未知时出现：对任意正实数 $a$，联合变换

$$
g_p' = a g_p,\qquad V_{pq}^{\rm model\,'}=V_{pq}^{\rm model}/a^2
$$

保持预测可见度不变；若模型通量由外部标尺固定，增益幅度原则上已被约束。极化校准中还会出现交叉手相位与绝对偏振角、泄漏与源偏振，以及平行手幅度比与 Stokes $V$ 之间的退化；方向相关校准中，每个方向的天空与增益自由度更复杂。

规范选择本身不是问题，但必须记录。参考天线变化和通量标尺模型变化会改变校准表的数值表达；相位中心变化则要求可见度相位、$uvw$ 与天空坐标一致重参考。若报告只比较增益曲线而不先对齐这些约定，很容易把等价表示误判为物理差异。


### 8.5.2 模型不完备与自校准偏差

自校准依赖当前天空模型。若模型缺少扩展结构、弱源、离轴强源或正确谱指数，求解器会试图用天线增益解释这些残差。这样做可能降低图像残差，却同时改变目标源通量、形态或光谱。自校准因此不是无条件的质量提升，而是一个模型和数据之间的闭环。


![模型不完备与校准偏差](figures/calibration_model_incompleteness_bias.png)

图 8.5.2 展示模型不完备如何影响校准。真实天空包含模型缺失结构时，求解增益可能吸收残差，最终导致目标图像的通量或形态偏差。自校准改善残差的同时，也可能改变科学量。


一个实用判断是比较自校准前后的稳定量、预测残差和科学量。纯天线基复增益校正按代数不会改变闭合相位或闭合振幅，所以“闭合量校正前后不变”只是实现一致性检查，不能证明某组增益正确；更有信息的是观测闭合量是否与天空模型预测一致，以及是否存在基线基非闭合残差。还应比较留出数据残差、强源和目标区域通量、位置、偏振、残差空间结构、解平滑性及不同模型/解间隔下的产品。若只有极短解间隔改善训练残差而留出残差或科学量变差，应怀疑过拟合。


### 8.5.3 解间隔、自由度与信噪比

解间隔决定名义校准自由度。过短间隔可以跟踪快速变化，但每个解的有效基线、带宽和样本减少，并增加把噪声吸收到增益中的风险；过长间隔信噪比较高，却可能无法跟踪真实变化。参数平滑、共享屏和先验会降低有效自由度，因此不能只用“数据点数减参数数”判断。合适解间隔应由物理时间尺度、模型可见度信噪比、连通性、正则化和科学目标共同决定。


![解间隔折中](figures/calibration_solution_interval_tradeoff.png)

图 8.5.3 展示解间隔选择的折中。过短、合适和过长分别对应不同的解信噪比、跟踪能力和过拟合风险。解间隔不是经验旋钮，而是决定自由度和物理可解释性的参数。


### 8.5.4 校准解可信度分级

校准解可信度不应只有“通过/失败”两种状态。关键标尺或参考约束缺失时，解不可用于相应科学测量；证据不完整但训练残差改善时，解只能作为候选；留出残差受控且转移稳定时，解可用于成像；只有当标尺、模型敏感性、残差、转移、注入恢复和误差预算共同支持时，解才可进入科学报告。求解器给出的形式协方差只覆盖所假设模型内的随机误差，不能替代模型版本、解间隔和方向自由度的系统敏感性测试。


![校准解可信度阶梯](figures/calibration_solution_trust_ladder.png)

图 8.5.4 把校准解可信度分成不可用、候选可用、可用于成像和可用于科学测量。可信度等级来自标尺、模型、残差、转移和科学量的一致性。


### 8.5.5 教学案例：自校准后通量改变

设一幅连续谱图像在 phase-only 自校准后动态范围提升，但目标扩展源积分通量下降。相位解没有直接重缩放可见度振幅，通量变化仍可通过相干恢复、模型位置锁定、去卷积与 mask 更新间接产生。若初始模型主要由致密源组成，扩展结构缺失，其可因错误相位解和后续模型闭环被削弱。合理处理应回到模型、mask 和解间隔，比较留出数据残差、扩展源通量、位置、残差结构及观测闭合量与模型闭合量，而不是把闭合量前后不变当作成功证据。

这个案例说明，校准可信度必须回到科学量。对点源动态范围有利的校准，不一定适合扩展通量测量；对成像残差有利的方向相关解，也可能因自由度过多而不适合精密通量或偏振测量。


### 8.5.6 本节结论

校准解是模型、数据、权重、约束和规范选择共同作用的结果。退化需要由参考天线和外部标尺固定，模型不完备会产生校准偏差，解间隔与正则化共同决定有效自由度。判断校准是否可信，不能只看训练残差，而应检查标尺、连通性、模型敏感性、留出残差、解平滑性、转移稳定性、闭合量与模型的一致性、注入恢复和最终科学量。

[第 9.38 节](../9_Practical/9_38_bima_measurement_set_calibration_replay.ipynb)把这些判据应用到真实 BIMA Measurement Set：用固定的训练/留出基线比较不同解区间，生成带参考天线、权重、flag 和逐时间残差的相对增益表，并明确说明缺少绝对通量模型时结论必须停在哪一级。
