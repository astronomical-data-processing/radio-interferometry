# 射电干涉数据处理实践

本书是《射电干涉孔径合成基础》的实践姊妹册，面向高年级本科实践课、研究生课程、暑期学校和研究训练。第一册负责建立辐射、Fourier、可见度、成像、观测系统和校准原理；本书把这些原理组织成数据检查、校准、成像、科学测量、质量控制和可复现项目。

当前文件沿用原第 9 章的 `9.x` 编号，以保持已有链接和课程材料有效。后续版本会先完成去重和内容重组，再决定是否重新编号；编号迁移不应先于科学结构稳定。


## 先修要求与环境边界

读者应理解第一册第 4 至第 8 章的可见度、成像、去卷积、RIME 和校准主线；只完成基础实践时，也可以在遇到公式时回查相应章节。本书默认环境仍由仓库根目录 `requirements.txt` 提供，真实轻量样例不要求 CASA 或 `python-casacore`。

CASA、WSClean、DP3/DDFacet、CARTA、PyBDSF、SoFiA 和大型公开归档数据属于课程扩展环境。使用这些工具时，应单独记录软件版本、数据身份、参数和运行日志，不把平台相关依赖加入基础教材环境。


## 四条阅读路径

| 路径 | 建议内容 | 目标 |

| 基础处理 | 9.1--9.6、9.14、9.16--9.18 | 完成检查、校准、连续谱成像、QA 和基本测量 |

这些路径不是严格先后关系。课程可以先用 9.36--9.38 的真实样例建立数据感，再回到相应讲授页补齐原理；也可以先完成基础处理主线，再选择一个科学专题深化。


## 页面类型

本书把页面标为四类。**讲授页**解释工作流背后的物理和统计判断；**实验页**包含可运行代码、样例数据和定量产物；**参考页**整理软件生态、特殊体制或进阶算法；**项目页**给出里程碑、量规和提交物。层级标签中的**核心**适合基础教学主线，**进阶**供专题选读，**课程**用于综合作业。页面类型不同，不意味着科学标准不同，但读者不必把参考页或课程页当作线性必修内容。完整分类同时记录在 `book_manifest.yaml`，并由测试检查。

## 学习成果与结论边界

完成基础路径后，读者应能说明每一步改变的是数据、模型、权重、flag 还是图像解释，并能用残差、噪声、通量、位置或选择函数判断结果。完成课程项目后，读者还应能提交数据清单、参数、运行记录、QA 证据、误差预算和限制声明。

实践训练不要求每个数据集都产生可发表图像。识别数据不适合某个科学问题、拒绝没有绝对标尺的 Jy 结论，或在证据不足时降低结论等级，都是合格且重要的学习成果。
