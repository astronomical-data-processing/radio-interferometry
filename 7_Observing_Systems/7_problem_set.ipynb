# 第 7 章定量问题集

本问题集共 100 分，把接收机、相关器、RIME、主波束、传播效应与 RFI 放进同一条定量链。所有题目都应写出单位、关键中间量以及采用的极化、基线和归一化约定；只给最终数字不能获得满分。[rime_lib.py](rime_lib.py) 核查 E 题的亮度和基线约定，[system_checks.py](system_checks.py) 可复现 A、B、D、F 的核对量，但提交仍须给出推导和适用边界。


## A. 从系统温度到图像热噪声（15 分）

一个由 27 面相同天线组成的阵列，口径 $D=15\ {\rm m}$，孔径效率 $\eta_a=0.70$，系统温度 $T_{\rm sys}=30\ {\rm K}$。先由

$$
A_{\rm eff}=\eta_a\pi(D/2)^2,\qquad
\mathrm{SEFD}=\frac{2k_{\rm B}T_{\rm sys}}{A_{\rm eff}}10^{26}\ {\rm Jy}
$$

计算单天线 SEFD。再取双极化 $n_{\rm pol}=2$、有效带宽 $\Delta\nu=1\ {\rm GHz}$、积分时间 $t=3600\ {\rm s}$、系统效率 $\eta_s=0.90$，用

$$
\sigma=\frac{\mathrm{SEFD}}{\eta_s\sqrt{n_{\rm pol}N(N-1)\Delta\nu t}}
$$

估计自然权重图像的热噪声。若实际残差 RMS 为 $18\ {\rm \mu Jy\,beam^{-1}}$，列出至少两种不能简单归入热噪声的可能原因。


## B. 相关器算量与数据率（15 分）

一个 64 天线、双线偏振阵列输出全部四个相关积，通道数为 4096，积分时间为 $1\ {\rm s}$。忽略自相关，每个复可见度以 8 byte 保存。计算基线数、每秒可见度数、数据率和一天的数据量。

若处理带宽为 $400\ {\rm MHz}$，X-engine 包含自相关，并把每个天线对、极化积和输入采样对应的一次复乘加计为一个操作，估计复乘加率：

$$
R_X=\frac{N(N+1)}{2}\,n_{\rm prod}\,\Delta\nu.
$$

最后说明天线数从 64 增至 128 时，算量和原始相关输出为什么都约增加四倍，而不是两倍。


## C. Jones 链的非交换性（10 分）

考虑一个差分增益和一个泄漏项

$$
G=\begin{pmatrix}1&0\\0&2\end{pmatrix},\qquad
D=\begin{pmatrix}1&0.1\\0&1\end{pmatrix}.
$$

分别计算 $GD$ 与 $DG$，再求 $\|GD-DG\|_{\rm F}$。解释为什么在 RIME 中交换馈源泄漏、增益、旋转或主波束 Jones 项的次序通常会改变物理模型。指出在什么特殊条件下两个 Jones 项可以交换。


## D. 泄漏引起的假偏振（10 分）

对一个本征无偏振点源，采用一阶近似

$$
\frac{V_{xy}}{V_I}\simeq d_{x,p}+d_{y,q}^*,
$$

其中 $V_I=I/2$ 是本章归一化下未偏振源的理想单个平行手尺度。若 $d_{x,p}=0.020$，$d_{y,q}=0.010\exp(i30^\circ)$，计算交叉手污染的复数值、幅度和相位。若把交叉手幅度直接解释为线偏振比例，会得到多大的假偏振？

进一步说明为什么只观测一个未知偏振源通常不能把源本征偏振、D-term 和交叉手相位完全分开。回答应联系校准源先验或随视差角变化的观测。


## E. 从矩阵 RIME 到四个相关积（20 分）

在线偏振基中取

$$
B=\frac{1}{2}\begin{pmatrix}I+Q&U+iV\\U-iV&I-Q\end{pmatrix},\qquad
J_p=\begin{pmatrix}g_{xp}&0\\0&g_{yp}\end{pmatrix}.
$$

从 $V_{pq}=J_pBJ_q^H$ 展开 $V_{xx}$、$V_{xy}$、$V_{yx}$ 和 $V_{yy}$。标出哪些相关积直接含有 $I\pm Q$，哪些含有 $U\pm iV$。然后证明所有天线增益同时乘以同一个单位模复数时，可见度不变，并说明这对应第 8 章校准中的哪一种退化。

扩展题：若加入非对角 D-term，原本只位于平行手的 Stokes $I$ 会通过哪些矩阵乘法路径进入交叉手？


## F. 主波束色度与谱指数偏差（12 分）

用圆对称高斯功率波束近似一面 $D=25\ {\rm m}$ 天线：

$$
P(\theta,\nu)=\exp\left[-4\ln2\left(\frac{\theta}{\theta_{\rm FWHM}}\right)^2\right],
\qquad
\theta_{\rm FWHM}=1.02\frac{c}{\nu D}.
$$

一个点源离轴 $12$ arcmin，本征谱指数为 $\alpha=-0.7$，即 $S\propto\nu^\alpha$。分别计算 1.4 GHz 与 2.8 GHz 的波束 FWHM 和该方向的 $P$。若不做主波束校正，用两频表观通量求出的谱指数 $\alpha_{\rm app}$ 是多少？若校正前两幅图的局部 RMS 都是 $10\ {\rm \mu Jy\,beam^{-1}}$，主波束校正后该位置的 RMS 各是多少？最后说明为什么波束模型误差在高频结果中更危险。


## G. 电离层与对流层相位标度（10 分）

在 150 MHz，取两天线方向上的差分电离层柱密度 $\Delta\mathrm{TEC}=0.05$ TECU。用第 7.7 节公式计算基线电离层相位和群时延。说明一个稳定、已知的相位偏移是否会自动降低可见度幅度，以及什么操作才会造成去相干。

再设 100 GHz 观测中的差分湿路径 RMS 为 $\sigma_L=0.20\ {\rm mm}$。计算相位 RMS

$$
\sigma_\phi=\frac{2\pi\nu}{c}\sigma_L
$$

以及高斯相位涨落下的平均相干度 $\gamma=\exp(-\sigma_\phi^2/2)$。比较两种介质的频率标度，并各给出一种观测或校准缓解手段。


## H. RFI 标记质量与灵敏度代价（8 分）

一个含 $10^6$ 个时频样本的数据块中，真 RFI 占 5%。某标记器的 recall 为 0.90、precision 为 0.75。计算 $TP$、$FN$、$FP$、总标记比例，以及干净样本中的误标比例。若所有样本权重相同，仅考虑积分时间损失，热噪声相对未标记数据增加多少？

解释为什么不能仅凭总 flag 比例评价算法，并说明 ADC 已被强 RFI 压缩或饱和时，为什么保留未超过像素阈值的数据也未必可靠。


## 参考核对量

- A：$A_{\rm eff}\simeq123.70\ {\rm m^2}$，$\mathrm{SEFD}\simeq670\ {\rm Jy}$，热噪声约 $10.5\ {\rm \mu Jy\,beam^{-1}}$。
- B：2016 条互相关基线，数据率约 $264.2\ {\rm MB\,s^{-1}}$，一天约 $22.8\ {\rm TB}$；X-engine 约需 $3.33\times10^{12}$ 次复乘加每秒。
- C：$GD$ 的右上元为 0.1，$DG$ 的右上元为 0.2，Frobenius 范数差为 0.1。
- D：污染为 $0.02866-0.00500i$，幅度约 0.0291，相位约 $-9.90^\circ$；直接解释会得到约 $2.91\%$ 的假偏振。
- E：例如 $V_{xy}=g_{xp}g_{yq}^*(U+iV)/2$；其余三项应具有相同的天线索引和共轭结构。
- F：两频 FWHM 约为 30.03 和 15.02 arcmin，$P\simeq0.642$ 和 0.170，$\alpha_{\rm app}\simeq-2.62$；校正后 RMS 约为 15.6 和 $58.7\ {\rm \mu Jy\,beam^{-1}}$。
- G：电离层相位约 $-2.82$ rad、群时延约 2.99 ns；100 GHz 湿路径对应 $\sigma_\phi\simeq0.419$ rad、$\gamma\simeq0.916$。
- H：$TP=45000$、$FN=5000$、$FP=15000$，总标记 6%，干净样本误标约 1.58%；仅按样本损失估计，热噪声乘以 $1/\sqrt{0.94}\simeq1.031$。

核对量只用于检查数量级和矩阵位置，不能替代推导。
