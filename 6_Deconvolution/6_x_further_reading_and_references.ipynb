## 6.x 延伸阅读与参考文献 <a id='deconv:sec:refs'></a>


第 6 章的重点不只是执行一次 CLEAN，而是将去卷积理解为包含模型假设、停止准则、质量评估和目录选择函数的反问题流程。本节按概念主线、经典文献、现代算法、软件实践和源搜索组织延伸阅读；建议阅读后使用 [6.P 去卷积综合问题集](6_problem_set.ipynb) 检查能否贯通各环节。阅读时应重点说明每个参数改变了哪一项模型假设、残差结构或计算折中。

### 6.x.1 本章主线回顾

第 6 章可以概括为从脏图像到科学目录的处理链：脏图像和 PSF 给出去卷积问题的观测形式；天空模型给出允许的解空间；次循环与主循环在模型域和数据一致性之间迭代；恢复图像和残图用于评价解的可信度；源搜索和参数拟合则将图像转化为可检验的科学量。

`6.1` 和 `6.2` 说明点源先验如何将病态反问题转化为可执行算法。`6.3` 区分算法思想和工程实现：Högbom、Clark、Cotton-Schwab、多尺度和多频率方法，是针对数据规模、视场、谱结构和计算资源作出的不同折中。`6.4` 和 `6.5` 进一步决定图像是否适合科学分析；若缺少残图、局部噪声、清理偏差、成像伪影和源拟合误差诊断，成像误差可能在后续目录中表现为科学偏差。

### 6.x.2 教材与讲义

[<cite data-cite='taylor1999synthesis'>Synthesis Imaging in Radio Astronomy II</cite> &#10548;](https://ui.adsabs.harvard.edu/abs/1999sira.conf.....T) 是理解综合成像传统方法的经典讲义，适合与本章并读，用来把教学示例接回真实阵列的数据处理流程。*Interferometry and Synthesis in Radio Astronomy*（Thompson, Moran, Swenson）是系统性最强的参考书，适合查阅公式、近似条件和工程限制。NRAO 的 [Essential Radio Astronomy &#10548;](https://science.nrao.edu/opportunities/courses/era) 对去卷积细节展开较少，但擅长把图像、噪声、角分辨率、亮温和动态范围放回完整射电观测语境。*Tools of Radio Astronomy* 更偏向仪器和观测系统背景，可用于补足接收机、噪声来源和观测误差对成像结果的影响。

### 6.x.3 CLEAN 家族与现代去卷积

[<cite data-cite='1974A&AS...15..417H'>Högbom (1974)</cite> &#10548;](https://ui.adsabs.harvard.edu/abs/1974A%26AS...15..417H) 是经典图像域 CLEAN 的起点，直接展示了“点源字典与逐步减去 PSF”的思想。[<cite data-cite='1980A&A....89..377C'>Clark (1980)</cite> &#10548;](https://ui.adsabs.harvard.edu/abs/1980A%26A....89..377C) 将 CLEAN 推向适用于大图像和大数据量的工程实现。[<cite data-cite='1984AJ.....89.1076S'>Schwab (1984)</cite> &#10548;](https://ui.adsabs.harvard.edu/abs/1984AJ.....89.1076S) 进一步明确了主循环、次循环与可见度一致性的关系。

扩展源和宽带成像需要更丰富的模型先验。[<cite data-cite='2008ISTSP...2..793C'>Cornwell (2008)</cite> &#10548;](https://arxiv.org/abs/0806.2228) 的多尺度 CLEAN 说明天空先验可以从纯点源推广到有限尺度分量。[<cite data-cite='2011A&A...532A..71R'>Rau 与 Cornwell (2011)</cite> &#10548;](https://arxiv.org/abs/1106.2745) 的多尺度多频率联合去卷积是宽带连续谱成像的重要入口。[<cite data-cite='2014MNRAS.444..606O'>Offringa 等（2014）</cite> &#10548;](https://arxiv.org/abs/1407.1943) 以 WSClean 展示了现代成像器如何将宽视场、并行计算和多种去卷积策略纳入统一软件。[<cite data-cite='2015A&A...576A...7D'>Dabbech 等（2015）</cite> &#10548;](https://arxiv.org/abs/1412.5387) 则提供了进入稀疏重构和一般反问题方法的入口。

### 6.x.4 宽视场、模型先验与联合推断

[<cite data-cite='1992A&A...261..353C'>Cornwell & Perley (1992)</cite> &#10548;](https://ui.adsabs.harvard.edu/abs/1992A%26A...261..353C) 是宽视场成像和非共面基线问题的经典文献，有助于理解去卷积为什么不能脱离成像几何。[<cite data-cite='2002ApJ...570..447C'>Chang & Refregier (2002)</cite> &#10548;](https://arxiv.org/abs/astro-ph/0107085) 用 shapelet 基进行源建模，展示了点源和高斯之外的函数基思路。[<cite data-cite='2015MNRAS.450.1308L'>Lochner et al. (2015)</cite> &#10548;](https://arxiv.org/abs/1501.05304) 从贝叶斯建模和不确定性传播角度提醒去卷积不只是得到一张较干净的图像，还涉及模型比较、误差刻画和后验推断。

本章目前仅建立这些方向的入口，尚未系统展开方向相关去卷积、联合校准与成像、偏振宽带联合重建、镶嵌成像与单天线数据联合去卷积等高级主题。这些内容是第 6 章的自然延伸，核心问题仍是在不完整、有噪声并包含系统误差的观测约束下，利用合适先验恢复可解释的天空模型。

### 6.x.5 源搜索、局部噪声与目录构建

[PyBDSF 文档 &#10548;](https://pybdsf.readthedocs.io/en/latest/) 可对应 `6.5` 中的均值图、RMS 图、高斯拟合、高斯分组、形状基函数（shapelet）或小波描述，以及偏振和谱指数等概念。[AegeanTools 文档 &#10548;](https://aegeantools.readthedocs.io/en/stable/) 除 `aegean` 本体外，还包含 `BANE` 等局部背景和局部噪声估计工具，可作为局部阈值方法的实践参照。Hancock 等（2012）的 [*Compact continuum source-finding for next generation radio surveys* &#10548;](https://arxiv.org/abs/1202.4500) 讨论下一代巡天中的完整率、可靠率和参数恢复；Hancock 等（2018）的 [*Aegean 2.0* &#10548;](https://arxiv.org/abs/1801.05548) 展示了源搜索器如何适应大规模图像和复杂科学场景。[Hydra II: source finder comparison &#10548;](https://arxiv.org/abs/2304.14357) 等基准测试可用于比较 `PyBDSF`、`Aegean` 和 `Selavy` 等软件在不同源形态与噪声条件下的偏差。

[<cite data-cite='1997PASP..109..166C'>Condon (1997)</cite> &#10548;](https://ui.adsabs.harvard.edu/abs/1997PASP..109..166C/abstract) 系统推导了相关图像噪声下椭圆高斯拟合参数的误差，是理解峰值、积分通量、位置和尺寸协方差的重要入口。源搜索的关键不只是阈值取几倍 RMS，还包括背景是否平稳、局部 RMS 是否可信、独立波束试验数、相邻源是否需要源分离、拟合模型是否与合成波束一致，以及完整率、可靠率、爱丁顿偏差和分辨率偏差是否经过注入恢复量化。图像质量评价与源目录构建应放在同一个验证闭环中理解。

### 6.x.6 软件与实践路径

[CASA `tclean` 文档 &#10548;](https://casadocs.readthedocs.io/en/latest/api/tt/casatasks.imaging.tclean.html) 把 `hogbom`、`multiscale`、`mtmfs`、`auto-multithresh`、`wproject` 等选项放在同一任务框架内，是理解参数组合与去卷积策略关系的重要资料。[WSClean deconvolution 手册 &#10548;](https://wsclean.readthedocs.io/en/latest/deconvolution.html) 系统整理了 `auto-masking`、`local RMS thresholding`、`wideband deconvolution`、`multi-scale cleaning`、`MORESANE` 和压缩感知相关模式，适合与 `6.3` 到 `6.5` 对照阅读。

实践上，可先复现 `6.2` 至 `6.4` 的合成示例，解释每次迭代为何停止以及残图中仍保留哪些结构；随后改变源形态、噪声水平、阈值和 PSF 宽度，考察算法的适用边界；最后转向真实连续谱图像，使用独立源搜索器生成目录，并将目录与残图、RMS 图和波束参数交叉检查。去卷积、图像质量评价和源搜索应作为同一推断流程的三个环节。
