# 射电干涉教材系列


本系列面向本科高年级及以上读者，由两册相互衔接的教材组成。第一册《射电干涉孔径合成基础》系统介绍辐射、数学、坐标、可见度、成像、去卷积、观测系统和校准；第二册《射电干涉数据处理实践》把这些原理接到数据检查、校准、成像、科学测量、质量控制和可复现课程项目。

两册采用统一的符号约定、章节链接和问题集结构。第一册建立的测量方程、成像算子和校准模型，会在第二册中通过数据检查、参数选择、质量控制和可复现报告得到检验；实践中出现的限制也应回溯到相应的物理假设和数学模型。


### 先修要求与学习路径

读者应具备大学物理、单变量微积分和基础 Python 阅读能力；复数、线性代数、概率统计与天球坐标可在第 2、3 章边学边补。公式推导默认读者愿意检查量纲、符号约定和近似条件，而不要求事先接触射电天文软件。

| 路径 | 建议范围 | 完成后应达到的能力 |
|---|---|---|
| 第一册核心 | 第 1 至第 6 章，第 7.1--7.6 节和第 8.1--8.3 节 | 从科学目标估算分辨率与灵敏度，推导二维可见度关系，解释 `uv` 覆盖、脏波束与 CLEAN，并读懂基本 Jones/RIME 和校准模型 |
| 第一册深化 | 完成第 1 至第 8 章及章末问题集 | 分析宽场、极化、传播、校准退化、模型不完备、残差统计与系统误差 |
| 第二册实践 | 从[实践册目录](../9_Practical/9_0_introduction.ipynb)选择基础处理、科学专题、可复现工作流或课程项目路径 | 完成数据处理、科学测量、QA 复查、误差预算和结论边界声明 |

无论选择哪条路径，都不应只以运行代码作为完成标准。最低检验包括：能够说明所用 Fourier、基线和极化约定；能从数据或图像中指出至少一种失败模式；能把最终科学量与热噪声、标度误差、模型误差和选择效应联系起来。
