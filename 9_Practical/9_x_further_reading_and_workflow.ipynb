## 9.x 延伸阅读与后续实践方向

本册围绕射电干涉数据处理的完整测量链展开。基础部分包括数据检查、数据标记、校准、成像、自校准和图像质量评估；专题部分包括平均展宽、谱线、宽带宽场、偏振、短间距、方向相关效应、极长基线干涉以及低频和高频特殊观测；科学产品部分包括图像测量、连续谱源表、谱线物理量、法拉第诊断和归档数据再分析；可复现实践部分则讨论处理流水线质量控制、项目结构、数据溯源、发布包、复查量规和综合课程项目。

进一步学习时，应始终将软件操作与测量方程对应起来，明确每一步改变的是原始数据、权重、校准解、天空模型、图像产品还是科学解释。只有保留处理选择、质量证据和误差边界，软件生成的产品才能成为可以复查的测量结果。

建议结合 CASA Guides、WSClean 文档、DDFacet 与 killMS 文档、PyBDSF 连续谱源搜索文档、SoFiA 谱线源搜索文档、RM-Tools 与 RM 综合材料，以及各望远镜的数据处理手册继续学习。阅读软件文档时，应重点识别其默认假设、适用频段、校准策略、产品定义和质量诊断方式，而不应仅复现命令序列。

本册的 `practical_metrics.py` 提供波束面积、Jy/beam 图像孔径积分、平均展宽、偏振角与高信噪比去偏置、归一化 RM 综合、考虑主波束的镶嵌以及 H I 质量计算等简洁实现。这些函数用于把教材公式转化为可以测试的计算基准，不能替代真实成像器中的点扩散函数与协方差处理、宽带波束模型、复杂偏振似然或完整的谱线辐射转移模型。


### 后续实践路线

后续训练可分为四个阶段。

1. **基础处理阶段**：选取一个规模较小的连续谱数据集，完成数据检查、射频干扰标记、方向无关初始校准、基础成像和仅相位自校准，并依据校准解、残差图和通量稳定性评价处理质量。
2. **科学产品阶段**：利用第 9.14--9.21 节的方法，将处理链延伸到连续谱源表、成像参数决策、图像科学测量、谱线物理量、偏振与法拉第诊断，以及短间距和镶嵌数据的通量恢复验证。
3. **专题算法阶段**：依据第 9.22--9.27 节，研究宽带宽场耦合、方向相关校准、观测设计、归档再分析、软件生态、VLBI、特殊频段和处理流水线质量控制。训练重点是理解算法的适用边界及其与物理误差项的对应关系。
4. **可复现项目阶段**：依据第 9.28--9.39 节，完成配置文件、环境记录、数据清单、产品溯源、源表验证、归档数据复盘、发布包、处理报告和综合项目。最终成果应能够由他人按照记录重新运行并独立复查。

每轮实践均应保存处理日志、关键诊断图、参数文件、失败案例和限制声明。训练目标不只是得到最终图像，而是能够识别失败模式、解释参数选择，并说明科学结论在何种数据范围和误差条件下成立。
