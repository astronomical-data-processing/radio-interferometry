### 1. 猎户座的位置（12 分）

下图给出了猎户座的恒星位置。

**图 4.P.1：猎户座天空示意图。**

Betelgeuse 和 Rigel 是猎户座最亮的两颗星，星座中心、Betelgeuse、Rigel 的赤道坐标为：

| 名称 | 赤经 ($\alpha$) | 赤纬 ($\delta$) |
|:---:|:---:|:---:|
| 中心 | 5h 30m $(\alpha_0)$ | $0^{\circ}$ $(\delta_0)$ |
| Betelgeuse | 5h 55m 10.3053s | $7^{\circ}~24'~25.42''$ |
| Rigel | 5h 14m 32.272s | $-8^{\circ}~12'~5.898''$ |

（1）计算猎户座中心、Betelgeuse 和 Rigel 的 $l$、$m$ 坐标。假定视场中心与猎户座中心重合。参考[第 3.4 节的方向余弦坐标系](../3_Positional_Astronomy/3_4_direction_cosine_coordinates.ipynb)。

$$
\begin{aligned}
    \Delta \alpha &=& \alpha-\alpha_0\\
    l &=& \cos \delta \sin \Delta \alpha \\
    m &=& \sin \delta \cos \delta_0 - \cos \delta \sin \delta_0\cos \Delta \alpha
\end{aligned}$$

```python
RA_rad = (np.pi/12) * np.array([5. + 30./60, 5 + 55./60 + 10.3053/3600, 5 + 14./60 + 32.272/3600])
DEC_rad = np.deg2rad([0., 7 + 24/60 + 25.42/3600, -(8 + 12/60 + 5.898/3600)])
print("RA (rad) of Sources and Field Center = ", RA_rad)
print("DEC (rad) of Sources = ", DEC_rad)

RA_delta_rad = RA_rad-RA_rad[0] #calculating delta alpha

l = np.cos(DEC_rad) * np.sin(RA_delta_rad)
m = (np.sin(DEC_rad) * np.cos(DEC_rad[0]) - np.cos(DEC_rad) * np.sin(DEC_rad[0]) * np.cos(RA_delta_rad))
print("l (dimensionless) = ", l)
print("m (dimensionless) = ", m)
```

（2）计算 Betelgeuse 在 `SIN` 投影平面中的无量纲半径 $\rho=\sqrt{l^2+m^2}$。

 

```python
D_Betel = np.sqrt(l[1]**2+m[1]**2)

print("projected radius rho (dimensionless) = ", D_Betel)
print("small-angle equivalent [deg] = ", np.rad2deg(D_Betel))
```

（3）利用前半球关系 $\rho=\sin\theta$ 求天球角距离 $\theta$，并说明反正弦函数的分支为何需要满足 $n>0$。



```python
D_Betel_celes = np.arcsin(D_Betel)
print("the angular distance from the field center to Betelgeuse on the celestial sphere (rad) = ", D_Betel_celes)
print("the angular distance from the field center to Betelgeuse on the celestial sphere (degrees) = ", 180./np.pi*D_Betel_celes)
```

（4）用 $\cos\theta=\sin\delta\sin\delta_0+\cos\delta\cos\delta_0\cos\Delta\alpha$ 独立核对角距离。



（5）解释为什么 $l,m$ 是无量纲方向余弦，却常在小视场图中以“角度等效值”显示；指出何时 $\rho$ 不能再直接当作以弧度计的 $\theta$。



（6）忽略折射，当 $\delta_0=0^\circ$ 的中心位于地平线时，求两个时角，并区分东升与西落。
