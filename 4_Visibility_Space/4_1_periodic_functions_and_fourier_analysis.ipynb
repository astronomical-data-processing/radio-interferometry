## 4.1 傅里叶分析与干涉条纹<a id='visibility:sec:periodic_fourier'></a>


第 2 章已经系统讨论傅里叶级数、连续与离散傅里叶变换、卷积和采样。本节不再重复数学推导，而是说明这些概念在干涉条纹和可见度中的物理含义。全书的指数符号和归一化沿用[第 2.0 节的标准约定](../2_Mathematical_Groundwork/2_0_introduction.ipynb#math:sec:canonical_conventions)。

干涉仪中的一条投影基线对应一组复指数条纹。二维近似下，位于 $(l_0,m_0)$ 的点源满足

$$I(l,m)=S\,\delta(l-l_0)\delta(m-m_0)\quad\Longrightarrow\quad V(u,v)=S e^{-2\pi i(ul_0+vm_0)}.$$

这一变换对概括了本章所需的基本傅里叶关系：点源通量决定可见度幅度，点源位置决定相位随基线的变化，多个互不相干的亮度成分则依据测量方程的线性性在可见度中相加。

| 图像域结构 | 傅里叶域/可见度域表现 | 干涉测量含义 |
|:---|:---|:---|
| 平滑的大尺度发射 | 集中在原点附近 | 需要短基线和零间距约束 |
| 紧致或尖锐结构 | 延伸到较高空间频率 | 长基线提高细节约束，但不补大尺度通量 |
| 整体位置平移 | 幅度不变，增加线性相位 | 相位中心和基线符号直接决定位置符号 |
| 有限边界或硬截断 | 产生振荡和旁瓣 | 视场窗、主波束和采样边界会进入响应 |

![天空与傅里叶平面的对应](../2_Mathematical_Groundwork/figures/fourier_transform_sky_uv_pair.png)

**图 4.1.1：天空亮度与二维傅里叶域的对应关系。** 幅度表示各空间频率成分的强弱，相位记录这些成分相对于所选原点的位置关系。

这里的 $uv$ 平面只适用于小视场二维近似。完整的测量方程还包含 $w$ 项、频率和时间依赖、偏振以及方向响应；[第 4.6 节](4_6_the_fourier_approximation_van_cittert-zernike_theorem.ipynb)将给出化为二维傅里叶关系所需的条件。
