## 8.6 延伸阅读与参考文献 <a id='cal:sec:refs'></a>

第 8 章把校准组织成一条从测量方程到实际解链的主线。进一步阅读时，材料大致可分为四类：基础干涉测量与 RIME、自校准和闭合量、快速求解器、方向依赖校准与宽场成像。不同材料的重点并不相同，有些适合建立理论框架，有些更适合理解真实管线中的工程选择。


### 8.6.1 校准与测量方程基础

**Thompson, Moran & Swenson, [*Interferometry and Synthesis in Radio Astronomy*](https://doi.org/10.1007/978-3-319-44431-4)** 是射电干涉领域最经典的系统教材之一。校准章节应与可见度、综合孔径成像、噪声和极化章节一起理解，因为校准并不是孤立步骤，而是测量方程的一部分。

**Smirnov (2011a), [“Revisiting the radio interferometer measurement equation. I. A full-sky Jones formalism”](https://doi.org/10.1051/0004-6361/201016082)** 建立了现代 RIME 的完整 Jones 形式，是把第 7 章仪器系统和第 8 章校准连接起来的关键文献。

**Smirnov (2011b), [“Revisiting the radio interferometer measurement equation. II. Calibration and direction-dependent effects”](https://doi.org/10.1051/0004-6361/201116434)** 将 RIME 明确推进到校准和方向依赖效应，是理解方向相关校准和误差分解方式的核心材料。


### 8.6.2 自校准、闭合量与经典成图思路

**Pearson & Readhead (1984), [“Image Formation by Self-Calibration in Radio Astronomy”](https://doi.org/10.1146/annurev.aa.22.090184.000525)** 是自校准经典文献。该文展示了自校准如何与成像模型共同迭代，也揭示了模型依赖性为什么是自校准的核心风险。

**Readhead & Wilkinson (1978), “The mapping of compact radio sources from VLBI data”** 是 hybrid mapping 思想的重要来源。VLBI 观测中绝对相位信息常常不可得，闭合量和自校准之间的关系在这一背景下非常清楚。

**Cornwell 与 Fomalont 的 synthesis imaging 相关综述** 有助于把闭合相位、闭合振幅、去卷积和图像质量控制联系起来。它们适合作为从理论方程走向实际成图判断的桥梁。


### 8.6.3 高效求解器与 3GC

**Salvini & Wijnholds (2014), [“Fast gain calibration in radio astronomy using alternating direction implicit methods”](https://doi.org/10.1051/0004-6361/201424487)** 介绍了 StEFCal，是理解大阵列快速方向无关增益求解的重要入门材料。它清楚展示了天线基结构如何被算法利用。

**Bhatnagar et al. (2008), [“Correcting direction-dependent gains in the deconvolution of radio interferometric images”](https://doi.org/10.1051/0004-6361:20079271)** 是 A-projection 的核心文献，适合用来区分“已知方向响应进入成像算子”和“从数据求未知方向增益”。

**Smirnov & Tasse (2015), [“Radio interferometric gain calibration as a complex optimization problem”](https://doi.org/10.1093/mnras/stv418)** 讨论复优化与 Wirtinger 形式；Tasse 及相关 DDFacet、killMS、faceting 工作则展示物理模型、局部方向求解和成像器耦合之间的实际折中。阅读软件论文时应分别确认哪些 Jones 项被固定、哪些被求解、采用何种规范和正则化。


### 8.6.4 稳健统计、不确定度与验证

**Kazemi & Yatawatta (2013), [“Robust radio interferometric calibration using the t-distribution”](https://ui.adsabs.harvard.edu/abs/2013MNRAS.435..597K/abstract)** 展示了重尾似然如何降低异常残差对增益解的影响。稳健损失不是 RFI 标记和天空建模的替代品；它只改变异常点在目标函数中的影响，并仍需报告有效权重和失效模式。

形式 Hessian 协方差只描述给定模型内的局部随机误差。科学用途还应检查不同模型、解间隔、参考规范和正则化下的敏感性，并使用留出时间/频率/基线、注入源恢复和校准前后科学量比较。残差下降而留出误差、通量、位置或偏振变差，是拒绝更复杂解的直接证据。


### 8.6.5 与第 9 章实践部分的衔接

进入 [第 9 章实践部分](../9_Practical/9_1_visualisation-inspection.ipynb) 后，重点不应只是记住软件命令，而是把命令和物理对象对应起来。`listobs` 与 `plotms` 对应数据质量判断，`gaincal` 与 `bandpass` 对应方向无关增益和频率响应，`applycal` 对应解转移，自校准则对应模型更新与增益求解的闭环。

第 8 章的练习页 [校准习题](8_problem_set.ipynb) 可作为进入实践章前的过渡。它把参考天线、最小二乘收敛、模型不完整和交替增益求解集中到同一组合成数据上，有助于把公式判断转化为数据诊断判断。
