# 第六章：成像中的去卷积<a id='deconv:sec:intro'></a>


在综合孔径成像里，人们通常把“成像”理解为两个相互关联的过程：首先用选定的加权与成像算子把可见度映射到图像域；其次利用该算子的点源响应（PSF）和天空先验求解欠定反问题。去卷积得到的是一个与数据、噪声模型和先验相容的天空模型，而不是由脏图唯一决定的“真实天空”。

正如上一章已经看到的，最终图像会明显受到阵列构型、带宽、观测时长、指向赤纬、权重和网格化方式的影响。只有在校准解与成像参数固定、算子线性，并可采用窄场标量和平移不变近似时，才可以用一幅由算子而非天空决定的 PSF，把脏图写成表观天空与 PSF 的卷积。宽场项、时变或方向相关响应会产生位置相关点源响应；自校准会让校准解依赖天空模型，自动 mask 和数据依赖停止准则则会使重建映射本身成为非线性的。因此，去卷积能分离的是给定算子下可辨识的天空结构与仪器响应，不能保证把二者完全拆开。

在日常说法里，人们常会说“把脏图变成洁图”。更准确地说，是利用已知成像算子和对源形态的先验，通过 CLEAN 及其变体构造模型、残图与恢复图，并用残差和参数稳定性检验这个解是否足以支持具体科学测量。

本章以第 4 章测量方程和第 5 章成像算子为先修。核心路径是 6.1--6.4：天空模型、点源 CLEAN、主循环/次循环和残图质量；6.5 是源搜索与选择函数的可选课程训练；6.6 是用正则化语言统一理解先验、过拟合和残差统计的拓展选读。章末问题集相应分成 60 分核心内容、18 分提高题和 22 分课程训练；完整研究生路径共 100 分。图像单位和归一化统一沿用[第 5 章约定表](../5_Imaging/5_0_introduction.ipynb#imaging:sec:normalization_units)。
