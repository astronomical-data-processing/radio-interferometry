# 第 4 章：可见度空间<a id='visibility:sec:intro'></a>


第 3 章把天空方向、台站指向和相位中心坐标串成了一条几何链。本章进入射电干涉测量的核心：干涉仪到底测量什么。单天线直接接收某个方向范围内的功率；干涉阵则把不同天线接收到的电压信号相乘并做时间平均。这个互相关结果不是图像中的一个像素，而是天空亮度分布的一个空间频率样本，称为*可见度*（visibility）。

这个观点是整个孔径合成的转折点。干涉阵不是先拍下一张模糊照片再做锐化，而是在 `uv` 平面上逐点采样天空亮度的傅里叶信息。每一条基线、每一个时刻、每一个频率频道都会给出一个不同的采样点；许多采样点共同构成 `uv` 覆盖。若覆盖足够密，并且小视场、远场、窄带等近似成立，就可以把这些可见度反变换为图像。若覆盖稀疏，反变换得到的就是带有脏波束卷积效应的脏图像。

![可见度测量链条](figures/chapter4_visibility_pipeline.png)

**图 4.0.1** 从天空亮度到可见度采样的测量链条。相关器把两台天线的电压信号转换为某条投影基线上的复可见度；许多基线、时间和频率样本共同形成可见度空间中的采样。

### 本章符号约定<a id='visibility:sec:conventions'></a>

下表把必须成套使用的定义固定在一处，并沿用[第 2.0 节的 Fourier、复统计与权重约定](../2_Mathematical_Groundwork/2_0_introduction.ipynb#math:sec:canonical_conventions)。外部软件或文献可以采用相反箭头或时间因子，但比较结果时必须同时转换整套定义，不能只改一个负号。

| 对象 | 本章定义 | 直接后果 |
|:---|:---|:---|
| 天空方向 | $\mathbf s$ 从阵列指向天源；相位中心为 $\mathbf s_0$ | 相位跟踪后使用 $\mathbf s-\mathbf s_0$ |
| 解析时间因子 | $e^{-2\pi i\nu t}$ | 正几何时间参数在电压中产生负相位 |
| 相关积 | $V_{pq}=\langle v_pv_q^*\rangle$ | 交换顺序给出 $V_{qp}=V_{pq}^*$ |
| 有向基线 | $\mathbf b_{pq}=\mathbf r_p-\mathbf r_q$，箭头从 $q$ 指向 $p$ | $\mathbf b_{qp}=-\mathbf b_{pq}$，必须与相关积同时交换 |
| 几何时间参数 | $\tau_{g,pq}=\mathbf b_{pq}\cdot\mathbf s/c$ | 单方向窄带核为 $e^{-2\pi i\nu\tau_{g,pq}}$ |
| `uvw` 与方向余弦 | $\mathbf b_{pq}/\lambda=(u,v,w)$，$\mathbf s=(l,m,n)$，$w$ 指向 $\mathbf s_0$ | 相位跟踪后的核为 $e^{-2\pi i[ul+vm+w(n-1)]}$；理论式中 `uvw` 是波长数，数据列也可能以米存储 |
| 天球分支 | 本章局部像面取 $n=+\sqrt{1-l^2-m^2}$ | 只覆盖相位中心前半球，且 $d\Omega=dl\,dm/n$ |

本章只在讨论标量可见度时省略偏振指标。第 7 章把同一相关次序扩展为亮度矩阵和 RIME；任何 Stokes 或馈源基约定都必须与这里的基线和共轭次序相容。

本章主要回答三个问题。第一，一条基线的互相关为什么对应天空亮度的一个空间频率分量；第二，基线在地球自转中如何投影成 `u,v,w` 坐标，并在 `uv` 平面上画出轨迹；第三，在什么近似下 van Cittert-Zernike 定理允许我们把可见度函数看成天空亮度的二维傅里叶变换。这三个问题连在一起，就构成了从物理接收机到成像方程的主线。

第 4.1 节是可跳过的 Fourier 桥接页，只把第二章数学翻译成干涉条纹语言。第 4.2 节定义物理基线以及它在不同参考系中的表示，并把它引向 `u,v,w` 坐标。第 4.3 节从最简单的二元干涉仪出发，说明相关器如何产生一个复测量样本。第 4.4 节解释可见度幅度、相位和角尺度，但不重复完整测量方程推导。第 4.5 节讨论地球自转、天线布局和频率综合如何改变 `uv` 覆盖。第 4.6 节是本章唯一的完整 van Cittert-Zernike 与宽场标量测量式推导。第 4.7 节继续讨论闭合量、缺短间距和 mosaicking，把可见度空间的采样约束连接到真实质量判断。
