## 9.35 综合课程项目与研究训练设计

本册前面的实践内容已经覆盖了从数据检查、校准、成像、自校准、源表、谱线、偏振、短间距、宽场宽带，到归档复盘、可复现工作流、样本包和处理报告的多个环节。若这些环节只以单节练习出现，训练会变成很多分散技能。综合课程项目的作用，是把技能重新组织成一条研究式路径：从科学问题出发，选择适合的数据，建立最小工作流，生成和验证产品，写出可复查报告，并形成带限制声明的发布包。

综合项目的目标不是完整执行所有软件命令，而是训练科学判断：评估数据对科学目标的支持程度，说明处理参数的依据，分析质量证据对结论的影响，并建立完整误差预算和明确结论边界。无论项目规模大小，每项结论都应能够追溯到相应数据、参数和证据。

本节给出适用于课程、暑期学校、研究组训练和自学的项目设计框架，并与第 9.33 节的轻量样本包和第 9.34 节的复查量规配套。框架包括项目阶段、能力评价、分层任务和项目库建设，可覆盖连续谱、谱线、偏振、短间距、宽场宽带和归档复盘等主题。

### 先修知识、学习目标与项目出口

综合项目以本册至少一条基础或科学专题链，以及 9.28--9.34 的项目材料链为先修。完成项目后，读者应能：将科学问题分解为数据适配、最小运行、产品验证和报告发布四个里程碑；为每个阶段定义停止、回退与降级条件；用能力矩阵评价概念、流程、诊断、测量和表达；根据基础、提高或研究层次裁剪任务范围。项目出口不是单一图像，而是科学问题说明、适用性记录、运行与参数身份、质量证据、测量及误差预算、复查报告、限制声明和发布包。

### 9.35.1 科学问题与综合项目

一个常见误区，是把综合项目设计成“完成某软件教程”。这样的项目能训练命令熟悉度，却很难训练科学判断。更好的起点是一个可以被数据回答或限制的科学问题。例如：一个连续谱场中某个扩展源的积分通量是否稳定；一个归档数据能否给出弥散辐射上限；一个谱线 cube 的 moment 图是否受 beam smearing 影响；一个偏振源的 Faraday 结构是否能由当前 $\lambda^2$ 覆盖分辨；一个短间距缺失的数据是否能支持总通量结论。

科学问题一旦明确，后续任务会自然展开。目标量决定频率、角尺度、速度分辨率、偏振覆盖或主波束区域；数据适配决定是否进入处理；最小工作流决定输入、环境和参数身份；产品验证决定图像、源表或 moment 图是否可信；误差预算决定结论置信度；限制声明决定哪些解释被支持，哪些仍然不能排除。

项目说明可以采用这样的句式：本项目关注某一目标量；给定数据或样本包保留了哪些证据；需要生成哪些产品；最终报告必须说明哪些限制。这样的设计让项目从一开始就带着科学方向，而不是在处理完成后才寻找解释。


### 9.35.2 停止、回退与降级里程碑

阶段项目应按里程碑推进，而不是最后一次性提交。第一个里程碑检查科学问题与数据适配：目标量是否可测，频率、尺度、灵敏度和质量证据是否足够。第二个里程碑建立最小运行与初始产品：输入、环境、参数和运行编号是否清楚，初始图像或源表是否生成。第三个里程碑进入产品验证与误差预算：残差、局部噪声、通量标尺、主波束和数据限制是否被纳入。第四个里程碑完成报告复查与发布包：结论、限制声明、清单和产品身份是否一致。

里程碑的价值在于它允许项目在合理位置停止或回退。若第一个里程碑发现目标不可测，最好的结果可能是改写问题，而不是硬做处理。若第二个里程碑发现输入或参数错误，项目应回退到清单和配置。若第三个里程碑发现证据不足，结论应降级为候选、上限或限制声明。真实研究中，停止、回退和降级并非失败，而是保持结论可信的正常机制。


![综合项目里程碑](figures/practical_capstone_milestone_ladder.png)

**图 9.35.2：综合项目里程碑。** 图 9.35.2 展示四个里程碑：科学问题与数据适配，最小运行与初始产品，产品验证与误差预算，报告复查与发布包。每个里程碑都应能形成可复查产物，并允许停止、回退或降低结论等级。


每个里程碑都应有明确产物。数据适配阶段可以提交目标量说明、数据清单和适配矩阵；最小运行阶段可以提交运行报告、参数文件和初始产品；产品验证阶段可以提交差异图、局部噪声、源表检查或上限计算；最终阶段可以提交处理报告、限制声明和发布包 manifest。这样复查不会只发生在最后，而是在项目进行中持续校正方向。


### 9.35.3 综合能力评价

综合项目的评价应覆盖概念、流程、诊断、测量和表达五个维度。概念理解关注能否把科学问题转化为可观测量和限制边界；流程组织关注能否建立可复查的运行和依赖关系；质量诊断关注能否识别异常、定位误差来源和判断模型不完备；科学测量关注能否给出通量、上限、源表、moment 或偏振量，并写出误差预算；证据表达关注能否提交可审查报告并根据复查意见修订。

评价标准可以分为基础层次、提高层次和研究层次。基础层次强调看懂数据和产品；提高层次强调能重跑、比较和解释取舍；研究层次强调能管理依赖、判断模型不完备、比较多种假设，并形成可审查结论。


![综合项目能力评价矩阵](figures/practical_capstone_assessment_matrix.png)

**图 9.35.3：综合项目能力评价矩阵。** 图 9.35.3 给出能力评价矩阵。评价不应只看最终图像，还应看能否解释目标量、组织流程、诊断误差、完成测量和表达证据。不同深度对应不同任务边界和证据要求。


这种评价方式能避免两个极端。一个极端是只看最终图像是否好看，忽视数据身份、误差和限制。另一个极端是只检查命令是否完整，忽视科学结论是否成立。综合项目应把二者连接起来：处理过程要可追溯，最终结论要可审查，图像和源表只是证据链中的一部分。


### 9.35.4 同一项目的三个层次

同一项目可以设置三个层次，而不必为不同训练层次更换完全不同的案例。基础层次可以要求看懂数据与产品：检查元数据、质量报告、图像、源表或残差，并写出初步判断。提高层次可以要求重跑与误差预算：改变成像权重、阈值、mask 或测量区域，比较产品差异并给出误差预算。研究层次可以要求完整复盘与发布包：说明数据适配、建立最小工作流、管理清单和依赖、形成限制声明，并整理可复查发布包。

三个层次可以共用同一输入，例如样本包、数据清单和图像产品；差异在于产物和证据要求。基础层次的产物可以是质量判读和简短报告；提高层次的产物可以是重跑产品、差异审计和误差预算；研究层次的产物可以是复盘报告、发布包和复查记录。这样的设计让课程材料具有连续性，也使不同阶段的训练可以相互衔接。


![同一项目的三个层次](figures/practical_capstone_depth_design.png)

**图 9.35.4：同一项目的三个层次。** 图 9.35.4 展示同一项目的基础、提高和研究三个层次。层次差异来自任务边界和证据要求，而不是换一套互不相干的案例。


例如，一个连续谱归档复盘项目在基础层次中可以只要求判断原图是否适合低面亮度测量；在提高层次中要求用 taper 重跑并比较共同 beam 下的上限；在研究层次中要求建立数据清单、质量证据板、产品 manifest、处理报告和限制声明。三者共享同一科学目标，却训练不同层次的能力。


### 9.35.5 项目库：覆盖关键科学量和失败模式

一个长期可用的项目库不需要每个案例都很大，但应覆盖关键科学量和常见失败模式。连续谱项目可以训练成像、源表、通量和动态范围；谱线项目可以训练速度约定、moment 图、mask、柱密度和上限；偏振项目可以训练泄漏、旋转量、退偏振和 Faraday 分辨率；短间距项目可以训练缺通量、mosaic、feather 和联合成像；宽场宽带项目可以训练主波束、方向依赖和频谱指数偏差；归档复盘项目可以训练证据链、限制声明和发布包。

项目库的组织方式应服务能力覆盖，而不是只按软件分类。同一个软件可以出现在多个项目中，同一个科学问题也可以用不同工具实现。关键是每个项目都说明自己训练哪些科学量、哪些处理判断、哪些误差来源和哪些报告能力。


![项目库能力覆盖图](figures/practical_capstone_portfolio_map.png)

**图 9.35.5：项目库能力覆盖图。** 图 9.35.5 展示项目库的能力覆盖。连续谱、谱线、偏振、短间距、宽场宽带和归档复盘分别覆盖不同科学量和失败模式。项目库不必每个案例都很大，但应避免所有案例都训练同一种能力。


项目库可以由三类案例构成。第一类是标准路径案例，用于建立基本流程和预期结果；第二类是失败模式案例，用于训练校准残差、主波束边缘伪检、短间距缺失、掩膜错误或源表选择函数变化等诊断；第三类是开放复盘案例，用于训练数据适用性判断、限制声明和报告修订。三类案例相互配合，使实践训练由流程复现进一步发展为依据明确的独立判断。


### 9.35.6 测量集纵向复现训练（提高）

完整的真实观测复现适合作为课程综合训练或研究组项目，不应成为教材基础路径的隐含前提。基础训练可以直接使用 [9.38 BIMA 样例](9_38_bima_measurement_set_calibration_replay.ipynb)，完成测量集身份、列结构、数据标记、$uv$ 覆盖、相对校准和证据边界判断；它不要求学生建立绝对 Jy 标度。

提高训练使用已经通过真实运行验证的 [9.39 VLA 3C391 案例](9_39_vla_3c391_archive_calibration.ipynb)，按四个阶段推进：建立归档身份、下载校验和再分发边界；检查初始测量集、3C286 通量模型、数据标记与权重语义；从教程起点生成延迟、带通和时间增益表，完成绝对标度转移与镶嵌成像；提交带参数、环境、产品校验、失败模式和限制声明的复盘报告。它补齐完整纵向链，但不把 CASA、大型数据或发表级成像变成教材基础路径要求。

建议把评分放在证据链而不是最终图像上，并使用一套可复查的 100 分量规：

| 评价项 | 分值 | 可检查产物 |
| --- | ---: | --- |
| 数据来源与许可证 | 15 | 归档身份、检索条件、许可证和数据清单 |
| 环境与可复现性 | 15 | 软件版本、参数、运行记录和校验和 |
| 标记与校准证据 | 20 | 数据标记统计、校准表诊断和转移关系 |
| 绝对通量标度判断 | 15 | 单位、通量模型和标度证据矩阵 |
| 成像与定量 QA | 20 | PSF、残差、噪声、通量或上限比较 |
| 失败模式与结论边界 | 15 | 限制声明、降级结论和改进方案 |

若证据不足以支持绝对标度，学生明确给出“不能建立该结论”同样可以是高质量结果。课程考查的是证据判断，不是强迫每组数据都产生可发表图像。

基础训练只使用仓库默认 `requirements.txt`；提高训练可以另建课程环境，按项目需要使用 CASA、`python-casacore` 或其他射电软件，并记录版本和配置。可选环境不应扩大教材默认依赖，也不应让外部网络、平台或大型数据成为基础 Notebook 回归的前置条件。


### 9.35.7 低面亮度连续谱综合项目案例

以低面亮度连续谱归档复盘为例，综合项目可以这样展开。科学问题是：某个目标区域内是否存在可由当前数据约束的弥散连续谱结构。第一个里程碑要求判断归档数据是否适合该问题，重点检查频率、最大可恢复尺度、主波束范围、pipeline 质量报告和强源残差。若最大可恢复尺度明显小于目标尺度，项目应在此处改写问题，只讨论较小尺度结构或给出新观测需求。

第二个里程碑建立最小运行。保留归档校准，使用适合低面亮度目标的权重和 taper 重新成像，记录参数、运行编号和产品身份。第三个里程碑比较原图和重跑图，卷积到共同 beam，检查局部噪声、残差、主波束裁剪和强源排除区域，并计算通量或上限。第四个里程碑形成报告和发布包，写明上限适用的角尺度、频率、主波束范围和数据限制。

基础层次可以只完成第一个里程碑和简短判读；提高层次完成重跑和误差预算；研究层次完成发布包与复查记录。这样同一项目既能训练基本概念，也能训练研究级证据链。


### 9.35.8 本节结论

综合课程项目把本册的分散实践技能组织成研究式训练。项目应从科学问题开始，以里程碑推进，以阶段产物接受复查，以能力矩阵评价概念、流程、诊断、测量和表达。不同层次的项目可以共享同一输入，通过不同任务边界和证据要求形成基础、提高和研究三个训练层次。

第 9.35 节的核心思想是：实践训练的最终目标不是完成软件步骤，而是形成可审查科学结论。数据选择、最小工作流、产品验证、误差预算、处理报告和发布包共同构成完整能力链。这样的项目设计能把教材内容落到真实训练场景，也能为后续建立项目库和样本包提供统一框架。
