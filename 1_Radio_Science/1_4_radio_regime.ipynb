## 1.4 射电波段<a id='science:sec:radio_regime'></a>


提到天文学时，我们脑海中首先浮现的往往是肉眼可见的星空：恒星、行星、银河以及少数明亮星云。但这种图景本身带有强烈的“人类中心”限制，因为它只对应极窄的一段可见光波长范围。宇宙并不会优先按照人眼最熟悉的方式发光；不同波段对应的是不同能量尺度、不同辐射机制和不同天体环境。

射电天文学的重要性正在于此。它并不是光学天文学的低能附属，而是打开了另一组物理现象：冷气体、磁场、相对论粒子、传播介质和时间域爆发，在射电波段往往比在其他波段更容易直接研究。因此，理解“射电波段是什么”，并不只是背一个频率范围，而是要理解：**为什么换一个波段，宇宙的样子和我们能提的问题都会发生变化。**

从数量级上看，地面射电天文学常处理从兆赫到太赫兹附近的电磁波，对应从数十米到亚毫米的波长。由于 $\lambda\nu=c$，同一个观测问题既可以用频率语言表达，也可以用波长语言表达。更重要的是，单个射电光子的能量很低，$h\nu/k\simeq 0.048\,{\rm K}\,(\nu/{\rm GHz})$。在 $1$ GHz 处，一个光子的等效温度只有约 $0.05$ K；即使到 $100$ GHz，也只有约 $5$ K。这意味着射电观测通常不是在数单个高能光子，而是在测量大量低能光子的电场相干性、功率、频谱结构和偏振状态。这个事实会贯穿后续关于接收机、噪声、相关器和可见度的讨论。

**图 1.4.1**：射电波段的频率、波长和典型物理对象。图中标出的边界不是严格定义，而是帮助建立数量级直觉：低频端逐渐受到电离层和传播效应控制，中频厘米波段适合大量连续谱与谱线工作，毫米和亚毫米端则越来越受大气水汽与相干时间限制。


### 1.4.1 射电天空为什么和光学天空不同

在高能端，伽马射线和 X 射线探测的是高温等离子体、激波与极端引力环境；在可见光波段，我们常直接看到恒星光球和电离气体的辐射；而在低能端的射电波段，主导的往往是冷气体、非热连续谱、谱线跃迁以及传播效应。由于不同波段响应的物理机制不同，同一片天空在不同波段上看起来往往完全不像同一个对象。

这意味着射电图像中的亮结构，并不一定对应光学上最亮的恒星或星系主体。很多时候，射电最亮的部分反而是喷流、射电瓣、冷气体盘、磁层辐射区或被尘埃遮蔽的恒星形成区。换句话说，射电波段所揭示的不是“光学天空的弱版本”，而是宇宙中另一组本来就不以可见光为主的物理成分。这个差异不是显示颜色的差异，而是物理选择函数的差异：射电波段让磁场、低温气体、等离子体传播效应和相干辐射过程进入可观测范围。


### 1.4.2 例子一：活动星系在射电上显露出隐藏结构

以天鹅座 A 为例，它是最亮的河外射电源之一，红移约为 $z=0.056$；在常用现代宇宙学参数下，其光度距离约为 250 Mpc，具体数值随所采用的宇宙学而变。若只看光学图像，我们看到的只是一个看似普通的星系所在天区；但在射电波段中，真正占主导地位的却是由中心活动星系核驱动产生的巨大喷流和射电瓣结构。这些结构在空间尺度上甚至可以远远超过光学星系主体本身。


<img src="figures/cygnus_a_optical_schematic.png" alt="Optical host-galaxy schematic" width="700">


**图 1.4.2**：天鹅座 A 光学形态的概念图。恒星光主要显示相对紧凑的宿主星系；本项目生成的示意图不代表实测图像，也不按角尺度复原。<a id='science:fig:cyg_a_comp'></a> <!--\label{science:fig:cyg_a_comp}-->


但如果改在射电波段观测，我们就能看到更多与高能粒子和磁场相关的结构：


<img src="figures/cygnus_a_radio_schematic.png" alt="Radio lobes and jets schematic" width="700">


**图 1.4.3**：天鹅座 A 类双瓣射电星系的形态示意。射电喷流把能量输运到宿主星系之外，并形成同步辐射瓣和热点；图中尺寸与亮度均为概念表达。<a id='science:fig:cyg_a_comp2'></a> <!--\label{science:fig:cyg_a_comp2}-->


这个例子说明了射电图像最典型的一个特点：它往往直接追踪 **相对论粒子、磁场和喷流反馈**，而不是恒星分布本身。天鹅座 A 两侧显著的射电瓣，本质上是等离子体结构，是活动星系核喷流与周围介质相互作用后留下的非热辐射区域。若只依赖光学观测，这些结构很容易被忽略；而在射电图像中，它们反而成为最突出的主角。

这也是为什么射电天文学在研究活动星系核反馈、星系际介质加热和大尺度喷流能量输运时如此重要。它不仅补充光学图像，更常常改变我们对目标“真正重要结构在哪里”的判断。


### 1.4.3 例子二：射电波段能看到冷气体与星系相互作用

另一个典型例子是 M81 系统。它由多个彼此引力束缚的星系组成。若只看光学图像，人们更容易注意到各星系自身的恒星盘和亮核；但在射电图像中，星系之间被潮汐相互作用拉扯出来的中性氢桥和气体尾结构会变得格外明显。


<img src="figures/interacting_galaxies_hi_schematic.png" alt="Interacting galaxies in starlight and neutral hydrogen" width="800">


**图 1.4.4**：以 M81 群为背景的相互作用星系概念图。恒星盘可以彼此分离，而 21 cm H I 辐射能够显露连接星系的低面密度气体桥和潮汐尾；示意结构不是观测数据复原。<a id='science:fig:m81'></a> <!--\label{science:fig:m81}-->


这些桥状结构在光学波段中并不显眼，因为它们并不是由大量恒星直接发光形成，而是由较冷的气体构成。射电谱线，特别是 HI 21 cm 线，使我们能够追踪这些气体在星系之间如何重新分布、被剥离、压缩或重新落回盘面。对星系演化研究来说，这种信息非常关键，因为恒星形成所需的燃料本质上就是气体，而不是光学图像中的恒星本身。


### 1.4.4 例子三：射电数据如何揭示暗物质问题

类似的思想也体现在星系旋转曲线研究中。旋转曲线描述的是轨道速度随半径的变化关系，由此可以反推出质量在星系中的分布。若星系质量几乎都集中在可见恒星盘内，按照牛顿动力学，远离中心后轨道速度应明显下降；然而实际情况并不是这样。

中性氢的射电观测可以延伸到远超光学盘的区域，因此能把旋转曲线追踪到更大半径。这一点非常重要，因为真正暴露“缺失质量”问题的，恰恰就是这些光学盘之外仍然保持较高转速的气体。下图给出了一个典型例子：


<img src="figures/rotation_curve_schematic.png" alt="Schematic galaxy rotation curve" width="760">


**图 1.4.5**：旋转曲线的物理示意。仅由可见重子物质给出的外盘速度会下降，而 H I 观测常显示更平坦的外侧曲线；阴影区强调中性氢可把测量延伸到明亮恒星盘之外。本图不是 M33 数据拟合。<a id='science:fig:gal_rot_cur'></a> <!--\label{science:fig:gal_rot_cur}-->


射电观测告诉我们：在可见恒星之外，仍有大量引力质量在控制气体运动。这是暗物质问题最早、也最经典的观测证据之一。这个例子清楚展示了一个事实：**射电并不只是“多看一种颜色”，而是能让我们测到原本在光学上根本无从获得的动力学信息。**


### 1.4.5 射电窗口及其物理边界

既然射电波段如此重要，就必须进一步问：地面望远镜到底能利用多宽的射电频率范围？答案并不是“任意低频到任意高频都可以”，因为地球大气和电离层本身就是一个强烈的频率选择滤波器。

低频端的主要限制来自电离层。电离层中存在自由电子，它们使电磁波在传播时具有等离子体色散关系。当频率接近或低于等离子体频率

$$
\nu_p={1\over 2\pi}\left({n_e e^2\over m_e\epsilon_0}\right)^{1/2}
\simeq 9\,\sqrt{n_e/{\rm cm}^{-3}}\ {\rm kHz},
$$

电磁波就难以像在真空中那样穿过介质，而会发生强烈折射、反射或吸收。实际电离层电子密度随高度、昼夜、太阳活动和地磁条件变化，因此地基射电天文学的低频边界不是一个固定常数；但数量级上，十几 MHz 以下的天文观测通常会非常困难，真正更低频的宇宙信号需要借助空间平台或月背环境来研究。

高频端的限制则主要来自中性大气，尤其是水汽和氧气分子的吸收线以及由此形成的连续不透明度。毫米和亚毫米波段的光深对可降水汽含量非常敏感，观测站址因此常选择高海拔、低水汽、气象稳定的地区。换句话说，射电窗口的两端分别由不同物理过程决定：低频端受等离子体传播控制，高频端受分子吸收和大气热辐射控制。下图给出了大气不透明度随频率变化的一个示意。


<img src="figures/atmospheric_radio_windows.png" alt="Schematic atmospheric radio transmission" width="800">


**图 1.4.6**：地面射电窗口的概念化天顶透过率。低频端受电离层截止限制，高频端受水汽连续吸收和分子谱线影响；实际透过率强烈依赖海拔、可降水量、天气与仰角，因此曲线只用于展示结构和数量级。<a id='science:fig:atm_opa'></a> <!--\label{science:fig:atm_opa}-->


从这张图可以看出，所谓“射电窗口”并不是一整块均匀可用的区域，而是包含多个透明度不同的子窗口。一般而言，地基射电天文学常把大约 $10$ MHz 到 $1$ THz 之间视作主要工作区间；再高频就逐渐进入远红外和红外天文学的主范围，而再低频则会受到电离层更强的限制。

这种频率选择性直接影响望远镜的设计和站址选择。低频阵需要同时面对电离层相位扰动、强射频干扰和极大视场带来的方向依赖校准问题；厘米波段阵列通常可以在地面较稳定地运行，因此成为射电干涉测量最成熟的主战场；毫米和亚毫米阵列则倾向于建设在高海拔、低水汽环境，例如阿塔卡马高原。也就是说，观测频段从一开始就决定了望远镜面对的是哪一种传播介质、哪一种噪声背景，以及哪一种校准困难。

大气透明也不等于频谱在工程上可用。广播、移动通信、卫星下行、雷达和望远镜自身电子设备都会产生射频干扰（RFI），其强度可比天文信号高许多个数量级。国际与国家频谱管理为若干射电天文频段提供保护，但保护带只覆盖部分频率，也不能自动消除带外泄漏、卫星旁瓣或非法发射。因此，选频还要同时检查目标谱线的红移范围、当地和卫星 RFI 环境、保护频段及接收机动态范围；偏远站址、射电静默区、实时监测和后处理旗标都是观测系统的一部分。


### 1.4.6 波段命名与观测语言

射电天文学中的波段命名带有很强的历史痕迹，字母顺序和频率顺序并不完全一致。这些命名最初来自不同年代的雷达、通信和天文传统，后来逐渐沿用到射电望远镜系统中。因此，波段代号不应被理解为自然界给出的严格边界，而应被理解为一种观测语言：当研究者说“L 波段观测”或“C 波段成像”时，实际上是在简洁地指代一个频率范围、一个典型接收机系统，以及一类常见科学任务。

下面列出一些常见波段。不同台站、接收机和文献中可能采用略有差别的边界，但表中的数量级已经足以支持本书后续讨论。

| 波段 | 频率（GHz） |
|:---:|---:|
| P | 0.23 - 0.47 |
| L | 1 - 2 |
| S | 2 - 4 |
| C | 4 - 8 |
| X | 8 - 12 |
| Ku | 12 - 18 |
| K | 19 - 26.5 |
| Ka | 26.5 - 40.0 |
| Q | 40.0 - 50.0 |


**表 1.4.1**：射电天文学中常见的观测波段。<a id='science:tab:bands'></a> <!--\label{science:tab:bands}-->

因此，当人们说“在 L 波段观测”时，通常指的是约 $1$ 到 $2$ GHz 的频率范围；而当说“毫米波观测”时，往往已经进入更高频、对大气条件、相位稳定性和天线表面精度更敏感的物理区间。用波段代替频率范围，是射电文献中非常常见的专业表达；重要的是始终把这个代号还原成具体的波长、角分辨率、主波束大小、接收机噪声和科学目标。


### 1.4.7 本节小结：射电波段带来了什么

射电波段的意义，可以概括为对宇宙低能、低温、磁化和等离子体环境的独特敏感性。它让光学上不显眼甚至不可见的成分进入观测范围，例如冷气体、喷流、射电瓣、传播介质效应和相干辐射源；它也把许多关键科学问题转化成可测量量，例如气体速度场、谱指数、亮温、偏振和时间变化。

与此同时，射电波段也对仪器提出了特殊要求。大气窗口决定了哪些频率能从地面观测，电离层和水汽决定了低频与高频观测的主要困难，目标角尺度则决定了单碟与干涉阵各自能恢复哪些空间信息。接下来的几节将进一步讨论射电辐射具体由哪些物理机制产生。也就是说，我们已经看到“射电天空长什么样”，下一步要回答的就是：**这些辐射为什么会出现。**
