# 1.12 第 1 章定量问题集

本问题集检验能否把辐射测量量、传输、辐射机制、谱线和阵列尺度转化为可核查的物理判断。除数值结果外，每题都必须写出单位、近似条件和适用边界；只写最终数字不能得到满分。

总分 100 分。建议先独立完成六题，再用末尾的参考量和评分要点自检。除非另有说明，取 $c=2.998\times10^8\,\mathrm{m\,s^{-1}}$、$k_B=1.381\times10^{-23}\,\mathrm{J\,K^{-1}}$、$1\,\mathrm{pc}=3.086\times10^{16}\,\mathrm{m}$。


## 1. 通量、光度与宇宙学传播（15 分）

一个红移 $z=0.50$ 的各向同性射电源在观测频率 1.4 GHz 处有 $S_{1.4}=2.0\,\mathrm{mJy}$，谱指数约定为 $S_\nu\propto\nu^\alpha$，其中 $\alpha=-0.8$。取光度距离 $D_L=2.9\,\mathrm{Gpc}$。

1. 1.4 GHz 观测对应的发射频率是多少？
2. 用 $L_\nu(\nu)=4\pi D_L^2S_{\nu_o}(\nu)(1+z)^{-(1+\alpha)}$ 计算静止系 1.4 GHz 谱光度。
3. 若错误地省略 K 修正，结果会偏高还是偏低，百分比约为多少？
4. 同一窄频比强度从发射端传播到观测端时仅考虑宇宙学红移，$I_{\nu,\rm obs}/I_{\nu,\rm em}$ 是多少？玻尔兹曼积分面亮度的比例又是多少？


## 2. 辐射传输与自由-自由连续谱（20 分）

一个均匀 H II 区的电子温度 $T_e=8000\,\mathrm{K}$，发射量 $EM=3.0\times10^6\,\mathrm{pc\,cm^{-6}}$，其后方有 $T_{\rm bg}=100\,\mathrm{K}$ 的背景。使用

$$\tau_{\rm ff}=0.328\left(\frac{T_e}{10^4\,\mathrm{K}}\right)^{-1.35}\left(\frac{\nu}{\mathrm{GHz}}\right)^{-2.1}\left(\frac{EM}{10^6\,\mathrm{pc\,cm^{-6}}}\right),$$

以及瑞利-金斯形式 $T_{b,\rm obs}=T_{\rm bg}e^{-\tau}+T_e(1-e^{-\tau})$。

1. 计算 1 GHz 和 10 GHz 的光学深度与观测亮温。
2. 求 $\tau_{\rm ff}=1$ 的转折频率。
3. 若 H II 区只填充恢复波束的 $f=0.25$，计算 10 GHz 连续谱扣除后的线外增量 $\Delta T_b=f(T_e-T_{\rm bg})(1-e^{-\tau})$。
4. 说明为什么低亮温既可能是光学薄，也可能是波束稀释；仅凭一幅图如何给出保守结论？


## 3. 同步辐射、偏振与法拉第旋转（15 分）

一个光学薄同步源位于 $B_\perp=10\,\mu\mathrm{G}$ 的磁场中，在 1.4 GHz 辐射最强，测得谱指数 $\alpha=-0.75$。前景法拉第屏可近似为 $n_e=0.030\,\mathrm{cm^{-3}}$、$B_\parallel=2.0\,\mu\mathrm{G}$、路径长度 $L=1.0\,\mathrm{kpc}$。

1. 用 $\nu_c\simeq4.2B_{\mu\mathrm{G}}\gamma^2\,\mathrm{Hz}$ 估算电子洛伦兹因子。
2. 由 $\alpha=-(p-1)/2$ 求电子能谱指数 $p$，再求均匀磁场中的最大线偏振度 $\Pi_{\max}=(p+1)/(p+7/3)$。
3. 用 $\mathrm{RM}=0.812\int n_eB_\parallel dl$ 计算 RM，并求波长 21 cm 与 6 cm 之间的位置角旋转差。
4. 观测亮温若为 $3\times10^{12}\,\mathrm{K}$，为什么不能只据此宣布测量错误？列出两个需要检查的物理或观测因素。


## 4. 谱线分辨率、线宽与 H I 质量（20 分）

一个星系位于 $z=0.020$，取 $D_L=86\,\mathrm{Mpc}$。其 H I 21 cm 积分通量为 $5.0\,\mathrm{Jy\,km\,s^{-1}}$；某个已分辨区域的连续谱扣除线亮温积分为 $120\,\mathrm{K\,km\,s^{-1}}$。冷 H I 的动力学温度取 100 K，观测线宽 $\mathrm{FWHM}_{\rm obs}=8.0\,\mathrm{km\,s^{-1}}$，仪器高斯响应 $\mathrm{FWHM}_{\rm inst}=2.0\,\mathrm{km\,s^{-1}}$。

1. 求 H I 线观测频率，以及 5 km/s 通道在该频率处对应的通道宽度。
2. 求氢原子的热速度弥散与热 FWHM；假定各项为独立高斯，求非热速度弥散。
3. 用 $N_{\rm HI}=1.823\times10^{18}\int T_{b,\rm line}dv$ 求该区域柱密度。
4. 用 $M_{\rm HI}/M_\odot=2.356\times10^5D_L^2(1+z)^{-1}\int Sdv$ 求星系 H I 质量。
5. 说明若未扣除连续谱，或冷 H I 已经光学厚，上述柱密度会怎样失效。


## 5. 从有效面积到空间尺度（15 分）

一个由 25 m 天线组成的阵列在 1.5 GHz 观测。单天线孔径效率 $\eta_a=0.70$、系统温度 $T_{\rm sys}=30\,\mathrm{K}$，最长与最短投影基线分别为 36 km 和 680 m。采用

$$A_{\rm eff}=\eta_a\pi D^2/4,\quad \mathrm{SEFD}=2k_BT_{\rm sys}/A_{\rm eff},$$
$$\theta_{\rm res}=\lambda/B_{\max},\quad \theta_{\rm PB}=1.02\lambda/D,\quad \theta_{\rm MRS}=0.6\lambda/B_{\min}.$$

1. 计算单天线 $A_{\rm eff}$ 和 SEFD，后者用 Jy 表示。
2. 计算名义分辨率、主波束 FWHM 与 MRS，并说明三个角尺度分别限制什么。
3. 若反射面均方根误差 $\sigma=0.20\,\mathrm{mm}$，用 Ruze 公式计算 100 GHz 时的效率比 $\eta/\eta_0$。
4. 说明为什么增大图像尺寸或减小像素不能补回超过 MRS 的结构。


## 6. 从科学目标反推仪器方案（15 分）

一个项目希望同时测量：$0.2''$ 的 AGN 核位置、约 $2'$ 的低面亮度同步晕、红移 $z=0.05$ 的 H I 线轮廓，以及 1--2 GHz 线偏振与 RM。请写一份不超过 15 行的观测方案，至少说明：

- 所需频率覆盖、通道宽度、时间采样和偏振相关产品；
- 为什么单一阵列构型很难同时满足 $0.2''$ 与 $2'$；
- 需要的长基线、短基线、mosaic 或单碟/总功率补充；
- 为公平测谱指数和 RM，如何匹配 $uv$ 覆盖与恢复波束；
- 至少三个会使最终物理量只能作为上限或下限的失败迹象。


## 选做：修正黑体与尘埃质量（不计入 100 分）

一个本地星系的积分尘埃连续谱在 230 GHz 为 $S_\nu=100$ mJy，距离 $D=10$ Mpc。假定尘埃等温且光学薄，$T_d=20$ K，且按尘埃质量定义的 $\kappa_{230}=0.050\,\mathrm{m^2\,kg^{-1}}$。

1. 计算 $x=h\nu/kT_d$ 以及 $B_\nu/B_\nu^{\rm RJ}=x/(e^x-1)$。
2. 用 $M_d=S_\nu D^2/[\kappa_\nu B_\nu(T_d)]$ 估算尘埃质量；再说明使用瑞利-金斯近似会使结果向哪个方向偏移。
3. 列出至少三个不能由这一个通量点检验、但会进入质量系统误差的假设。


## 参考量与评分要点

**第 1 题：** 发射频率 2.1 GHz；$L_{1.4}\simeq1.86\times10^{24}\,\mathrm{W\,Hz^{-1}}$。省略 K 修正会偏高约 8.4%。窄频比强度比为 $(1.5)^{-3}\simeq0.296$，玻尔兹曼面亮度比为 $(1.5)^{-4}\simeq0.198$。频率与单位 3 分，光度 6 分，偏差 3 分，面亮度 3 分。

**第 2 题：** $\tau_{1\rm GHz}\simeq1.33$、$T_b\simeq5.9\times10^3\,\mathrm{K}$；$\tau_{10\rm GHz}\simeq0.0106$、$T_b\simeq183\,\mathrm{K}$。转折频率约 1.15 GHz；10 GHz 的波束稀释增量约 21 K。光深和亮温各 6 分，转折 3 分，填充因子 2 分，退化解释 3 分。

**第 3 题：** $\gamma\simeq5.8\times10^3$，$p=2.5$，$\Pi_{\max}\simeq72.4\%$；$\mathrm{RM}\simeq48.7\,\mathrm{rad\,m^{-2}}$，21 cm 与 6 cm 间旋转差约 1.97 rad（113 度，实际角度有 $\pi$ 模糊）。亮温超过 $10^{12}$ K 时应检查源系 Doppler boosting、相干机制、角尺寸上限、散射展宽和变源时间尺度。前三项各 4 分，边界解释 3 分。

**第 4 题：** $\nu_{\rm obs}\simeq1.3926\,\mathrm{GHz}$，5 km/s 对应约 23 kHz；$\sigma_{\rm th}\simeq0.91\,\mathrm{km\,s^{-1}}$，热 FWHM 约 2.14 km/s，扣除热与仪器方差后 $\sigma_{\rm nt}\simeq3.16\,\mathrm{km\,s^{-1}}$。$N_{\rm HI}\simeq2.19\times10^{20}\,\mathrm{cm^{-2}}$，$M_{\rm HI}\simeq8.54\times10^9M_\odot$。前四项各 4 分，失效条件 4 分。

**第 5 题：** $A_{\rm eff}\simeq344\,\mathrm{m^2}$，$\mathrm{SEFD}\simeq241\,\mathrm{Jy}$；$\theta_{\rm res}\simeq1.15''$，$\theta_{\rm PB}\simeq28.0'$，$\theta_{\rm MRS}\simeq36.4''$。100 GHz 的 Ruze 效率比约 0.50。面积与 SEFD 5 分，三个角尺度 5 分，Ruze 3 分，采样解释 2 分。

**第 6 题：** 没有唯一设施组合。满分方案必须把每项设置连接到一个科学尺度或误差项，明确高分辨率与低面亮度产品职责不同，并讨论 H I 频率约 1.353 GHz、连续谱扣除、全 Stokes、宽带偏振、短间距和共同分辨率。可接受的失败迹象包括缺短间距导致负碗与通量随 taper 变化、核未分辨使亮温仅为下限、带宽/时间平均造成离轴衰减、波束与泄漏模型不足、RM 对 $\lambda^2$ 非线性却仍强行单分量拟合。覆盖需求 5 分，空间尺度方案 5 分，QA 与降级规则 5 分。

**选做题：** $x\simeq0.552$，$B_\nu/B_\nu^{\rm RJ}\simeq0.75$；完整普朗克函数给出 $M_d\simeq3.9\times10^6M_\odot$，瑞利-金斯近似约给出 $2.9\times10^6M_\odot$，低估约 25%。必须说明该质量只含尘埃，并讨论温度混合、$\kappa_\nu$ 归一化与 $\beta$、光深、距离、非尘埃连续谱污染或气尘比中的至少三项。
