## 1.10 单碟望远镜的局限与价值

单碟望远镜能够直接接收射电信号并测量总功率，但其角分辨率受口径和观测波长限制。射电波长远长于可见光波长，单纯增大连续孔径难以在工程上实现亚角秒乃至更高角分辨率，因此需要采用由离散阵元组成的干涉阵。与此同时，单碟望远镜在总功率测量、大尺度结构恢复和绝对亮度标定等方面仍具有不可替代的作用。

本节比较单碟望远镜与干涉阵在收集面积、角分辨率、空间尺度响应、视场、绝对零点和数据处理流程方面的差异。这些差异决定了两类系统的适用科学目标，也构成后续可见度、成像和校准方法的仪器基础。


### 1.10.1 单碟望远镜的持续科学价值

“单碟望远镜”通常指以单个连续反射面作为主要收集孔径的射电望远镜。在比较干涉阵时，还可将概念推广到单阵元系统；阵元既可以是抛物面反射面，也可以是低频偶极子或经过站内波束形成的台站。干涉阵中的每个阵元本身仍具有独立接收和方向响应能力。

单阵元系统承担总功率测量、大尺度弥散辐射恢复、绝对亮温或零点测量、大面积扫描，以及部分谱线和时变源观测任务。干涉阵主要解决连续孔径难以进一步提高角分辨率的问题，并不替代单碟望远镜的全部科学功能。

### 1.10.2 收集面积、口径与角分辨率的尺度约束

单台望远镜最直观的优势，是用有限效率把几何面积转换成有效收集面积。若把望远镜近似为圆形孔径，则有

$$A_{\rm eff}=\eta_a\frac{\pi D^2}{4},$$

这里 $D$ 是口径，$\eta_a<1$ 是包含照明、遮挡、表面误差等损失的孔径效率。更大的 $A_{\rm eff}$ 通常意味着更高的灵敏度，但系统温度也同样关键；常用的等效系统通量密度为

$$\mathrm{SEFD}=\frac{2k_BT_{\rm sys}}{A_{\rm eff}},$$

它越小，望远镜对同一通量密度源的信噪比越高。单碟的衍射极限角分辨率大致满足

$$\Delta \theta \approx 1.22\frac{\lambda}{D}$$

其中 $\lambda$ 是观测波长。由于射电波长通常远长于可见光波长，在相同角分辨率要求下，射电望远镜需要大得多的等效口径。

这里必须区分两个常被混淆的量。**有效收集面积与系统温度**共同控制灵敏度；**口径尺度**控制角分辨率。虽然在单碟里增大口径通常会同时改善二者，但它们并不是同一回事。到了干涉阵，二者甚至会以不同方式分离开来：阵列灵敏度取决于各阵元的有效面积、系统温度、带宽和相关效率，而最高角分辨率主要由最长基线决定。


| 目标角分辨率 $\Delta \theta$ | 可见光 <br> $\lambda=500$ nm | 红外 <br> $\lambda=10\,\mu$m | 毫米波 <br> $\lambda=10$ mm，30 GHz | 米波 <br> $\lambda=1$ m，300 MHz |
|:---:|:---:|:---:|:---:|:---:|
| $0.8''$ | 15 cm | 3 m | 3 km | 300 km |
| $0.25''$ | 50 cm | 10 m | 10 km | 1000 km |
| $0.12''$ | 1 m | 21 m | 21 km | 2100 km |
| $0.015''$（15 mas） | 8 m | 168 m | 168 km | 16800 km |


**表 1.10.1**：不同观测波长下达到给定角分辨率所需的等效口径 $D$。


表中数值表明，在 300 MHz 附近达到典型光学角分辨率需要数十至数千千米的等效口径，无法通过建造单个连续反射面实现。下面的交互程序用于计算给定波长和目标角分辨率对应的口径需求。


```python
def aperture_diameter(wavelength_m=1.0, angular_resolution_deg=15e-3 / 3600):
    """Print the circular-aperture diameter for a target resolution."""
    frequency_mhz = 299_792_458.0 / wavelength_m / 1e6
    diameter_m = 1.22 * wavelength_m / np.radians(angular_resolution_deg)
    print(f"Frequency: {frequency_mhz:.3f} MHz (wavelength {wavelength_m:.3g} m)")
    print(f"Required diameter: {diameter_m:.3g} m")
    print(f"Resolution: {angular_resolution_deg * 3600:.3g} arcsec")


ipywidgets.interact(
    aperture_diameter,
    angular_resolution_deg=(15e-3 / 3600, 10, 1e-5),
    wavelength_m=(0.5e-6, 1, 1e-7),
)
```

### 1.10.3 单碟口径扩展的工程限制

单碟角分辨率由 $\lambda/D$ 决定，但连续增大口径会受到多项工程约束。

首先，大型可转向结构必须在重力作用下保持足够高的面形精度和指向精度。口径越大，自重引起的挠曲越严重，结构设计、驱动系统、基础设施和维护成本都会急剧上升。其次，观测频率越高，允许的面形误差就越小。若反射面均方根误差为 $\sigma$，Ruze 近似给出高频孔径效率损失

$$\frac{\eta}{\eta_0}=\exp\left[-\left(\frac{4\pi\sigma}{\lambda}\right)^2\right].$$

因此，在毫米和亚毫米波段，反射面误差必须远小于观测波长，大口径与高面形精度难以同时实现。传统单馈源系统一次形成一个波束，多馈源接收机和相控阵馈源（PAF）则可同时形成多个波束并提高巡天速度；但单台反射面内部不会产生阵元之间的多条独立基线，因而不能提供干涉阵的二维空间频率采样。


大型单碟的发展体现了这些工程约束。可转向的 300 英尺绿岸望远镜显示了大口径结构面临的机械和材料限制。阿雷西博将反射面固定于天然地形，以有限天空可达范围换取较大口径。中国 FAST 采用主动反射面和馈源舱系统，在更大口径下实现局部抛物面跟踪，但其可观测天区、频率范围和系统性能仍受设计边界限制。单碟技术仍可继续发展，但口径扩展需要同时解决结构、驱动、面形、指向和维护成本问题。


![口径与分辨率缩放](figures/reflector_scaling_schematic.png)

### 1.10.4 单碟望远镜的独特功能

单碟望远镜的重要优势是能够直接测量天空总功率。干涉阵测量阵元之间的互相关，通常缺少 $B=0$ 的零间距测量，因此对足够大尺度的平滑结构不敏感。

因此，延展射电源即使具有较高总通量，也可能在干涉阵的已采样基线上被分辨。银河系弥散同步辐射、分子云大尺度 CO 发射、星系盘外延 H I 气体、太阳和附近延展星云的部分结构，都可能需要单碟数据恢复总通量和低空间频率信息。毫米波和厘米波观测因而常采用单碟与干涉阵联合成像或短间距补偿。


单碟的另一个重要价值在于大面积扫描和总功率光谱测量。对于巡测、宽视场弥散结构研究，以及某些要求绝对亮度或零点的谱线项目，单碟往往更直接。但“包含零间距”不等于自动得到可靠绝对亮度：增益漂移、$1/f$ 噪声、驻波、扫描条纹、旁瓣拾取和基线扣除都可能改变大尺度结构，必须依靠扫描策略、交叉扫描、基线模型和外部标定控制。干涉阵虽然分辨率高，也并不天然适合所有科学问题。这里最重要的判断是：更高角分辨率不等于更完整的信息。


<img src="figures/single_dish_interferometer_spatial_scales.png" alt="single_dish_interferometer_spatial_scales.png" width="820">


**图 1.10.6**：单碟与干涉阵对空间尺度的响应示意。单碟包含零间距信息，能够测量总功率和大尺度结构；干涉阵由有限的最短与最长基线采样空间频率，因此会同时面对最大可恢复尺度和最高角分辨率两个边界。


### 1.10.5 从单碟到合成孔径的概念转变

当连续孔径无法进一步扩展时，可以通过离散阵元采样和数字相关提高等效尺度。下面的示意图说明从连续孔径到稀疏孔径的概念转变：干涉阵将连续孔径替换为多个独立阵元，并在电子学和计算过程中保留和重建阵元之间的相位关系。

**图 1.10.7a**：一个简化的单碟望远镜模型。入射平面波（红色虚线）沿着光线追迹路径（青色）被抛物面反射并汇聚到焦点处的接收机。


对于单碟望远镜，远处天体的波前可近似为平面波，并由反射面汇聚至焦点接收机。不同孔径位置的电磁场通过反射面光路在硬件中完成相干叠加，接收机输出相应的总信号。


![cartoon_2.png](figures/cartoon_2.png)


**图 1.10.7b**：可以把整面反射镜看成由多个较小的反射面组成，每个子反射面都有自己通向焦点的光路。


将连续反射面分解为许多微小子孔径时，单碟可视为连续孔径采样的极限，而干涉阵则属于稀疏孔径系统。第 4 章将使用孔径平面和可见度空间的数学语言描述这一关系。


![cartoon_3.png](figures/cartoon_3.png)


**图 1.10.7c**：每个子孔径各自接收信号，随后按选定方向补偿延迟并相干求和，形成一个数字波束；这并不会自动补回未被阵元覆盖的连续孔径。


若各子孔径能够独立接收并数字化信号，就可以用电子学延迟和数字处理替代同一焦点处的被动汇聚。射电信号可由接收机下变频、采样和后处理，从而实现数字波束形成或相关测量。


![cartoon_4.png](figures/cartoon_4.png)


**图 1.10.7d**：子阵元被布置到地面上并配有独立信号链。加入适当时延可以对某个方向相干求和，但稀疏阵列的收集面积、旁瓣和尺度响应仍由真实阵元分布决定。


这些子孔径无需排列成完整抛物面。已知阵元间几何时延并进行相应补偿后，可将阵元分散布置形成阵列。该过程不会直接生成图像，但能将最高角分辨率对应的尺度由单阵元口径扩展到阵列最大基线。


![cartoon_5.png](figures/cartoon_5.png)


**图 1.10.7e**：相关器保留阵元对之间的复相关，产生供后续成像反演使用的多条可见度样本；相关器输出本身仍不是天空图像。


多个阵元信号经延迟补偿后直接相加，对应**波束形成**，系统在指定方向形成等效波束并输出合成信号。对阵元信号进行两两互相关则得到**可见度**。相关器保留不同基线上的复数测量，使阵列能够采样二维天空亮度分布的空间频率信息。


![cartoon_6.png](figures/cartoon_6.png)

干涉阵不需要在整个等效孔径内连续布置阵元。有限离散位置形成的基线可以获得由最大基线决定的高角分辨率，但稀疏孔径同时引入有限灵敏度、旁瓣、空间频率采样不完整和大尺度结构恢复不足等问题。


### 1.10.6 干涉阵的能力与代价

干涉阵将最高角分辨率对应的尺度由单阵元口径扩展到最长基线。若最长基线为 $B_{\max}$，角分辨率近似按 $\theta_{\rm res}\sim\lambda/B_{\max}$ 缩放；若最短投影基线为 $B_{\min}$，最大可恢复尺度可作数量级估计 $\theta_{\rm MRS}\sim 0.6\lambda/B_{\min}$。系数 0.6 不是普适常数，实际 MRS 还取决于投影基线分布、权重、观测时长、mosaic 策略和源形态。单阵元口径 $D_{\rm ant}$ 则决定主波束视场，近似为 $\theta_{\rm PB}\sim\lambda/D_{\rm ant}$。角分辨率、最大可恢复尺度和主波束视场是三个不同的角尺度。

干涉阵不能等同于具有相同外径的连续孔径望远镜。它将角分辨率提高到最长基线对应的尺度，同时引入零间距缺失、大尺度通量恢复不完整、主波束视场限制、有限 $uv$ 覆盖以及更严格的相位稳定性和延迟模型要求。复可见度还需经过校准、权重选择、成像和去卷积，才能形成可供科学分析的图像。

角分辨率与灵敏度也必须分别讨论。理想热噪声按有效带宽和积分时间的平方根下降，即 $\sigma\propto(\Delta\nu\,t)^{-1/2}$；增加带宽或积分时间不能恢复未采样的空间尺度。点源灵敏度主要由总有效收集面积、系统温度和权重决定，面亮度灵敏度还强烈依赖恢复波束和短基线密度；使用较小恢复波束通常会降低每波束面亮度灵敏度。完整辐射计方程和 SEFD 权重将在 7.3 与 7.9 节讨论。

高质量射电成像需要综合权衡不同仪器能力：单碟数据提供总功率和大尺度结构，干涉阵数据提供高角分辨率和结构分解能力，二者联合可获得更完整的空间尺度信息。后续短间距、成像权重、主波束和校准章节将进一步讨论这些问题。


### 1.10.7 本节小结

单碟望远镜的收集面积与口径平方相关，而角分辨率近似按 $\lambda/D$ 缩放。由于射电波长很长，若只靠单碟获得高角分辨率，所需口径会很快进入工程上不可实现的范围。干涉阵通过离散阵元和相关处理把分辨率提升到最长基线尺度，这是现代高分辨率射电科学得以成立的关键。

大型单碟望远镜仍具有重要作用，可测量总功率、恢复大尺度结构并提供零间距信息。现代射电观测通常综合利用单碟望远镜与干涉阵的互补能力。下一节将进一步比较现代射电干涉阵在频率范围、阵元类型、视场和基线分布方面的设计差异。
