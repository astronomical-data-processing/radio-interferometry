## 5.x 延伸阅读与参考文献 <a id='imaging:sec:refs'></a>


### 综合成像总论

1. [<cite data-cite='taylor1999synthesis'>Synthesis Imaging in Radio Astronomy II（Lecture 7）</cite> &#10548;](https://ui.adsabs.harvard.edu/abs/1999sira.conf.....T)：本章中关于成像、加权与采样函数的很多基础内容都可追溯到这份讲义。
2. [<cite data-cite='thompson2008interferometry'>Interferometry and Synthesis in Radio Astronomy（Chapter 10）</cite> &#10548;](https://doi.org/10.1002/9783527617845)：可与 *Synthesis Imaging in Radio Astronomy II* 第 7 讲配合阅读，是一章很好的补充材料。

### 权重函数

1. [<cite data-cite='briggs1995high'>High Fidelity Deconvolution of Moderately Resolved Sources</cite> &#10548;](https://www.aoc.nrao.edu/dissertations/dbriggs/)：提出鲁棒权重的原始工作，也对各类权重方案作了较系统的综述。

### 宽视场效应

1. [<cite data-cite='taylor1999synthesis'>Synthesis Imaging in Radio Astronomy II（Lecture 19）</cite> &#10548;](https://ui.adsabs.harvard.edu/abs/1999sira.conf.....T)：更深入地讨论宽视场成像问题。
2. [<cite data-cite='cornwell2008noncoplanar'>The noncoplanar baselines effect in radio interferometry: The W-projection algorithm</cite> &#10548;](https://doi.org/10.1109/JSTSP.2008.2005290)：详细介绍 W-projection 算法。
3. [<cite data-cite='cornwell1992radio'>Radio-interferometric imaging of very large fields: The problem of non-coplanar arrays</cite> &#10548;](https://ui.adsabs.harvard.edu/abs/1992A%26A...261..353C)：概述宽视场问题以及多面体解法，并讨论其他相关技术。
4. [<cite data-cite='bhatnagar2008correcting'>Correcting direction-dependent gains in the deconvolution of radio interferometric images</cite> &#10548;](https://doi.org/10.1051/0004-6361:20079284)：A-projection 及方向相关增益进入成像算子的基础论文。
5. [<cite data-cite='offringa2014wsclean'>WSCLEAN</cite> &#10548;](https://doi.org/10.1093/mnras/stu1368)：从现代实现角度讨论 W-stacking、宽场成像性能和工程取舍。

### 宽带成像

1. [<cite data-cite='rau2011multi'>A multi-scale multi-frequency deconvolution algorithm for synthesis imaging</cite> &#10548;](https://doi.org/10.1051/0004-6361/201117104)：MT-MFS 的测量方程、Taylor 项及误差边界。

### 网格化与去网格化

1. [<cite data-cite='rhodes1970spheroidal'>On the Spheroidal Functions</cite> &#10548;](https://cdm16009.contentdm.oclc.org/digital/collection/p13011coll6/id/61576)
2. [<cite data-cite='tan1986aperture'>Aperture-synthesis mapping and parameter estimation</cite> &#10548;](https://ethos.bl.uk/OrderDetails.do?uin=uk.bl.ethos.384529)：关于最优网格化与去网格化函数的进一步讨论。
3. 本章代码中的伴随内积测试应与任何生产级 gridder 一起使用；它能发现索引、复共轭和 FFT 归一化错误，但不能单独证明抗混叠误差已经足够小。
