## 1.5 热辐射

射电连续谱通常分为热辐射和非热辐射两大类。热辐射并非仅指高温物体的辐射，而是指辐射场与物质温度之间存在确定联系：辐射强度可以由热平衡直接决定，或由局部热平衡条件下的粒子速度分布描述。同步辐射、脉冲星辐射和其他相干辐射过程则属于非热辐射。

热辐射为射电观测提供了若干基本物理量和近似关系。普朗克谱给出热平衡辐射的基准谱形；瑞利-金斯近似建立亮度与温度之间的低频关系；亮温将比强度表示为等效温度；自由-自由辐射则把电离气体的物理状态与射电连续谱联系起来。本节将据此建立温度、光学深度、谱形和物理环境之间的关系。


### 1.5.1 黑体辐射<a id='science:sec:blackbody_emission'></a>

黑体辐射是热辐射的理想化基准。当辐射与物质充分交换能量并达到热平衡时，辐射场的比强度只由温度决定，而与材料成分、表面形状和具体发射机制无关。严格满足黑体条件的天体系统较少，但黑体谱仍是判断热辐射、定义亮温和理解光学厚极限的出发点。

黑体比强度由普朗克公式给出：

\begin{equation}
B_{\nu}(T)=\frac{2h\nu^{3}}{c^{2}}\frac{1}{\exp(h\nu/kT)-1},
\end{equation}

其中 $h$ 是普朗克常数，$k$ 是玻尔兹曼常数，$c$ 是光速。频率 $\nu$ 与温度 $T$ 在普朗克函数中通过下列无量纲参数共同决定谱形：

$$
x={h\nu\over kT}.
$$

当 $x\gg 1$ 时，光子能量远高于热运动能量，指数项主导，普朗克谱进入维恩尾端，

$$
B_{\nu}(T)\simeq {2h\nu^3\over c^2}\exp\left(-{h\nu\over kT}\right).
$$

当 $x\ll 1$ 时，可以展开 $\exp(x)-1\simeq x$，于是得到瑞利-金斯近似：

$$
B_{\nu}(T)\simeq {2kT\nu^2\over c^2}.
$$

瑞利-金斯低频极限是射电天文学中的基础近似。在许多厘米波和毫米波问题中，热平衡辐射的比强度与温度近似成正比，频率依赖主要表现为 $\nu^2$。亮温定义、谱线亮温 $T_b(v)$ 以及 Jy/beam 与 K 之间的换算均建立在这一近似之上。


普朗克谱峰值的位置由维恩位移定律给出。若用频率形式表示，峰值频率近似为

$$
\nu_{\max}\simeq 58.8\,T\ {\rm GHz},
$$

其中温度单位为 K。宇宙微波背景的温度约为 $2.725$ K，因此其按 $B_\nu$ 表示的谱峰位于约 $160$ GHz；温度更高的热源，其谱峰向更高频率移动。图 1.5.1 同时给出完整普朗克谱和瑞利-金斯近似：二者在低频端重合，而接近谱峰后瑞利-金斯近似明显失效。

#### 1.5.1.1 亮温的物理意义

瑞利-金斯近似自然引出亮温。对于任意辐射源，只要已知它在某个方向上的比强度 $I_\nu$，都可以定义

$$
T_b \equiv {c^2\over 2k\nu^2} I_\nu.
$$

亮温的定义不要求辐射源本身是黑体；它表示产生同一比强度的瑞利-金斯黑体所对应的等效温度。因此，亮温通常不是物质的实际温度，而是以温度单位表示比强度的观测量。对于光学厚热源，$T_b$ 可以接近电子温度、尘埃温度或激发温度；对于非热源，$T_b$ 主要作为判断辐射机制的诊断量。

**拓展内容：普朗克等效亮温。** 当 $h\nu/kT$ 不再远小于 1 时，可以直接反解普朗克函数，定义 Planck 等效亮温

$$T_{\rm P}=\frac{h\nu/k}{\ln\!\left(1+\dfrac{2h\nu^3}{c^2I_\nu}\right)}.$$

$T_{\rm P}$ 与瑞利-金斯亮温 $T_b$ 在低频极限一致，但在毫米波和亚毫米波可能显著不同。还应区分图像中的 $\mathrm{K_{RJ}}$ 与 CMB 工作中的热力学温标 $\mathrm{K_{CMB}}$：后者用小扰动关系 $\delta I_\nu=(\partial B_\nu/\partial T)_{T_{\rm CMB}}\delta T_{\rm CMB}$ 定义，并不是把同一强度简单代入瑞利-金斯公式。

亮温与干涉成像关系尤其密切。实际图像常以 Jy/beam 给出，而物理解释常需要 K。若合成波束近似为二维高斯，主轴半高全宽分别为 $\theta_{\rm maj}$ 和 $\theta_{\rm min}$（单位为角秒），则常用换算为

$$
T_b({\rm K})\simeq 1.222\times 10^6\,{S_\nu({\rm Jy\ beam}^{-1})\over
\nu_{\rm GHz}^2\,\theta_{\rm maj}('')\,\theta_{\rm min}('')}.
$$

该式表明，亮温同时取决于通量密度和角分辨率。相同的 Jy/beam 在较小波束中对应较高的面亮度；合成波束发生变化时，图像单位 Jy/beam 与 K 之间的换算也随之改变。后续成像和科学解释必须同时报告频率与波束参数。


#### 1.5.1.2 从黑体到真实天体

黑体辐射是理想模型，真实天体通常只在有限频段和一定光学深度条件下接近黑体。行星、宇宙微波背景、光学厚 H II 区、部分恒星大气以及尘埃连续谱，都可能在特定频率范围内使用黑体或修正黑体模型描述。模型适用性取决于辐射是否接近热平衡、光学深度是否足够大、瑞利-金斯近似是否成立，以及亮温能否代表相应的物理温度。

光学深度决定了热辐射的观测含义。在光学厚极限下，辐射场趋向 $B_\nu(T)$，观测量主要反映源函数或局部温度；在光学薄极限下，观测量反映沿视线累积的发射率，亮温通常低于粒子温度。许多热辐射谱形和亮度差异都可由这两个极限及其过渡状态解释。


#### 1.5.1.3 选读：修正黑体与尘埃质量

毫米和亚毫米连续谱中的热尘埃通常不是理想黑体。对温度近似为 $T_d$ 的等温尘埃层，可写成

$$I_\nu=B_\nu(T_d)(1-e^{-\tau_\nu}),\qquad \tau_\nu=\kappa_\nu\Sigma_d,$$

其中 $\Sigma_d$ 是尘埃面密度，$\kappa_\nu$ 是按尘埃质量定义的吸收不透明度。有限频段内常用 $\kappa_\nu=\kappa_0(\nu/\nu_0)^\beta$，但 $\beta$ 和 $\kappa_0$ 依赖颗粒成分、尺寸与所采用的尘埃模型，并不是普适常数。

在光学薄、等温、距离可按欧氏几何处理，并且使用源的积分通量时，尘埃质量满足

$$S_\nu\simeq\frac{\kappa_\nu M_d B_\nu(T_d)}{D^2},\qquad M_d\simeq\frac{S_\nu D^2}{\kappa_\nu B_\nu(T_d)}.$$

这里没有额外的 $4\pi$：各向发射的 $4\pi$ 与通量的几何稀释因子相消。对宇宙学源必须在发射频率评价 $\kappa_\nu B_\nu$ 并使用一致的红移因子，不能把本地公式只换成 $D_L$。若同时满足瑞利-金斯近似和 $\kappa_\nu\propto\nu^\beta$，光学薄尾部才有 $S_\nu\propto T_d\nu^{2+\beta}$。例如 $T_d=20$ K、$\nu=230$ GHz 时 $h\nu/kT_d\simeq0.55$，且 $B_\nu/B_\nu^{\rm RJ}\simeq0.75$；此时直接使用瑞利-金斯近似会把质量低估约 25%。

因此，单频尘埃通量不能同时独立确定 $T_d$、$\beta$、$\tau_\nu$ 和 $M_d$。温度混合、未分辨的光学厚结构、不透明度归一化，以及从尘埃质量换算气体质量所需的气尘比，通常主导系统误差；这条公式应被视为带模型条件的质量估计，而不是直接称量。


### 1.5.2 热轫致辐射<a id='science:sec:bremsstrahlung_emission'></a>

热轫致辐射是射电天文学中的重要热连续谱机制，也称自由-自由辐射（free-free emission）。它产生于自由电子在离子库仑场中的偏转和加速过程。“自由-自由”表示电子在相互作用前后均处于自由态，未被俘获到原子能级。

这类辐射广泛存在于 H II 区、行星状星云、恒星风、超新星遗迹中的热等离子体、星系核区和星暴星系。与理想黑体辐射相比，自由-自由辐射更直接反映电离气体的密度、温度和几何分布。将其与同步辐射区分，可以分别约束电离气体和相对论电子这两类物理成分。


自由-自由辐射的经典描述建立在加速电荷辐射的基础上。在经典电动力学中，带电粒子的辐射功率与加速度平方成正比。采用高斯单位制，拉莫尔公式为

$$
P={2\over 3}{q^2 a^2\over c^3}.
$$

当电子掠过离子时，库仑力会改变电子速度方向，从而产生瞬时加速度。由于电子质量远小于离子，在同样电场作用下电子的加速度大得多，因此天体自由-自由辐射主要由电子在离子电场中的运动贡献。单次相互作用的轨迹会形成连续频谱；在天体等离子体中，观测到的是大量电子、离子和速度分布平均后的总体辐射。


<img src="figures/bremsstrahlung_fig.png" alt="bremsstrahlung_fig.png" width="330">


**图 1.5.2**：自由电子在离子库仑场中偏转并产生轫致辐射的示意图。


若电子速度服从麦克斯韦分布，对速度和冲量参数积分后，自由-自由辐射的发射率可以写成近似形式

$$
j_\nu^{\rm ff}\propto n_e n_i Z^2 T_e^{-1/2}
\exp\left(-{h\nu\over kT_e}\right)\,\bar g_{\rm ff},
$$

其中 $n_e$ 和 $n_i$ 分别是电子和离子数密度，$Z$ 是离子电荷数，$T_e$ 是电子温度，$\bar g_{\rm ff}$ 是 Gaunt 因子，用于修正经典近似与量子力学结果之间的差别。在射电频率下通常有 $h\nu\ll kT_e$，指数项接近 1，因此发射率对频率的依赖较弱，而对密度乘积的依赖较强。自由-自由辐射由此成为电离气体发射量的重要示踪量。

相应的吸收系数在射电近似下近似满足

$$
\kappa_\nu^{\rm ff}\propto n_e n_i Z^2 T_e^{-3/2}\nu^{-2}\bar g_{\rm ff}.
$$

更常用的观测量是沿视线积分得到的发射量

$$
EM=\int n_e^2\,dl,
$$

以及由它决定的自由-自由光学深度。对氢等离子体，一个常用近似为

$$
\tau_\nu^{\rm ff}\simeq 8.24\times 10^{-2}
\left({T_e\over {\rm K}}\right)^{-1.35}
\left({\nu\over {\rm GHz}}\right)^{-2.1}
\left({EM\over {\rm pc\ cm}^{-6}}\right),
$$

其中指数 $-2.1$ 已经把 Gaunt 因子的缓慢频率变化合并进去。由于这个数值系数对单位极为敏感，同一个近似更不容易误用的规范化写法是

$$\tau_\nu^{\rm ff}\simeq0.328\left(\frac{T_e}{10^4\,\mathrm{K}}\right)^{-1.35}\left(\frac{\nu}{\mathrm{GHz}}\right)^{-2.1}\left(\frac{EM}{10^6\,\mathrm{pc\,cm^{-6}}}\right).$$

两式适用于典型射电条件下、Gaunt 因子采用上述近似的氢等离子体；精密计算应使用与离子成分和频率相匹配的 Gaunt 因子。该关系表明，低频辐射更容易受到自由-自由吸收，高密度或长路径的电离气体更容易达到光学厚状态，而较高温度会降低吸收。


将自由-自由辐射代入辐射转移方程，可得到常用的等温介质关系。若电离气体近似等温并处于局部热平衡，源函数为 $B_\nu(T_e)$，于是

$$
I_\nu=B_\nu(T_e)\left(1-e^{-\tau_\nu}\right).
$$

在瑞利-金斯极限下，这个式子等价于

$$
T_b=T_e\left(1-e^{-\tau_\nu}\right).
$$

这条关系直接连接了观测亮温和电离气体物理状态。对固定立体角的均匀等温薄层，若 $\tau_\nu\gg 1$，亮温趋向电子温度，谱形接近光学厚黑体的 $S_\nu\propto\nu^2$；若 $\tau_\nu\ll 1$，则 $T_b\simeq T_e\tau_\nu$，而通量密度近似满足 $S_\nu\propto\nu^{-0.1}$，表现为非常平坦的热连续谱。真实源若有密度梯度，不同视线会在不同频率达到 $\tau_\nu\sim1$，谱指数可以落在两极限之间；经典稳态、等温、恒速电离恒星风给出约 $S_\nu\propto\nu^{0.6}$。正是这一厚薄转换，使紧致 H II 区、恒星风和星系中的电离气体可以通过宽频连续谱观测被区分出来。


<img src="figures/freefree_spectral_regimes.png" alt="freefree_spectral_regimes.png" width="760">


**图 1.5.3**：自由-自由辐射从光学厚到光学薄的谱形转换。低频端若光学深度较大，谱形接近 $S_\nu\propto\nu^2$；高频端进入光学薄状态后，谱形变为近似平坦的 $S_\nu\propto\nu^{-0.1}$。


#### 1.5.2.1 热轫致辐射的观测诊断

自由-自由辐射主要产生于电离气体。因此，年轻大质量恒星周围的 H II 区、行星状星云、恒星风以及强恒星形成星系的射电连续谱中常含有显著的自由-自由成分。它与同步辐射有两项主要差别：光学薄自由-自由谱较为平坦，而典型同步辐射谱随频率升高明显下降；自由-自由亮温受电离气体温度和光学深度限制，通常不会达到极高的非热亮温。

实际观测中，热与非热成分往往混合在一起。低频可能受到自由-自由吸收影响，高频可能受到尘埃连续谱影响，中间厘米波段则常同时包含自由-自由和平滑的同步辐射背景。因此，区分辐射机制不能只看一张图像，而要结合谱指数、亮温、偏振、空间位置和多波段环境。自由-自由辐射通常无显著线偏振，并与 H$\alpha$、红外尘埃加热和复合线等电离气体示踪量相关；同步辐射则更直接追踪相对论电子和磁场。


H II 区是自由-自由辐射最经典的天体环境之一。年轻高温恒星发出的紫外光电离周围氢气，形成温度约 $10^4$ K 的电离气体云。由于射电自由-自由辐射不易受到尘埃消光影响，它能够揭示光学波段被遮蔽的电离区，并与分子云、尘埃连续谱和谱线数据一起构成恒星形成研究的重要观测基础。


<img src="figures/hii_region_freefree_schematic.png" alt="H II region optical and radio schematic" width="760">


**图 1.5.4**：H II 区的光学与射电自由-自由辐射概念对照。尘埃可以显著削弱光学辐射，而厘米波自由-自由连续谱仍能追踪视线方向的发射量；等值线仅表示教学用的平滑结构，不对应某个真实天体或观测波束。


以猎户座星云为例，同一区域同时包含热连续谱、复合线、分子云边界和几何结构信息。连续谱可约束电离气体和能量输入，谱线可约束速度、化学组成和密度结构。连续谱与谱线的联合分析因而能够提供较完整的物理图景。


### 1.5.3 本节小结

黑体辐射是热平衡辐射场的基准模型，瑞利-金斯近似建立了比强度与亮温之间的联系，亮温则连接图像单位与物理解释。自由-自由辐射表明，电离气体即使不满足理想黑体条件，也会通过热电子与离子的相互作用产生可观测的射电连续谱。

热连续谱的解释主要依赖光学深度、温度和谱形。光学厚时，亮温趋向相应的物理温度；光学薄时，亮温反映沿视线累积的发射量。下一节将讨论非热连续谱的主要机制之一，即同步辐射。
