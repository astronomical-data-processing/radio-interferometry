# 射电干涉数据处理实践

本册是《射电干涉孔径合成基础》的实践教材，面向高年级本科实践课、研究生课程、暑期学校和研究训练。第一册建立辐射、Fourier、可见度、成像、观测系统和校准原理；本册将这些原理落实为数据检查、校准、成像、科学测量、质量控制和可复现项目。

本册沿用第 9 章的 `9.x` 编号，以保持现有链接和课程材料兼容。章节编号在结构稳定前不作迁移；读者可按本页给出的阅读路径选择相应的实践单元。


## 先修要求与环境边界

读者应理解第一册第 4 至第 8 章的可见度、成像、去卷积、RIME 和校准主线；只完成基础实践时，也可以在遇到公式时回查相应章节。本书默认环境仍由仓库根目录 `requirements.txt` 提供，真实轻量样例不要求 CASA 或 `python-casacore`。

CASA、WSClean、DP3/DDFacet、CARTA、PyBDSF、SoFiA 和大型公开归档数据属于课程扩展环境。使用这些工具时，应单独记录软件版本、数据身份、参数和运行日志，不把平台相关依赖加入基础教材环境。

## 页面类型

本书把页面标为四类。**讲授页**解释工作流背后的物理和统计判断；**实验页**包含可运行代码、样例数据和定量产物；**参考页**整理软件生态、特殊体制或拓展算法；**项目页**给出里程碑、量规和提交物。层级标签中的**核心**适合基础教学主线，**提高**供专题选读，**课程**用于综合作业。页面类型不同，不意味着科学标准不同，但读者不必把参考页或课程页当作线性必修内容。完整分类同时记录在 `book_manifest.yaml`，并由测试检查。

| [9.11](9_11_polarization_calibration_and_rm_synthesis.ipynb) | 校准与 $\lambda^2$ 采样允许恢复哪些 Faraday 结构？ | 校准证据、dirty spectrum、RMSF 与算法停止条件 |

## 短间距测量链

短间距专题先建立缺失尺度与输入数据的诊断，再进入定量方法选择和验收。9.10 是基础讲授页；需要报告总通量、柱密度、质量、速度场或功率谱时，应继续完成 9.21，不能以图像外观代替 transfer-function 验证。

| 页面 | 唯一问题 | 交付物 |
| --- | --- | --- |
| [9.10](9_10_short_spacing_and_feather.ipynb) | 数据缺少哪些尺度，feather 输入是否满足共同坐标、亮度、beam、PB 和标尺契约？ | 科学尺度、输入语义、可信 overlap 与不可恢复尺度的组合前诊断 |
| [9.21](9_21_short_spacing_mosaic_joint_imaging.ipynb) | Mosaic、多配置、feather 或 joint deconvolution 实际增加什么约束，结果怎样验收？ | Response/MRS、方法选择、通量恢复、残差、拆分/注入检验与限制声明 |


## 宽带宽场测量链

宽带宽场专题按科学产品、已知成像算子和未知方向响应分成三层。前两层没有通过时，不应直接增加方向相关校准自由度；需要谱指数、偏振或高动态范围科学量时，三层都要给出相应证据。

| 页面 | 唯一问题 | 交付物 |
| --- | --- | --- |
| [9.8](9_8_wideband_and_widefield_imaging.ipynb) | 怎样得到参考频率、Taylor 协方差、PB 语义和失效区域都明确的宽带产品？ | $I_0$、谱指数/曲率、协方差、PB/noise/selection 与分频段验证 |
| [9.12](9_12_direction_dependent_widefield_imaging.ipynb) | 已知 $w$ 项、A-term 和 pointing 怎样组成相容的 forward/adjoint operator？ | 相位预算、算子选择、A-term 来源、位置相关 PSF/sensitivity 与注入检验 |
| [9.22](9_22_wideband_widefield_algorithm_boundaries.ipynb) | 哪些剩余结构可归因于未知 DDE，怎样受约束求解并拒绝过拟合？ | 固定/求解/忽略项清单、规范与正则化、训练/留出比较、科学量稳定性和停止条件 |

## 学习成果与结论边界

完成基础路径后，读者应能说明每一步改变的是数据、模型、权重、flag 还是图像解释，并能用残差、噪声、通量、位置或选择函数判断结果。完成课程项目后，读者还应能提交数据清单、参数、运行记录、QA 证据、误差预算和限制声明。

实践训练不要求每个数据集都产生可发表图像。识别数据不适合某个科学问题、拒绝没有绝对标尺的 Jy 结论，或在证据不足时降低结论等级，都是合格且重要的学习成果。
