## 1.6 同步辐射

前一节讨论的热辐射建立了射电亮度与温度、电离气体和光学深度之间的联系。同步辐射则是射电天文学中最重要的非热连续谱机制之一。许多射电源在厘米波和米波段具有很高亮度，其能量来源不是热平衡过程，而是磁场中的相对论电子。喷流、射电瓣、超新星遗迹壳层、星系盘非热连续谱和星系团弥散射电晕均以同步辐射为主要机制。

本节依次讨论单电子辐射、电子能谱与射电谱指数的关系、偏振对磁场几何的约束以及谱老化对粒子冷却历史的记录，并进一步说明这些物理性质对多频、宽带、偏振和多空间尺度干涉观测提出的要求。


### 1.6.1 单个相对论电子的辐射

同步辐射的物理起点是带电粒子在磁场中的螺旋运动。洛伦兹力改变粒子速度方向，因而产生横向加速度；只要带电粒子被加速，就会辐射电磁波。在非相对论极限下，这种辐射称为回旋辐射，频率接近回旋频率。当天体中的电子达到相对论能量时，辐射会被强烈束射到电子瞬时速度方向附近，观测到的频率也会因相对论效应显著提高，这就是同步辐射。

若电子洛伦兹因子为 $\gamma$，磁场中垂直于速度方向的分量为 $B_\perp$，则同步辐射总功率的标度可以写成

$$
P_{\rm syn}\propto \gamma^2 B_\perp^2.
$$

更完整地说，辐射损失率可写为

$$
P_{\rm syn}=2\sigma_T c\,\beta^2\gamma^2 U_B\sin^2\alpha_p,
$$

其中 $\sigma_T$ 是 Thomson 截面，$U_B=B^2/8\pi$ 是磁场能量密度，$\alpha_p$ 是电子速度与磁场方向之间的夹角（pitch angle）。若该夹角的分布各向同性，则 $\langle\sin^2\alpha_p\rangle=2/3$，平均功率可写成 $\langle P_{\rm syn}\rangle=(4/3)\sigma_Tc\,\beta^2\gamma^2U_B$。区分单电子公式与各向同性平均公式，可以避免混用不同假设。该关系同时表明，电子能量和磁场强度越高，同步辐射损失率越大。

单个电子的辐射并不是一条无限窄的谱线，而是在一个特征频率附近具有较宽的连续谱。常用特征频率为

$$
\nu_c\simeq {3e\over 4\pi m_e c}\,B_\perp\gamma^2.
$$

若把磁场写成微高斯量级，可以记成数量级关系

$$
\nu_c\simeq 4.2\,B_{\mu{\rm G}}\gamma^2\ {\rm Hz}.
$$

因此，在 $10\,\mu{\rm G}$ 的磁场中，$\gamma\sim 10^4$ 的电子主要在 GHz 量级辐射。同一天体的不同观测频率对应不同能量电子的主要辐射贡献，多频射电观测由此能够约束电子能量分布。

### 1.6.2 从电子能谱到射电谱指数

真实天体中的同步辐射由相对论电子群体共同产生。喷流、射电瓣、超新星遗迹和星系盘中的电子通常由激波、湍流或磁重联等过程加速，许多加速机制会形成近似幂律的电子能量分布：

$$
N(E)\,dE=N_0 E^{-p}\,dE.
$$

当这群电子在磁场中产生光学薄同步辐射时，观测到的通量密度也呈幂律：

$$
S_\nu\propto \nu^\alpha,
\qquad
\alpha=-{p-1\over 2}.
$$

该关系把电子能谱指数 $p$ 与可观测的射电谱指数 $\alpha$ 联系起来。典型光学薄同步辐射常有 $\alpha\sim -0.5$ 至 $-1.0$，明显陡于热自由-自由辐射的光学薄谱指数 $\alpha\simeq -0.1$。宽带通量测量和谱指数图因而是区分热与非热连续谱的基本工具。

高能电子具有较高的辐射功率，同时也具有较短的冷却时间。随着时间推移，高频辐射对应的高能电子优先损失能量，使同步辐射谱在高频端变陡，这一现象称为谱老化。若只考虑同步损失，电子在频率 $\nu$ 附近的典型寿命近似满足 $t_{\rm syn}\propto B^{-3/2}\nu^{-1/2}$；若同时考虑宇宙微波背景等光子场引起的逆康普顿损失，则有效损失率取决于 $B^2+B_{\rm IC}^2$，其中 $B_{\rm IC}\simeq 3.25(1+z)^2\,\mu{\rm G}$。因此，射电瓣、星系团射电晕和老化电子群体的高频辐射较弱，而低频阵列更容易探测较早期注入的非热结构。

### 1.6.3 偏振、亮温与低频翻转

同步辐射的另一个关键特征是线偏振。单个电子辐射的电矢量与磁场投影有确定关系；若电子能谱为幂律，均匀磁场中的光学薄同步辐射理论最大线偏振度为

$$
\Pi_{\max}={p+1\over p+7/3}.
$$

对常见的 $p\sim 2$--$3$，理论最大值可达约 $70\%$。真实天体的观测偏振度通常较低，因为望远镜波束内可能混合多个磁场方向，湍流会降低有序磁场分量，法拉第旋转和法拉第退偏也会改变不同频率下的偏振叠加。偏振成像仍是研究磁场几何、喷流结构和传播介质的重要方法。

在简单的外部法拉第屏模型中，线偏振位置角满足

$$\chi(\lambda^2)=\chi_0+\mathrm{RM}\,\lambda^2,$$

$$\mathrm{RM}=0.812\int n_e(\mathrm{cm^{-3}})\,B_\parallel(\mu\mathrm{G})\,dl(\mathrm{pc})\quad\mathrm{rad\,m^{-2}},$$

其中 RM 的正负还取决于视线磁场和积分方向约定。法拉第旋转校正之前不能直接把观测电矢量位置角当成源处方向；校正后，光学薄同步辐射的电矢量通常垂直于天球面上的有序磁场投影。若发射与旋转介质混合，$\chi$ 对 $\lambda^2$ 未必是一条直线，需要后续章节的宽带偏振和 RM synthesis 方法。

亮温也可用于判断辐射机制。热源通常不能长期维持远高于物质温度的亮温；若图像推得的 $T_b$ 显著超过热平衡温度，则支持非热辐射解释。对静止源系中的非相干同步辐射，逆康普顿灾变通常将可持续亮温限制在约 $10^{12}$ K 的数量级，接近能量均分时的特征尺度往往更低。相对论多普勒增亮可以提高表观亮温，相干辐射也不受同一限制，因此该数量级只能作为机制诊断，不能作为严格上限。

致密同步辐射源还可能受到同步自吸收影响：低频光子在源内部被同一相对论电子群体吸收，使低频谱由下降幂律转为上升谱。理想均匀自吸收源的光学厚谱可接近 $S_\nu\propto\nu^{5/2}$，外部自由-自由吸收也可产生低频翻转。因此，低频谱形变化不足以单独判定吸收机制，还需结合角尺度、亮温、源环境和多频结构进行分析。


### 1.6.4 同步辐射的观测诊断

同步辐射的识别通常依赖多条证据同时成立。最直接的证据是光学薄幂律连续谱；若谱指数明显为负，并且比自由-自由辐射更陡，就强烈指向非热电子。线偏振提供第二条证据，因为热自由-自由辐射本身几乎不产生显著线偏振，而同步辐射天然带有偏振。高亮温提供第三条证据，尤其是在紧致射电核、喷流和脉冲星风云中。空间上的谱指数变化则提供更深入的信息：核心、喷流、热点和射电瓣常具有不同谱指数，反映粒子注入、输运和冷却的历史。

这些诊断也直接影响干涉观测策略。研究同步辐射需要可靠的多频通量标定、匹配的 $uv$ 覆盖和共同分辨率下的谱指数成图；偏振研究还需要处理泄漏项、绝对偏振角、法拉第旋转和带宽退偏；研究弥散射电瓣或星系晕时，则必须保证短基线对大尺度结构具有足够的敏感性。同步辐射因此构成了连接辐射机制与干涉观测设计的典型案例。


### 1.6.5 典型同步辐射源

同步辐射需要磁场和相对论电子，这两项条件广泛存在于不同天体环境。超新星遗迹的激波可以加速电子并形成壳状非热辐射；脉冲星风云由高速粒子风和磁场共同塑造；活动星系核喷流将能量输运到星系外并形成热点和射电瓣；星系盘中的宇宙线电子在微高斯磁场中产生弥散非热连续谱；星系团中的弥散射电晕和射电遗迹则记录大尺度结构形成过程中的湍流与激波。

射电星系天鹅座 A 是典型实例之一，其核心、喷流、热点和射电瓣结构可用于说明粒子加速、能量输运和谱老化之间的关系。


<img src="figures/cygnus_a_radio_schematic.png" alt="Cygnus A radio morphology schematic" width="700">


**图 1.6.3**：以天鹅座 A 为物理原型的双射电瓣概念图。核心、喷流、热点和射电瓣共同说明活动星系核如何把能量输运到星系外介质；图中位置和尺度不复原具体观测。


在这类系统中，中央活动星系核向喷流注入能量，喷流在远端与环境介质相互作用形成热点，随后带电粒子回流并填充射电瓣。热点接近近期粒子加速区，通常具有较平的频谱；距离加速区较远、冷却时间较长的射电瓣则更容易表现出高频变陡的老化特征。形态与频谱的联合分析是射电天体物理中的常用方法。


```python
frequency_mhz = np.array([151.0, 327.5, 1345.0, 4525.0, 8514.9, 14650.0])
lobe_flux_jy = np.array([4746.0, 2752.7, 749.8, 189.4, 83.4, 40.5])
hotspot_flux_jy = np.array([115.7, 176.4, 69.3, 45.2, 20.8, 13.4])

def fit_power_law(frequency, flux):
    return np.polyfit(np.log10(frequency), np.log10(flux), 1)

lobe_fit = fit_power_law(frequency_mhz, lobe_flux_jy)
hotspot_fit_all = fit_power_law(frequency_mhz, hotspot_flux_jy)
hotspot_fit_high = fit_power_law(frequency_mhz[1:], hotspot_flux_jy[1:])
fit_frequency = np.geomspace(frequency_mhz.min(), frequency_mhz.max(), 200)

fig, ax = plt.subplots(figsize=(7.5, 5))
ax.loglog(frequency_mhz, lobe_flux_jy, 'o', color='tab:blue', label='Lobe L data')
ax.loglog(frequency_mhz, hotspot_flux_jy, 's', color='tab:orange', label='Hotspot L1 data')
ax.loglog(fit_frequency, 10**np.polyval(lobe_fit, np.log10(fit_frequency)),
          color='tab:blue', label=f'Lobe fit: alpha={lobe_fit[0]:.2f}')
ax.loglog(fit_frequency[fit_frequency >= frequency_mhz[1]],
          10**np.polyval(hotspot_fit_high, np.log10(fit_frequency[fit_frequency >= frequency_mhz[1]])),
          color='tab:orange', label=f'Hotspot fit above 151 MHz: alpha={hotspot_fit_high[0]:.2f}')
ax.set_xlabel('Frequency [MHz]')
ax.set_ylabel('Flux density [Jy]')
ax.grid(alpha=0.25, which='both')
ax.legend()
plt.tight_layout()
plt.show()
print(f'Lobe all-band alpha: {lobe_fit[0]:.3f}')
print(f'Hotspot all-band alpha: {hotspot_fit_all[0]:.3f}')
print(f'Hotspot alpha above 151 MHz: {hotspot_fit_high[0]:.3f}')
```

**图 1.6.4**：天鹅座 A 的 `L` 射电瓣与较亮 `L1` 热点通量谱。数值逐项取自 [Steenbrugge, Heywood & Blundell (2010) 表 1](https://doi.org/10.1111/j.1365-2966.2009.15663.x)。直线是在 $\log S_\nu$-$\log \nu$ 空间中的未加权描述性拟合；论文表中的图像噪声不是这些积分通量的完整协方差，因此这里不给拟合误差，也不把直线当作物理谱老化模型。


射电瓣六点拟合给出 $\alpha\simeq-1.04$，明显陡于热点在 327.5 MHz 以上的 $\alpha\simeq-0.65$，与电子离开加速区后逐渐冷却的图景一致。然而，热点通量在 151--327.5 MHz 之间上升；若对全部六点采用单一幂律拟合，斜率将变为约 $-0.51$。该差异可能包含低频吸收、分辨率匹配、背景扣除和不同观测数据集通量标度等效应，仅凭这六个数据点无法区分。可靠的谱老化分析还需考虑通量协方差、共同 $uv$ 范围与分辨率、绝对标度系统误差以及弯曲谱模型。


### 1.6.6 选读：与逆康普顿散射的联系

产生同步辐射的相对论电子还会与周围的低能光子场相互作用，将光子散射到更高能量，这一过程称为逆康普顿散射。若被散射的是同一电子群体产生的同步辐射光子，则称为同步自康普顿散射；若目标光子来自宇宙微波背景、恒星光场或尘埃红外辐射，则形成相应的多波段辐射分量。

同步辐射强度同时依赖电子数密度和磁场强度，二者之间存在退化。将射电同步辐射与 X 射线或毫米波逆康普顿信号联合分析，可以对电子能谱和磁场强度给出更严格的约束。对喷流和射电瓣而言，逆康普顿辐射是评估能量损失和磁场强度的重要工具；对星系团而言，热电子对宇宙微波背景的康普顿化还会产生 Sunyaev-Zel'dovich 效应，连接射电与毫米波宇宙学观测。


### 1.6.7 本节小结

同步辐射来自磁场中的相对论电子，是典型的非热连续谱机制。电子能谱若近似幂律，光学薄同步辐射也表现为幂律谱；谱指数、偏振、亮温和低频翻转共同构成识别同步辐射的主要证据。由于高能电子冷却更快，多频成像和谱指数分布还能追踪粒子加速、输运和老化历史。

对干涉测量而言，同步辐射源的复杂性直接转化为观测需求：宽频覆盖用于测量谱形，偏振校准用于研究磁场，短基线用于恢复弥散结构，长基线则用于分辨喷流和热点。下一节将讨论谱线辐射及其对物质成分、速度和气体物理状态的约束。
