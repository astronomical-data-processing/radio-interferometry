# 附录：常用坐标与相关器恒等式

本附录汇总正文中反复使用的五个推导。为避免只记公式而混淆符号，先固定约定：方位角 $\mathcal A$ 从北向东增加，高度角 $\mathcal E$ 从地平面向天顶增加，时角 $H$ 从本地子午圈向西增加；相位中心为 $(\alpha_0,\delta_0)$，并定义 $\Delta\alpha=\alpha-\alpha_0$。基线沿用全书约定 $\mathbf b_{pq}=\mathbf r_p-\mathbf r_q$。若采用不同的方位角零点、时角方向或相关积顺序，矩阵中的相应符号必须一起改变。


## A.1 从地平坐标 $(\mathcal A,\mathcal E)$ 到时角-赤纬 $(H,\delta)$

设台站纬度为 $L_a$。球面三角形 $STZ$ 给出地平坐标与赤道坐标之间的关系；相关球面余弦定理、正弦定理和五部公式见 [第 2.13 节](../2_Mathematical_Groundwork/2_13_spherical_trigonometry.ipynb)。

<img src='figures/conversion-comp.svg' width='70%'>

<div align="center"><b>图 A.1</b>：球面三角形 $STZ$ 及本节采用的角度约定。</div>

由三条球面三角关系分别得到

$$
\begin{aligned}
\sin\delta
  &=\sin\mathcal E\sin L_a+\cos\mathcal E\cos L_a\cos\mathcal A,\\
\cos\delta\sin H
  &=-\cos\mathcal E\sin\mathcal A,\\
\cos\delta\cos H
  &=\cos L_a\sin\mathcal E-\sin L_a\cos\mathcal E\cos\mathcal A.
\end{aligned}
$$

第二式的负号来自“方位角向东增加、时角向西增加”这两个相反方向。它是最容易在不同教材约定之间被翻转的一项。


把单位方向写成赤道直角分量，可得

$$
\begin{pmatrix}
\cos\delta\cos H\\
\cos\delta\sin H\\
\sin\delta
\end{pmatrix}
=
\begin{pmatrix}
\cos L_a\sin\mathcal E-\sin L_a\cos\mathcal E\cos\mathcal A\\
-\cos\mathcal E\sin\mathcal A\\
\sin L_a\sin\mathcal E+\cos L_a\cos\mathcal E\cos\mathcal A
\end{pmatrix}.
$$

数值实现应使用双参数反正切而不是 $\arctan(y/x)$：

$$
\delta=\arcsin z_{\rm eq},\qquad
H=\operatorname{atan2}(y_{\rm eq},x_{\rm eq}).
$$

这样能保留象限，并在 $x_{\rm eq}=0$ 时保持稳定。在天极 $\cos\delta=0$ 处，时角本身没有唯一意义；程序应保留这一坐标奇点，而不是伪造一个高精度角度。


## A.2 赤经-赤纬与方向余弦 $(l,m,n)$

以相位中心 $(\alpha_0,\delta_0)$ 为局部 $n$ 轴，$l$ 指向局部东，$m$ 指向局部北。对天空方向 $(\alpha,\delta)$，令 $\Delta\alpha=\alpha-\alpha_0$，则

<img src='figures/conversion2-comp.svg' width='70%'>

<div align="center"><b>图 A.2</b>：相位中心附近的方向余弦几何。</div>

$$
\begin{aligned}
l&=\cos\delta\sin\Delta\alpha,\\
m&=\sin\delta\cos\delta_0-\cos\delta\sin\delta_0\cos\Delta\alpha,\\
n&=\sin\delta\sin\delta_0+\cos\delta\cos\delta_0\cos\Delta\alpha.
\end{aligned}
$$

旋转保持单位长度，因此 $l^2+m^2+n^2=1$。相位中心本身对应 $(l,m,n)=(0,0,1)$；通常成像前半球取 $n=+\sqrt{1-l^2-m^2}$。


### A.2.1 反变换

把上式看成三维旋转并取逆，可直接得到

$$
\sin\delta=m\cos\delta_0+n\sin\delta_0,
$$

以及

$$
\cos\delta\sin\Delta\alpha=l,\qquad
\cos\delta\cos\Delta\alpha=n\cos\delta_0-m\sin\delta_0.
$$

所以稳定的反变换为

$$
\boxed{\delta=\arcsin(m\cos\delta_0+n\sin\delta_0)},
$$

$$
\boxed{\alpha=\alpha_0+
\operatorname{atan2}\!\left(l,\,n\cos\delta_0-m\sin\delta_0\right)}.
$$

计算后应把赤经按所需范围折回，例如 $[0,2\pi)$。原点在天极时赤经存在坐标奇点；有限精度输入还应先把反正弦的参数截断到 $[-1,1]$。


## A.3 单基线的 $uv$ 轨迹椭圆

令 $(X,Y,Z)$ 为基线 $\mathbf b_{pq}$ 在正文采用的赤道坐标架中的米制分量，$H$ 和 $\delta$ 为跟踪方向的时角和赤纬。以波长为单位的坐标为

$$
u=\frac{X\sin H+Y\cos H}{\lambda},
$$

$$
v=\frac{-X\sin\delta\cos H+Y\sin\delta\sin H+Z\cos\delta}{\lambda}.
$$

消去 $H$，有

$$
u^2+\left(\frac{v-Z\cos\delta/\lambda}{\sin\delta}\right)^2
=\frac{X^2+Y^2}{\lambda^2}.
$$

定义 $a=\sqrt{X^2+Y^2}/\lambda$ 和 $v_0=Z\cos\delta/\lambda$，可写为

$$
\boxed{\frac{u^2}{a^2}+
\frac{(v-v_0)^2}{a^2\sin^2\delta}=1}.
$$

因此轨迹中心为 $(0,v_0)$，半轴为 $a$ 与 $a|\sin\delta|$。当 $\delta=0$ 时椭圆退化为直线 $v=Z/\lambda$；当 $X=Y=0$ 时 $a=0$，轨迹退化为单点。有限时角覆盖只描出这条椭圆的一段，而不是完整椭圆。


## A.4 方向余弦中的立体角元

在单位球前半球令

$$
l=\sin\theta\cos\phi,\qquad
m=\sin\theta\sin\phi,\qquad
n=\cos\theta=\sqrt{1-l^2-m^2}.
$$

<img src='figures/cosine2.svg' width='70%'>

<div align="center"><b>图 A.3</b>：球坐标 $(\theta,\phi)$ 与方向余弦 $(l,m,n)$。</div>

记 $r=\sqrt{l^2+m^2}=\sin\theta$。由 $\theta=\arcsin r$、$\phi=\operatorname{atan2}(m,l)$ 得

$$
\left|\frac{\partial(\theta,\phi)}{\partial(l,m)}\right|
=\frac{1}{nr}.
$$

代入 $d\Omega=\sin\theta\,d\theta\,d\phi$，得到

$$
\boxed{d\Omega=\frac{dl\,dm}{n}}\qquad(n>0).
$$

虽然 $(\theta,\phi)$ 在 $r=0$ 处有极坐标奇点，乘积中的 $r$ 会抵消，$d\Omega$ 的极限有限。若处理另一半球，应写成 $dl\,dm/|n|$ 并另外记录 $n$ 的符号；单独的 $(l,m)$ 不能区分前后半球。


## A.5 延迟信号的乘积恒等式

相关器推导常用以下积化和差公式：

$$
\begin{aligned}
\cos(x-y)\cos x
 &=\tfrac12\cos y+\tfrac12\cos(2x-y),\\
\sin(x-y)\sin x
 &=\tfrac12\cos y-\tfrac12\cos(2x-y),\\
\sin(x-y)\cos x
 &=-\tfrac12\sin y+\tfrac12\sin(2x-y),\\
\cos(x-y)\sin x
 &=\tfrac12\sin y+\tfrac12\sin(2x-y).
\end{aligned}
$$

例如第一式由 $\cos a\cos b=[\cos(a-b)+\cos(a+b)]/2$ 直接得到。若 $x=\omega t$ 是快速载波相位、$y=\omega\tau$ 是几何延迟相位，理想低通或足够长的时间平均会抑制 $2x-y$ 的和频项，留下与 $\cos y$ 或 $\sin y$ 成正比的相关输出。有限积分时间、带宽和滤波器响应决定抑制是否充分，因此不能在推导之外默认高频项严格为零。
