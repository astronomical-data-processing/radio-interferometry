## 术语表 <a id='preface:sec:glossary'></a>

### A

**adjoint operator（伴随算子）:** 采用 $\langle\mathbf x,\mathbf y\rangle=\mathbf x^H\mathbf y$ 时，算子 $A$ 的伴随 $A^H$ 满足 $\langle A\mathbf x,\mathbf y\rangle=\langle\mathbf x,A^H\mathbf y\rangle$。成像中的 gridder 应与预测可见度的 degridding 算子构成伴随对；FFT 采用不同归一化时等式会带相应尺度因子。

**aliasing（混叠）:** 采样后不同连续频率或空间频率产生相同离散样本的现象；超过 Nyquist 范围的成分会折返并污染主带，采样后通常无法唯一撤销。

**azimuth（方位角）:** 在天球地平坐标系中，从正北沿东向量取的角距离。

**altitude（高度角）:** 天体与地平面之间的夹角。

**altitude-azimuth mount（高度-方位式支架）:** 一种让天线绕竖直的方位轴和水平的高度轴转动，从而跟踪天空目标的天线架设方式。

**A-projection / AW-projection（A/AW 投影）:** 在卷积网格化与反网格化核中应用已给定的方向相关天线响应；AW-projection 还联合处理 $w$-term。它们是预测/成像算子，本身不会自动从数据估计未知或错误的波束。

### B

**baseline（基线）:** 两天线位置之差形成的有向向量。本教材采用 $\mathbf b_{pq}=\mathbf r_p-\mathbf r_q$，并与 $V_{pq}=\langle v_pv_q^*\rangle$ 及相应 Fourier 指数配套；交换天线顺序会翻转基线并使理想可见度复共轭。

**beam filling factor（波束填充因子）:** 发射区域在望远镜波束内所占的有效比例。未分辨或只部分填充波束的源会发生波束稀释。

**brightness temperature（亮温）:** 把比强度转换为等效温度的表示量；瑞利-金斯亮温为 $T_b=c^2I_\nu/(2k_B\nu^2)$，不一定等于物质的动力学或热力学温度。

**brightness/coherency matrix（亮度/相干矩阵）:** 双极化电场二阶统计量 $\mathbf B=\langle\mathbf e\mathbf e^H\rangle$。本教材在线极化基中采用 $\mathbf B=\tfrac12\begin{pmatrix}I+Q&U+iV\\U-iV&I-Q\end{pmatrix}$，因此 $\operatorname{tr}\mathbf B=I$；物理亮度矩阵应为 Hermitian 半正定。

### C

**calibration（校准）:** 给定测量方程、天空模型、权重和规范约束，从数据估计仪器或传播响应的反问题。较低训练残差不自动意味着模型正确或科学量无偏。

**celestial equator（天球赤道）:** 地球赤道所在平面向天球延伸后得到的大圆。

**CLEAN（迭代去卷积）:** 用选定基函数字典在残图中寻找分量，并逐步减去相应 PSF 响应的算法家族；mask、模型字典和停止准则形成模型选择，循环增益主要是数值步长。

**closure phase（闭合相位）:** 同一时频点上沿闭合天线三角形有向相位之和；方向无关天线相位在和中抵消，但绝对位置、基线误差和热噪声不会由此恢复或消失。

**closure amplitude（闭合振幅）:** 四条基线幅度构成、可抵消方向无关天线振幅增益的比值；它不保留绝对通量尺度，低信噪比时还会有非线性偏差。

**completeness（完整率）:** 给定通量密度、尺寸和位置条件下，真实或注入源中被检测流程成功找回的比例；它描述漏检，不等同于输出目录的可靠率。

**condition number（条件数）:** 线性算子最大与最小奇异值之比，用于衡量输入扰动被解放大的最坏程度；秩亏矩阵的 2-范数条件数为无穷大。

**celestial horizon（天球地平圈）:** 观测者的水平面在天球上的投影，是地平坐标系的基本参考平面。

**celestial sphere（天球）:** 一个包围地球的想象单位球面，宇宙中所有天体都可视为投影其上。

### D

**deapodization / gridding correction（去切趾/网格校正）:** 卷积网格化后，在图像有效区域内除以卷积核的 Fourier 响应，以校正核造成的方向相关幅度 taper；响应接近零处不能稳定相除。

**declination（赤纬）:** 天体沿其时圈相对于天球赤道的角距离；北天球为正，南天球为负。

**density weighting function（密度加权函数）:** 见 *weighting, density*。

**direction cosine coordinates（方向余弦坐标）:** 把天空单位方向投影到以参考方向为 $n$ 轴、局部东和北为 $l,m$ 轴的正交基底所得坐标。前半球有 $n=+\sqrt{1-l^2-m^2}$；背半球需取负支，投影边界处不能继续沿用普通切平面近似。

**direction-dependent effects（方向依赖效应）:** 随观测方向变化而变化的传播或仪器效应。

**direction-independent effects（方向无关效应）:** 在所有方向上都相同的传播或仪器效应。

**dirty beam（脏波束，也称脏束）:** 可见度采样与成像权重产生的点扩散函数，通常含旁瓣；本文以“脏波束”为主术语。

**dirty image（脏图像）:** 见 *image, dirty*。

### E

**Earth-centered Earth-fixed, ECEF（地心地固坐标）:** 原点位于地心并随地球固连旋转的直角坐标表示；精密应用必须说明所采用的地球参考架及历元。

**Earth orientation parameters, EOP（地球定向参数）:** 描述天球参考系与地球参考系瞬时关系的一组参数，主要包括 `UT1-UTC`、极移 $x_p,y_p$ 和天球中间极改正 $dX,dY$。

**Eddington bias（爱丁顿偏差）:** 当源数随通量密度快速变化时，阈值附近更多暗源被正噪声散射进目录、较少亮源被散射出去而造成的统计偏差，常表现为 flux boosting。

**ecliptic（黄道）:** 太阳在天球上视运动所描出的轨迹。

**elevation（仰角）:** 见 *altitude*。

**equatorial coordinate system（赤道坐标系）:** 天文学中广泛使用的坐标系，其基本平面由地球赤道向天球延伸得到。

**equatorial mount（赤道式支架）:** 一种使天线绕极轴转动来跟踪天体的架设方式，该极轴指向天极。

**equinox（分点，坐标语境）:** 传统赤道坐标轴所采用的平春分点时刻标签；它不同于星表位置参数适用的历元。

**epoch（历元）:** 星表位置、自行或台站坐标等参数适用的参考时刻。`J2000.0` 可能在不同上下文中标记历元或分点，使用时须查看元数据。

**effective noise bandwidth, ENBW（有效噪声带宽）:** 把滤波器对白噪声通过的总功率等效为理想矩形带宽；它由完整通道响应决定，不一定等于 FFT/PFB 的通道中心间隔。

### F

**field centre（视场中心）:** 见 *phase centre*。

**Faraday depth（法拉第深度）:** 从发射位置到观测者沿视线积分 $n_eB_\parallel$ 得到的传播量，单位为 $\mathrm{rad\,m^{-2}}$；其符号依赖路径与磁场方向约定。只有简单外部 Faraday-thin 屏下，观测 RM 才可直接等同于单一 Faraday depth。

**Faraday rotation measure, RM（法拉第旋转量）:** 线偏振位置角随波长平方变化的斜率，简单外部屏下有 $\chi=\chi_0+\mathrm{RM}\lambda^2$，单位为 $\mathrm{rad\,m^{-2}}$。

**Faraday rotation measure synthesis, RM synthesis（旋转量综合）:** 把复线偏振 $P(\lambda^2)=Q+iU$ 在采样的 $\lambda^2$ 轴上反演到 Faraday 深度谱 $F(\phi)$ 的方法。有限且不规则的 $\lambda^2$ 覆盖使重建结果与 rotation measure spread function（RMSF）卷积；其分辨率、最大可探测 Faraday 厚度和最大无混叠 Faraday 深度分别受 $\Delta\lambda^2$、最小 $\lambda^2$ 和频道宽度制约。它不是对每个像素简单拟合一条 $\chi$--$\lambda^2$ 直线。

**fringe pattern（条纹图样）:** 一条投影基线在天空方向上产生的复指数或正弦/余弦响应；天空亮度与该条纹加权积分形成对应可见度。

**first point of Aries（白羊宫第一点）:** 太阳从天球赤道南侧穿越到北侧时对应的交点，也就是春分点。

**full width at half maximum, FWHM（半高全宽）:** 指函数在峰值一半处两侧位置之间的宽度，常用于描述分布或波束的尺度。

### G

**gain, complex（复增益）:** 作用在电场电压上的复乘性响应；其模改变电压幅度、相位改变相位，基线可见度响应为 $g_pg_q^*$。功率增益是电压增益模的平方。

**gauge freedom（规范自由度）:** 参数的某种共同变换不改变预测数据的不可辨识方向。例如所有标量天线增益乘同一单位模相位不会改变 $g_pg_q^*$；参考天线只选择表示规范，不增加观测信息。

**great circular arc（大圆弧）:** 大圆上的一段弧。

**great circle（大圆）:** 通过球心的平面与球面相交所得的圆。

### H

**Hermitian transpose（Hermitian 转置/共轭转置）:** 矩阵先逐元素复共轭再转置，记作 $\mathbf A^H$；它不是只做复共轭的 $\mathbf A^*$。

**horizontal coordinate system（地平坐标系）:** 一种基于观测者本地天空的坐标系统，用于描述天体在本地天空中的位置。

**hour circle（时圈）:** 经过天体和北天极、并与天球赤道垂直的大圆。

**hour angle（时角）:** 天体时圈与本地子午圈之间沿天球赤道向西量取的角距离，通常以小时计。

### I

**International Celestial Reference System/Frame, ICRS/ICRF（国际天球参考系统/架）:** `ICRS` 是准惯性天球参考系统的定义，`ICRF` 是由河外射电源位置给出的具体实现；系统和实现不是可互换的两个名称。

**International Terrestrial Reference System/Frame, ITRS/ITRF（国际地球参考系统/架）:** `ITRS` 定义随地球固连的参考系统，`ITRF20xx` 用带历元和速度的站坐标实现它。

**integrated flux density（积分通量密度）:** 对源所占立体角积分得到的通量密度。Jy/beam 图中高斯源的积分通量密度需按拟合源面积与恢复波束面积之比换算，并考虑 residual scaling。本文以“通量密度”为主术语；其他中文资料也常简称“流量”。

**image, dirty（脏图像）:** 对干涉观测中采样后的可见度做傅里叶变换所得到的图像，近似等于理想天空图像与阵列 PSF 的卷积结果。它通常是后续反卷积算法的输入图像。

### J

**Jones matrix（Jones 矩阵）:** 作用在双分量复电场上的 $2\times2$ 复线性变换。单个 Jones 矩阵只描述确定性、乘性和非退偏响应；加性噪声、非线性失真及随机平均造成的退偏需要额外模型。

### K

**K-correction（K 修正）:** 因宇宙学红移使固定观测频段对应不同静止系频段而进行的谱修正；其形式依赖谱指数定义、滤波带和所比较的静止系频率。

### L

**latitude（纬度）:** 地球表面某地点相对于赤道的南北角距离。

**local meridian（本地子午圈）:** 连接天顶与天球北极所形成的时圈。

**local sidereal time, LST（本地恒星时）:** 本地子午圈相对于相应平均或真春分点的时角；东经为正时等于对应格林尼治恒星时加台站经度。平均/视 LST 必须与相应赤经定义匹配。

**longitude（经度）:** 地球表面某地点相对于格林尼治子午线的东西角距离。

**luminosity distance（光度距离）:** 由通量-光度反平方关系 $F=L/(4\pi D_L^2)$ 定义的宇宙学距离；它通常不等于角直径距离或简单欧氏路径长度。

**low noise amplifier, LNA（低噪声放大器）:** 模拟前端链路中的第一级放大器，用于放大微弱天体信号，同时尽量只引入较少的系统噪声。

### M

**matched filter（匹配滤波器）:** 在给定模板和噪声协方差下最大化检测信噪比的线性滤波器。白噪声下等价于数据与模板互相关；相关噪声下 Fourier 权重与模板复共轭成正比、与噪声功率谱成反比。

**maximum recoverable scale, MRS（最大可恢复尺度）:** 干涉数据能够可靠恢复的最大角尺度的经验指标，主要受最短有效投影基线控制，也依赖权重、观测轨迹、mosaic、软件定义和源形态；它不是硬截止。

**multi-term multi-frequency synthesis, MT-MFS（多项多频综合）:** 在联合多频可见度成像时，用参考频率附近的多个 Taylor 项同时描述亮度和谱形的宽带去卷积方法；它不同于只合并频率以改善 $uv$ 覆盖的单项 MFS。

**mosaicking（镶嵌观测/联合拼接）:** 用多个相邻指向及其主波束模型联合约束较大天空区域的方法。它能扩展每条基线附近的 Fourier 约束，但通常不能替代真正的零间距总功率。

**Measurement Set, MS（测量集）:** 射电干涉数据的表格化容器；MAIN 表保存可见度、时间、天线对、uvw、权重和标记，子表保存频率、极化、场和天线等解释数据所需的元数据。

**Mueller matrix（Mueller 矩阵）:** 作用在 Stokes 四向量上的 $4\times4$ 线性映射。确定性非退偏 Jones 过程可导出 Mueller 矩阵，但含退偏的一般 Mueller 映射未必来自单个 Jones 矩阵。

### N

**nadir（天底）:** 天球上与天顶相对的位置。

**natural weighting（自然加权）:** 见 *weighting, natural*。

**north celestial pole, NCP（北天极）:** 地球北极向天球延伸后与天球的交点。

### O

### P

**parallactic angle（视差角）:** 由“源到天顶的大圆”和“源到北天极的大圆”之间形成的球面夹角。

**peeling（逐源剥离校准）:** 对少数亮方向建立局部方向相关模型、求解并减去其贡献的流程。它会引入额外天空-增益自由度，必须检查源通量和未参与求解的数据。

**phase centre（相位中心）:** 用于定义名义相位或局部方向余弦的天空参考方向。相关器延迟跟踪中心、天线指向中心和成像参考方向常相同但概念上可不同，不能与“相位校准所指向的位置”简单等同。

**point spread function, PSF（点扩散函数）:** 测量系统在图像域中对点源产生的响应，本质上可视为可见度采样函数的傅里叶变换。在较早文献中，它也常被称为 *synthesized beam*（综合波束）。

**Polaris（北极星）:** 一颗位于北天极附近的恒星。

**precession（岁差）:** 旋转天体自转轴方向随时间发生的缓慢变化。

**primary beam（主波束）:** 天线增益随方向、频率和极化变化的响应。电场波束是 Jones 响应；相同天线的标量功率波束为 $P=|E|^2$，异质基线响应一般为 $E_pE_q^*$。

**polyphase filter bank, PFB（多相滤波器组）:** 把多抽头原型滤波器与 FFT 结合的通道化器，可降低通道旁瓣和泄漏；通道间隔不必等于其有效噪声带宽或独立谱分辨率。

### Q

### R

**radiation pattern（辐射方向图）:** 见 **primary beam**。

**radio frequency interference, RFI（射频干扰）:** 人为产生并污染射电观测的电磁信号，常呈非高斯、非平稳结构；强干扰还可使前端饱和或产生互调，不能把它简单并入热系统温度。

**Radio Interferometer Measurement Equation, RIME（射电干涉测量方程）:** 用亮度矩阵、几何相位和两端 Jones 链描述极化可见度的矩阵方程；基线方向、相关积次序、亮度归一化和 Fourier 符号必须成套一致。

**rotation measure spread function, RMSF（旋转量扩散函数）:** RM synthesis 在 Faraday 深度域的点响应，由加权的 $\lambda^2$ 采样函数决定；观测到的 Faraday 谱是理想谱与 RMSF 的卷积，因此 RMSF 主瓣宽度和旁瓣必须与峰值一起报告。

**regularization（正则化）:** 在数据拟合目标中加入平滑、稀疏、幅度或其他先验约束，以限制数据不能稳定决定的自由度。

**reliability / purity（可靠率/纯度）:** 输出候选中真实检测所占的比例；它描述误检，不等同于真实源被找回的完整率。

**residual scaling（残差缩放）:** CLEAN 恢复时将 Jy/dirty beam 残差转换到 Jy/clean beam 归一化的处理，通常涉及两个有效波束面积之比，并会同时改变残差 RMS。

**restored image（恢复图）:** 恢复波束卷积后的去卷积模型与单位相容残图之和；其他资料也称复原图，本文以“恢复图”为主术语。

**restoring beam（恢复波束）:** 去卷积模型在最终恢复图中所卷积的理想化波束，通常是拟合 PSF 主瓣的椭圆高斯；它不同于含旁瓣的脏波束。

**right ascension（赤经）:** 天体时圈与春分点之间沿天球赤道向东量取的角距离。

**Ruze efficiency（Ruze 效率）:** 反射面随机误差造成的高频效率损失近似，$\eta/\eta_0=\exp[-(4\pi\sigma/\lambda)^2]$。

**robust weighting（稳健加权）:** 见 *weighting, robust*。

### S

**sampling function（采样函数）:** 表示哪些位置、时间或空间频率被测量以及采用何种权重的函数或分布。理想规则采样可写成 Shah 函数；干涉成像中它通常是位于实际 $uv$ 样本处的加权 delta 之和。

**self-calibration（自校准）:** 用目标场自身的天空模型迭代求天线响应、校正和重建模型的闭环。它可提高相干性和动态范围，也可把模型缺陷吸入增益，因此必须保留外部标尺并验证科学量。

**specific intensity（比强度）:** 单位投影面积、单位频率、单位立体角的辐射功率，是辐射转移和成像的基本量。静态欧氏真空传播时沿射线守恒；宇宙学传播的不变量是 $I_\nu/\nu^3$。

**sidereal day（恒星日）:** 地球相对于春分点完成一周旋转的时间，约为 23 h 56 min 4 s 平均太阳时；恒星时的精确定义还区分 mean 与 apparent。

**solar day（太阳日）:** 太阳回到天空同一位置所需的时间。

**solution interval, solint（校准解间隔）:** 一组校准参数被假定恒定并共同使用数据的时间/频率范围。间隔过短会降低解信噪比并增加自由度，过长则无法跟踪真实变化。

**spatial domain（空间域）:** 一种信号表示域，其中样本之间的相对距离直接反映信号在物理空间中的相对位置，例如星场图像。

**spatial frequency domain（空间频率域）:** 用每单位角度或长度的周期数表示结构尺度的域。每个坐标对应一个复正弦基函数；复系数的模和相位分别编码该尺度分量的振幅与位置偏移，而坐标间距表示空间频率差，不表示样本“相似度”。

**spherical triangle（球面三角形）:** 由三段大圆弧两两相交形成的三角形。

**south celestial pole, SCP（南天极）:** 地球南极向天球延伸后与天球的交点。

**synthesized beam（综合波束）:** 通常指采样和权重产生的干涉 PSF，也称 dirty beam；部分语境会把拟合其主瓣的 Gaussian restoring beam 也简称为 beam，因此报告数值时必须明确是 dirty/synthesized beam 还是 restoring/clean beam。

**system equivalent flux density, SEFD（等效系统通量密度）:** 把系统噪声表示为输入通量密度的灵敏度指标，单阵元常写成 $2k_BT_{\rm sys}/A_{\rm eff}$，其中 $A_{\rm eff}$ 已含口径效率；数值越小越灵敏。

**system temperature（系统温度）:** 在指定参考面上把天空、接收机、大气、溢出和损耗噪声表示为等效温度的量。引用数值时需说明是否已作大气不透明度等校正；RFI 通常不作为热系统温度分量。

### T

**total electron content, TEC（总电子含量）:** 沿传播路径积分的电子柱密度；$1\,\mathrm{TECU}=10^{16}\,\mathrm{electrons\,m^{-2}}$。差分 TEC 使干涉相位近似按 $\nu^{-1}$、群时延按 $\nu^{-2}$ 变化。

**International Atomic Time, TAI（国际原子时）:** 连续原子时标；它与 UTC 的整数秒差会在插入闰秒后改变。

**Terrestrial Time, TT（地球时）:** 用于地心星历等计算的均匀时标，定义关系为 $\mathrm{TT}=\mathrm{TAI}+32.184\,\mathrm{s}$。

**transit（中天）:** 当天体穿过本地子午圈时所处的状态。

**taper weighting function（加窗加权函数）:** 见 *weighting, taper*。

### U

**Universal Time 1, UT1（世界时 UT1）:** 由地球实际自转确定的时标，是计算 ERA 和地球姿态的时间参数；`UT1-UTC` 由 EOP 提供。

**Coordinated Universal Time, UTC（协调世界时）:** 含闰秒的民用与归档时标；它适合标记观测事件，但不能在未知 `UT1-UTC` 时替代 UT1。

**uv coverage（uv 覆盖）:** 阵列在时间、频率和天线对上实际采样的二维投影空间频率位置及权重分布；共轭点不是额外独立天空信息。

**uvw coordinates（uvw 坐标）:** 基线在以参考方向为 $w$ 轴、局部东和北为 $u,v$ 轴的正交架中的分量。理论式常以波长表示，数据表也可能以米存储。

**uniform weighting（均匀加权）:** 见 *weighting, uniform*。

### V

**van Cittert-Zernike theorem（van Cittert-Zernike 定理）:** 在空间非相干、准稳态与远场等条件下，把接收平面两点的互相干函数与源亮度分布联系起来的定理；常见二维 Fourier 成像式还需小视场或可处理的 $w$ 项等条件。

**Van Vleck correction（Van Vleck 校正）:** 把低比特量化后的相关统计映回模拟相关系数的校正。理想零均值联合高斯信号经 1 bit 零阈值符号量化时有 $c=(2/\pi)\arcsin\rho$；非理想阈值和非高斯输入需更一般的校正。

**vernal equinox（春分点）:** 见 *first point of Aries*。

**visibility（可见度）:** 两天线复电压的互相关，经过延迟跟踪和标定后成为天空互相干函数的复样本。其幅度与相位共同编码结构；基线顺序、相关积顺序和指数符号必须一致。

### W

**weighting, density（密度加权）:** 在合并重叠的可见度样本时，用于设定样本显著性的函数，是合成图像时的重要步骤。

**weighting, natural（自然加权）:** 对独立可见度采用逆噪声方差权重、从而最小化图像热噪声的方式；等噪声时每个样本权重相同。它常得到较宽波束，但旁瓣取决于阵列覆盖。

**weighting, robust（稳健加权）:** 一种可通过参数调节、在分辨率与灵敏度之间折中的加权方案。

**weighting, taper（加窗加权）:** 一种充当空间滤波器的加权函数，用于在成像时选择特定空间尺度。

**weighting, uniform（均匀加权）:** 用局部 `uv` 单元内的总自然权重对各样本再归一化，使被采样网格单元的总贡献更接近一致。它通常提高分辨率并增加热噪声；PSF 旁瓣如何变化取决于二维覆盖和密度估计尺度。

### Z

**zenith（天顶）:** 天球上位于观测者正上方的点。

**zero-padding（零填充）:** 在有限序列外补零。它可防止 FFT 线性卷积回绕，或让频谱采样更密；后者只是插值，不会增加观测时长或真实分辨能力。

### 数学符号 <a id='preface:sec:symbols'></a>

同一字母在不同物理语境中可能复用；标量、向量和矩阵的字体以及下标是定义的一部分。出现歧义时，正文必须写出限定词或显式定义，不能只凭字母猜测。

* $\lambda$ : 波长（m）
* $\nu$ : 频率（Hz）

* $\mathbf{B}$ : 亮度相干矩阵
* $\boldsymbol{\mathcal{D}}$, $\boldsymbol{\mathcal{M}}$ : 非极化可见度矩阵
* $\mathbf{e}$ : 电场向量
* $\boldsymbol{\mathcal{G}}$ : 非极化增益矩阵
* $\boldsymbol{\mathscr{G}}$ : $\mathbf{g}\mathbf{g}^H$
* $I_\nu(l,m)$ : 天空比强度；仅在上下文明确时简写为 $I$
* $I,Q,U,V$ : Stokes 参数；其中 Stokes $V$ 的符号由本表所列亮度矩阵固定，不要与标量可见度 $V_{pq}$ 混淆
* $\mathbf{J}$ : Jones 矩阵，常见例子包括：
    * $\mathbf{D}$ : 馈源泄漏
    * $\mathbf{E}$ : 主波束
    * $\mathbf{F}$ : 法拉第旋转
    * $\mathbf{G}$ : 增益
    * $\mathbf{K}$ : 几何延迟
    * $\mathbf{P}$ : 视差角
* Jy : Jansky，通量密度单位，$1\,Jy = 10^{-26}\,W\,m^{-2}\,Hz^{-1}$
* $\boldsymbol{\mathcal{M}}$ : 见 $\boldsymbol{\mathcal{D}}$
* $P$ : 功率（W 或 erg s$^{-1}$）；偏振章节中的 $P=Q+iU$ 会显式称为复线偏振
* $\mathbf{S}$ : Stokes 向量，包含 $I,Q,U,V$；无粗体 $S(u,v)$ 留给采样函数
* $T$ : 温度（K）
* $\mathbf{v}$ : 电压向量
* $\mathscr{V}$ : 可见度函数，表示底层连续函数而非离散测量值
* $V_{pq}$ : 天线对 $p,q$ 的标量可见度；不要与 Stokes $V$ 混淆
* $\mathbf{V}_{pq}$ : 天线对 $p,q$ 的 $2\times2$ 极化可见度矩阵
