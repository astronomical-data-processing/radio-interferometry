# 第 3 章：射电干涉测量中的位置天文学<a id='pos:sec:intro'></a>

射电干涉观测的几何描述由天源方向、观测时刻和台站位置共同确定。星表以赤经和赤纬给出天源方向；观测时刻和台站经度确定地方恒星时，进而得到天源的时角；时角、赤纬和台站纬度共同确定望远镜的方位角与高度角；以相位中心为参考，还可进一步建立方向余弦坐标，为可见度测量方程和成像坐标系奠定基础。

在这条变换链中，天源方向具有三种相互联系的表示。第一种是天球参考架中的星表方向；第二种是台站地平坐标系中的本地指向；第三种是相对于相位中心的局部方向，并与基线投影共同构成可见度相位的几何变量。坐标、时间或符号约定若不一致，将直接引入几何延迟、可见度相位和图像位置误差。

本章着重建立射电干涉测量所需的位置天文学基础，不展开完整的天体测量算法。岁差、章动、光行差、极移、地球定向参数和大气折射等改正通常由标准天文软件实现，但理解其物理作用和适用边界仍是正确使用软件的前提。本章将系统说明星表位置、观测时刻、台站位置、相位中心和局部成像坐标之间的关系。


#### 本章知识主线

本章按照从星表方向到观测几何的顺序展开。赤道坐标用于表示源目录位置和相位中心；地方恒星时与时角描述天源方向相对于地球自转的位置；地平坐标给出台站本地指向和可观测性约束；方向余弦则描述相位中心邻域内的天空方向。下一章将在同一相位中心下引入基线坐标 $(u,v,w)$，并建立天空方向与空间频率之间的关系。


#### 本章结构

第 3.1 节介绍赤道坐标，以及星表位置所涉及的参考系统、参考架、分点和历元。第 3.2 节讨论地方恒星时、时角和过中天，并区分平恒星时与视恒星时、UT1 与 UTC。第 3.3 节将赤道坐标转换为方位角和高度角，进而讨论观测窗口、视差角和坐标奇点。第 3.4 节引入方向余弦 $(l,m,n)$，说明半球分支、SIN 投影和 $w$ 项相位判据。第 3.5 节综合讨论 ICRS/ICRF、ITRS/ITRF、地球定向参数、地心地固坐标、局部东-北-天顶坐标和相位参考误差。章末问题集用于检验整条坐标变换链。
