# 第 4 章：可见度空间<a id='visibility:sec:intro'></a>


第 3 章建立了天空方向、台站指向和相位中心之间的几何关系。本章在此基础上讨论射电干涉仪的基本测量量。单天线测量某一方向范围内的功率；干涉阵则对不同天线接收的电压信号进行互相关和时间平均。互相关结果不是图像中的一个像素，而是天空亮度分布在某一空间频率处的复测量量，称为可见度（visibility）。

孔径合成的基本思想是：干涉阵在 $uv$ 平面上采样天空亮度的傅里叶信息，而不是先形成一幅图像再进行锐化。每条基线在每个时刻和频率频道上都提供一个空间频率采样点，所有采样点共同构成 $uv$ 覆盖。在小视场、远场和窄带等近似成立且采样充分时，可由可见度反演天空图像；采样不完整时，反变换结果是天空亮度与脏波束卷积得到的脏图像。

![可见度测量链条](figures/chapter4_visibility_pipeline.png)

**图 4.0.1：从天空亮度到可见度采样的测量链条。** 相关器将两台天线的电压信号转换为某条投影基线上的复可见度，多个基线、时刻和频率样本共同形成可见度空间中的采样。

### 本章符号约定<a id='visibility:sec:conventions'></a>

下表集中列出本章必须配套使用的定义，并沿用[第 2.0 节的傅里叶变换、复统计与权重约定](../2_Mathematical_Groundwork/2_0_introduction.ipynb#math:sec:canonical_conventions)。不同软件或文献可能采用相反的基线方向或时间因子；比较结果时必须同时转换全部相关定义，不能只改变某一个符号。

| 对象 | 本章定义 | 直接后果 |
|:---|:---|:---|
| 天空方向 | $\mathbf s$ 从阵列指向天源；相位中心为 $\mathbf s_0$ | 相位跟踪后使用 $\mathbf s-\mathbf s_0$ |
| 解析时间因子 | $e^{-2\pi i\nu t}$ | 正几何时间参数在电压中产生负相位 |
| 相关积 | $V_{pq}=\langle v_pv_q^*\rangle$ | 交换顺序给出 $V_{qp}=V_{pq}^*$ |
| 有向基线 | $\mathbf b_{pq}=\mathbf r_p-\mathbf r_q$，方向由 $q$ 指向 $p$ | $\mathbf b_{qp}=-\mathbf b_{pq}$，必须与相关积同时交换 |
| 几何时延 | $\tau_{g,pq}=\mathbf b_{pq}\cdot\mathbf s/c$ | 单方向窄带核为 $e^{-2\pi i\nu\tau_{g,pq}}$ |
| $(u,v,w)$ 与方向余弦 | $\mathbf b_{pq}/\lambda=(u,v,w)$，$\mathbf s=(l,m,n)$，$w$ 指向 $\mathbf s_0$ | 相位跟踪后的核为 $e^{-2\pi i[ul+vm+w(n-1)]}$；理论式中的 $(u,v,w)$ 以波长为单位，数据列也可能以米存储 |
| 天球分支 | 本章局部像面取 $n=+\sqrt{1-l^2-m^2}$ | 只覆盖相位中心前半球，且 $d\Omega=dl\,dm/n$ |

本章讨论标量可见度时省略偏振指标。第 7 章将同一相关次序扩展为亮度矩阵和射电干涉测量方程（RIME）；斯托克斯参数（Stokes）和馈源基底的约定必须与本章的基线方向及共轭次序相容。

本章围绕三项基本关系展开：基线互相关与天空空间频率分量的关系；物理基线随地球自转投影为 $(u,v,w)$ 坐标和 $uv$ 轨迹的关系；以及范·锡特特-泽尼克定理在何种近似下将可见度函数化为天空亮度的二维傅里叶变换。这三项关系构成从接收电压到成像方程的主线。
